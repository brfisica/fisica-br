## Punto 1 — Capacitor de placas paralelas (Ing. Electrónica/Eléctrica)  ★★  [CL, UFT]
En el diseño de un filtro para una fuente de alimentación, se necesita un capacitor de placas paralelas de área  A = 0,0200 m²  y separación  d = 1,00 mm.

**Capacitancia:**  C = ε₀·A/d       **Carga:**  Q = C·V       **Energía:**  U = ½·C·V²

> 🔧 **Aplicación en ingeniería:** dimensionar filtros, temporizadores y fuentes en circuitos electrónicos parte de calcular C. En subestaciones y equipos eléctricos, la geometría (área y separación) fija la capacitancia y el voltaje que soporta antes de la ruptura dieléctrica.

## Punto 5 — Sistema de flash industrial: arreglo serie/paralelo (Ing. Fotónica/Eléctrica)  ★★★  [PRC, UCM, ULC]

En inventario solo hay capacitores de  100 µF  con voltaje nominal máximo de  250 V.

**Serie (reparte voltaje):**  1/C = Σ 1/Cᵢ ,  V_rama = Σ Vᵢ       **Paralelo (suma capacidad):**  C = Σ Cᵢ
**Potencia media de descarga:**  P = U/t ,  con  U = ½·C·V²
