## Punto 1 — Capacitor de placas paralelas (Ing. Electrónica/Eléctrica)  ★★  [CL, UFT]
En el diseño de un filtro para una fuente de alimentación, se necesita un capacitor de placas paralelas de área  A = 0,0200 m²  y separación  d = 1,00 mm.

**Capacitancia:**  C = ε₀·A/d       **Carga:**  Q = C·V       **Energía:**  U = ½·C·V²

> 🔧 **Aplicación en ingeniería:** dimensionar filtros, temporizadores y fuentes en circuitos electrónicos parte de calcular C. En subestaciones y equipos eléctricos, la geometría (área y separación) fija la capacitancia y el voltaje que soporta antes de la ruptura dieléctrica.


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

eps0 = 8.85e-12
k = 1/(4*np.pi*eps0)

def esquema():
    fig,ax=plt.subplots(figsize=(5,3))
    ax.plot([0,0],[-1,1],lw=5,color='red'); ax.plot([1.5,1.5],[-1,1],lw=5,color='blue')
    ax.text(0,1.2,'+Q',color='red',ha='center'); ax.text(1.5,1.2,'−Q',color='blue',ha='center')
    for yy in np.linspace(-0.7,0.7,4):
        ax.annotate('',xy=(1.35,yy),xytext=(0.15,yy),arrowprops=dict(arrowstyle='->',color='gray'))
    ax.annotate('',xy=(1.5,-1.3),xytext=(0,-1.3),arrowprops=dict(arrowstyle='<->')); ax.text(0.75,-1.5,'d',ha='center')
    ax.text(0.75,0.15,'área A',ha='center',fontsize=9)
    ax.set_xlim(-0.5,2); ax.set_ylim(-1.8,1.6); ax.axis('off'); ax.set_title('Capacitor de placas paralelas')
    plt.tight_layout(); plt.show()
esquema()

def capacitor(A_cm2=200.0, d_mm=1.0, V=12.0):
    A=A_cm2*1e-4; d=d_mm/1000; C=eps0*A/d; Q=C*V; U=0.5*C*V**2
    dd=np.linspace(0.2,5,200)/1000; Cd=eps0*A/dd
    plt.figure(figsize=(9,4.2)); plt.plot(dd*1000,Cd*1e12,lw=2,color='darkred')
    plt.scatter([d_mm],[C*1e12],color='blue',s=60,zorder=5,label=f'd={d_mm:.1f} mm → C={C*1e12:.1f} pF')
    plt.xlabel('separación d (mm)'); plt.ylabel('C (pF)'); plt.title('Capacitancia vs separación')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(capacitor,
    A_cm2=FloatSlider(value=200.0,min=50,max=500,step=25,description='A (cm²)'),
    d_mm=FloatSlider(value=1.0,min=0.2,max=5,step=0.1,description='d (mm)'),
    V=FloatSlider(value=12.0,min=1,max=50,step=1,description='V (voltios)'));

def solucion():
    A=0.02; d=0.001; C=eps0*A/d; V=12.0
    print(f"C = ε₀A/d = (8,85e-12·0,0200)/0,00100 = {C:.3g} F = {C*1e12:.3g} pF")
    print(f"Q = CV = {C*V:.3g} C = {C*V*1e9:.3g} nC")
    print(f"U = ½CV² = {0.5*C*V**2:.3g} J")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Capacitancia (A=0,0200 m², d=1,00 mm):",177,0.04,"pF","C=ε₀A/d≈177 pF.")
quiz_opcion_multiple("Si acercas las placas (d menor), la capacitancia:",
    ["Disminuye","Aumenta (C ∝ 1/d)","No cambia","Se anula"],1,"C=ε₀A/d: menor d ⟹ mayor C.")
quiz_numerico("Carga almacenada con V=12,0 V:",2.12,0.05,"nC","Q=CV≈2,12 nC.")
```

## Punto 2 — Efecto del dieléctrico (Ing. Electrónica/Materiales)  ★★  [CL, UCM]
Al insertar un material dieléctrico de constante  κ = 3,50  entre las placas, la capacitancia
cambia (aplicación: condensadores comerciales, aislantes).

**Con dieléctrico:**  C' = κ·C₀       (κ = constante dieléctrica del material)

> 🔧 **Aplicación en ingeniería:** los condensadores comerciales usan dieléctricos (cerámica, poliéster) para lograr alta capacitancia en poco volumen. Elegir el material κ correcto define el tamaño, el costo y el voltaje máximo del componente en un diseño real.

## Punto 3 — Asociación en serie y paralelo (Ing. Eléctrica/Sistemas)  ★★★  [PRC, ULC]
En un banco de condensadores para corrección del factor de potencia de una planta industrial, tres unidades  C₁ = 2,00 µF,  C₂ = 3,00 µF,  C₃ = 6,00 µF  se conectan de dos formas.

**Paralelo:**  C = C₁+C₂+C₃       **Serie:**  1/C = 1/C₁ + 1/C₂ + 1/C₃

> 🔧 **Aplicación en ingeniería:** en bancos de condensadores para corrección del factor de potencia (industria, RETIE) se combinan unidades en serie y paralelo para alcanzar la capacitancia y el voltaje requeridos por la red. Saber asociarlos es diseño eléctrico puro.


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

eps0 = 8.85e-12
k = 1/(4*np.pi*eps0)

def esquema():
    fig,(a1,a2)=plt.subplots(1,2,figsize=(9,3))
    # paralelo
    for i,yy in enumerate([0.6,0,-0.6]):
        a1.plot([0,1],[yy,yy],'k-'); a1.add_patch(plt.Rectangle((0.4,yy-0.08),0.05,0.16,color='navy'))
        a1.add_patch(plt.Rectangle((0.55,yy-0.08),0.05,0.16,color='navy'))
    a1.plot([0,0],[-0.6,0.6],'k-'); a1.plot([1,1],[-0.6,0.6],'k-')
    a1.set_title('Paralelo'); a1.set_xlim(-0.3,1.3); a1.set_ylim(-1,1); a1.axis('off')
    # serie
    xs=[0.2,0.5,0.8]
    a2.plot([0,1.1],[0,0],'k-')
    for x in xs:
        a2.add_patch(plt.Rectangle((x-0.02,-0.1),0.02,0.2,color='navy'))
        a2.add_patch(plt.Rectangle((x+0.03,-0.1),0.02,0.2,color='navy'))
    a2.set_title('Serie'); a2.set_xlim(-0.2,1.3); a2.set_ylim(-0.6,0.6); a2.axis('off')
    plt.tight_layout(); plt.show()
esquema()

def asociacion(C1=2.0, C2=3.0, C3=6.0):
    Cpar=C1+C2+C3; Cser=1/(1/C1+1/C2+1/C3)
    plt.figure(figsize=(8,4))
    plt.bar(['C₁','C₂','C₃','Paralelo','Serie'],[C1,C2,C3,Cpar,Cser],
            color=['gray','gray','gray','green','orange'])
    plt.ylabel('Capacitancia (µF)'); plt.title('Comparación de asociaciones'); plt.grid(alpha=.3,axis='y')
    plt.tight_layout(); plt.show()
    print(f"Paralelo: {Cpar:.3g} µF  |  Serie: {Cser:.3g} µF")

interact(asociacion,
    C1=FloatSlider(value=2.0,min=1,max=10,step=0.5,description='C₁ (µF)'),
    C2=FloatSlider(value=3.0,min=1,max=10,step=0.5,description='C₂ (µF)'),
    C3=FloatSlider(value=6.0,min=1,max=10,step=0.5,description='C₃ (µF)'));

def solucion():
    C1,C2,C3=2.0,3.0,6.0
    print(f"Paralelo: C=C₁+C₂+C₃ = {C1+C2+C3:.3g} µF")
    print(f"Serie: 1/C=1/2+1/3+1/6 → C = {1/(1/C1+1/C2+1/C3):.3g} µF")
    print("En paralelo la capacitancia SUMA; en serie es menor que el más pequeño.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("Capacitancia en PARALELO (2, 3, 6 µF):",11,0.03,"µF","C=2+3+6=11 µF.")
quiz_numerico("Capacitancia en SERIE (2, 3, 6 µF):",1.0,0.05,"µF","1/C=1/2+1/3+1/6=1 ⟹ C=1 µF.")
quiz_opcion_multiple("En serie, la capacitancia total es:",
    ["Mayor que todas","Menor que la más pequeña","Igual al promedio","La suma"],1,
    "En serie C es menor que el capacitor más pequeño.")
```

> 🔧 **Aplicación en ingeniería:** un desfibrilador almacena energía en un capacitor y la libera en milisegundos; lo mismo hacen los flashes y los respaldos (UPS). Dimensionar C y V para entregar la energía justa —y hacerlo con seguridad— es ingeniería biomédica y eléctrica.


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

eps0 = 8.85e-12
k = 1/(4*np.pi*eps0)

def esquema():
    fig,ax=plt.subplots(figsize=(5,2.5))
    ax.add_patch(plt.Rectangle((0.3,0.2),0.05,0.6,color='navy'))
    ax.add_patch(plt.Rectangle((0.45,0.2),0.05,0.6,color='navy'))
    ax.plot([0,0.3],[0.5,0.5],'k-'); ax.plot([0.5,0.8],[0.5,0.5],'k-')
    ax.text(0.4,0.9,'C=100 µF',ha='center'); ax.text(0.4,0.05,'V=50 V',ha='center')
    ax.text(1.1,0.5,'⚡ energía\nalmacenada',ha='center',va='center',fontsize=9,color='orange')
    ax.set_xlim(-0.2,1.5); ax.set_ylim(-0.1,1.1); ax.axis('off'); ax.set_title('Capacitor cargado')
    plt.tight_layout(); plt.show()
esquema()

def energia(C_uF=100.0, V=50.0):
    C=C_uF*1e-6; U=0.5*C*V**2; Q=C*V
    vs=np.linspace(0,100,200); Us=0.5*C*vs**2
    plt.figure(figsize=(9,4.2)); plt.plot(vs,Us*1000,lw=2,color='crimson')
    plt.scatter([V],[U*1000],color='blue',s=60,zorder=5,label=f'V={V:.0f} V → U={U*1000:.1f} mJ')
    plt.xlabel('voltaje V (V)'); plt.ylabel('energía U (mJ)'); plt.title('Energía almacenada (U ∝ V²)')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(energia,
    C_uF=FloatSlider(value=100.0,min=10,max=500,step=10,description='C (µF)'),
    V=FloatSlider(value=50.0,min=10,max=100,step=5,description='V (voltios)'));

def solucion():
    C=100e-6; V=50.0
    print(f"Energía U = ½CV² = ½·100e-6·50,0² = {0.5*C*V**2:.3g} J")
    print(f"Carga Q = CV = {C*V:.3g} C = {C*V*1e3:.3g} mC")
    print("La energía crece con el CUADRADO del voltaje: al doblar V, U se cuadruplica.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Energía almacenada (C=100 µF, V=50,0 V):",0.125,0.05,"J","U=½CV²=0,125 J.")
quiz_numerico("Carga almacenada:",5.0,0.05,"mC","Q=CV=5,00 mC.")
quiz_opcion_multiple("Si duplicas el voltaje, la energía:",
    ["Se duplica","Se cuadruplica (U ∝ V²)","No cambia","Se reduce"],1,
    "U=½CV²: al doblar V, U crece 4 veces.")
```
