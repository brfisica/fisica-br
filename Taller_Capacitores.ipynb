## Punto 1 — Capacitor de placas paralelas (Ing. Electrónica/Eléctrica)  ★★  [CL, UFT]
En el diseño de un filtro para una fuente de alimentación, se necesita un capacitor de placas paralelas de área  A = 0,0200 m²  y separación  d = 1,00 mm.

**Capacitancia:**  C = ε₀·A/d       **Carga:**  Q = C·V       **Energía:**  U = ½·C·V²

> 🔧 **Aplicación en ingeniería:** dimensionar filtros, temporizadores y fuentes en circuitos electrónicos parte de calcular C. En subestaciones y equipos eléctricos, la geometría (área y separación) fija la capacitancia y el voltaje que soporta antes de la ruptura dieléctrica.


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


eps0 = 8.85e-12
k = 1/(4*np.pi*eps0)

def esquema():
    fig,(ax1,ax2)=plt.subplots(1,2,figsize=(9,4))
    # IZQUIERDA: más área = más capacidad
    ax1.plot([1,1],[1,4],color='#c0392b',lw=6); ax1.plot([3,3],[1,4],color='#2980b9',lw=6)
    ax1.plot([5,5],[0.5,4.5],color='#c0392b',lw=6); ax1.plot([7,7],[0.5,4.5],color='#2980b9',lw=6)
    ax1.text(2,4.6,'placa pequeña',ha='center',fontsize=8,color='#7f8c8d')
    ax1.text(6,4.9,'placa grande',ha='center',fontsize=8,color='#7f8c8d')
    ax1.annotate('',xy=(6.5,2.5),xytext=(3.5,2.5),arrowprops=dict(arrowstyle='-|>',color='#27ae60',lw=2))
    ax1.text(4.5,2.8,'más área\n→ más C',ha='center',fontsize=9,color='#27ae60',fontweight='bold')
    ax1.set_xlim(0,8); ax1.set_ylim(0,5.3); ax1.axis('off')
    ax1.set_title('Mayor área → mayor capacidad',fontsize=10)
    # DERECHA: menor separación = más capacidad
    ax2.plot([2,2],[1,4],color='#c0392b',lw=6); ax2.plot([2.6,2.6],[1,4],color='#2980b9',lw=6)
    ax2.plot([5,5],[1,4],color='#c0392b',lw=6); ax2.plot([6.8,6.8],[1,4],color='#2980b9',lw=6)
    ax2.annotate('',xy=(2.6,0.6),xytext=(2,0.6),arrowprops=dict(arrowstyle='<->',color='#e67e22',lw=1.5))
    ax2.text(2.3,0.2,'d chica',ha='center',fontsize=8,color='#e67e22')
    ax2.annotate('',xy=(6.8,0.6),xytext=(5,0.6),arrowprops=dict(arrowstyle='<->',color='#e67e22',lw=1.5))
    ax2.text(5.9,0.2,'d grande',ha='center',fontsize=8,color='#e67e22')
    ax2.text(2.3,4.5,'más C',ha='center',fontsize=9,color='#27ae60',fontweight='bold')
    ax2.text(5.9,4.5,'menos C',ha='center',fontsize=9,color='#c0392b',fontweight='bold')
    ax2.set_xlim(1,7.5); ax2.set_ylim(-0.2,5); ax2.axis('off')
    ax2.set_title('Menor separación → mayor capacidad',fontsize=10)
    plt.suptitle('Capacidad: C = ε₀·A/d (geometría del capacitor)',fontsize=11,fontweight='bold',color='#2c3e50')
    plt.tight_layout(); plt.show()
esquema()

def capacitor(A_cm2=200.0, d_mm=1.0, V=12.0):
    A=A_cm2*1e-4; d=d_mm/1000; C=eps0*A/d; Q=C*V; U=0.5*C*V**2
    dd=np.linspace(0.2,5,200)/1000; Cd=eps0*A/dd
    plt.figure(figsize=(9,4.2)); plt.plot(dd*1000,Cd*1e12,lw=2,color='darkred')
    plt.scatter([d_mm],[C*1e12],color='blue',s=60,zorder=5,label=f'd={d_mm:.1f} mm → C={C*1e12:.1f} pF')
    plt.xlabel('separación d (mm)'); plt.ylabel('C (pF)'); plt.title('Capacitancia vs separación')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(capacitor,
    A_cm2=FloatSlider(value=200.0,min=50,max=500,step=25,description='A (cm²)'),
    d_mm=FloatSlider(value=1.0,min=0.2,max=5,step=0.1,description='d (mm)'),
    V=FloatSlider(value=12.0,min=1,max=50,step=1,description='V (voltios)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    A=0.02; d=0.001; C=eps0*A/d; V=12.0
    _pc(f"C = ε₀A/d = (8,85e-12·0,0200)/0,00100 = {C:.3g} F = {C*1e12:.3g} pF")
    _pc(f"Q = CV = {C*V:.3g} C = {C*V*1e9:.3g} nC")
    _pc(f"U = ½CV² = {0.5*C*V**2:.3g} J")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Un capacitor de placas paralelas tiene área de 0,0200 m² y separación de 1,00 mm. Encuentre su capacitancia.",177,0.04,"pF","C=ε₀A/d≈177 pF.", id_preg="Capacito_p1_1", peso=2.0)
quiz_opcion_multiple("En un capacitor de placas paralelas, si se acercan las placas (menor separación), ¿qué ocurre con la capacitancia?",
    ["Disminuye","Aumenta (C ∝ 1/d)","No cambia","Se anula"],1,"C=ε₀A/d: menor d ⟹ mayor C.", id_preg="Capacito_p1_2", peso=2.0)
quiz_opcion_multiple("En un capacitor, Q=C·V. Si se mantiene el mismo capacitor pero se duplica el voltaje aplicado, ¿qué ocurre con la carga almacenada?",["No cambia","Se duplica (Q es proporcional a V)","Se cuadruplica","Se reduce a la mitad"],1,"Para un capacitor dado, la carga es proporcional al voltaje (Q=CV). Al duplicar V se duplica Q. La capacitancia C es una constante geométrica del capacitor, no cambia con el voltaje.", id_preg="Capacito_p1_3", peso=2.0)

# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---
```

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


eps0 = 8.85e-12
k = 1/(4*np.pi*eps0)

def esquema():
    fig,ax=plt.subplots(figsize=(7,4))
    V=np.linspace(0,4,100); Q=1.5*V  # Q=C·V
    ax.plot(V,Q,color='#2980b9',lw=2.5)
    ax.fill_between(V,0,Q,color='#3498db',alpha=0.25)
    ax.text(2.3,2,'Energía = ½·Q·V\n(área del triángulo)',ha='center',fontsize=12,color='#2980b9',fontweight='bold')
    ax.annotate('',xy=(4,6.3),xytext=(0,6.3),arrowprops=dict(arrowstyle='<->',color='#7f8c8d',lw=1))
    ax.text(2,6.5,'voltaje V',ha='center',fontsize=9,color='#7f8c8d')
    ax.set_xlabel('voltaje V'); ax.set_ylabel('carga Q')
    ax.set_title('Energía almacenada en un capacitor (U = ½QV = ½CV²)')
    ax.set_xlim(0,4.3); ax.set_ylim(0,7); ax.grid(alpha=0.2,linestyle='--')
    plt.tight_layout(); plt.show()
esquema()

def energia(C_uF=100.0, V=50.0):
    C=C_uF*1e-6; U=0.5*C*V**2; Q=C*V
    vs=np.linspace(0,100,200); Us=0.5*C*vs**2
    plt.figure(figsize=(9,4.2)); plt.plot(vs,Us*1000,lw=2,color='crimson')
    plt.scatter([V],[U*1000],color='blue',s=60,zorder=5,label=f'V={V:.0f} V → U={U*1000:.1f} mJ')
    plt.xlabel('voltaje V (V)'); plt.ylabel('energía U (mJ)'); plt.title('Energía almacenada (U ∝ V²)')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(energia,
    C_uF=FloatSlider(value=100.0,min=10,max=500,step=10,description='C (µF)'),
    V=FloatSlider(value=50.0,min=10,max=100,step=5,description='V (voltios)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    C=100e-6; V=50.0
    _pc(f"Energía U = ½CV² = ½·100e-6·50,0² = {0.5*C*V**2:.3g} J")
    _pc(f"Carga Q = CV = {C*V:.3g} C = {C*V*1e3:.3g} mC")
    _pc("La energía crece con el CUADRADO del voltaje: al doblar V, U se cuadruplica.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Un capacitor de 100 µF se carga a 50,0 V. Encuentre la energía almacenada.",0.125,0.05,"J","U=½CV²=0,125 J.", id_preg="Capacito_p4_9", peso=3.0)
quiz_opcion_multiple("La energía de un capacitor es ½CV². Si se duplica el voltaje, ¿cómo cambia la energía almacenada?",["Se duplica","Se cuadruplica (la energía va con V²)","No cambia","Se reduce a la mitad"],1,"La energía va con el CUADRADO del voltaje (½CV²). Al duplicar V, la energía se cuadruplica. Por eso los capacitores de alto voltaje almacenan mucha más energía y pueden ser peligrosos.", id_preg="Capacito_p4_10", peso=3.0)
quiz_opcion_multiple("En un capacitor, si se duplica el voltaje aplicado, ¿qué ocurre con la energía almacenada?",
    ["Se duplica","Se cuadruplica (U ∝ V²)","No cambia","Se reduce"],1,
    "U=½CV²: al doblar V, U crece 4 veces.", id_preg="Capacito_p4_11", peso=3.0)
```
