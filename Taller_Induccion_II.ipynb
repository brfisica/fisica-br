## Punto 1 — Generador síncrono de imanes permanentes (PMSG) (Ing. Eléctrica/Energética)  ★★★  [PRC, UCM, ULC]
El rotor de un PMSG lleva imanes permanentes; al girar con velocidad angular ω, el campo que atraviesa
la bobina del estátor varía como B(t) = B₀·cos(ωt), induciendo una FEM alterna sinusoidal.

**Datos:** N = 200 espiras,  A = 50,0 cm² = 50,0×10⁻⁴ m²,  B₀ = 0,800 T,  n = 1500 rpm,
R_bob = 4,00 Ω,  R_L = 20,0 Ω.

**Velocidad angular:**  ω = 2πn/60       **Flujo concatenado:**  Λ(t) = N·B₀·A·cos(ωt)
**FEM (Faraday):**  ℰ(t) = −dΛ/dt = N·B₀·A·ω·sen(ωt)   ⟹   **ℰ_pico = N·B₀·A·ω**
**Circuito resistivo:**  I_pico = ℰ_pico/(R_bob+R_L) ,  V_L,pico = I_pico·R_L ,  P = I²·R

> 🔧 **Aplicación en ingeniería:** los PMSG son el corazón de los aerogeneradores de pequeña y mediana
> potencia y de muchos vehículos eléctricos. La FEM crece con la velocidad de giro (ℰ_pico ∝ ω), por eso
> los aerogeneradores de velocidad variable ajustan las rpm para entregar el voltaje deseado. Que la
> resistencia de carga sea mucho mayor que la de la bobina (R_L ≫ R_bob) es clave para que la mayor
> parte de la potencia llegue a la red y no se pierda como calor en el devanado.

## Punto 2 — Generador de una central hidroeléctrica: Hidroituango (Ing. Eléctrica/Civil/Energética)  ★★★  [PRC, UCM, ULC]
En la central Ituango (río Cauca, Antioquia) cada unidad tiene una turbina Francis unida por el eje a un generador
síncrono de eje vertical. El agua del embalse mueve la turbina; el rotor del generador —un electroimán con muchos
polos N y S alternados— gira dentro del estátor, y el flujo de cada polo que barre las bobinas del estátor induce
la FEM (ley de Faraday). Es el mismo principio del Punto 1, a escala de cientos de megavatios.

**Datos públicos de una unidad:**  P_el = 300 MW,  Q = 168,75 m³/s,  salto neto H = 197 m,  n = 180 rpm,
frecuencia de la red colombiana f = 60,0 Hz.
**Datos didácticos (supuestos, no del fabricante):**  B_med = 0,800 T en la cara de cada polo,  A_polo = 1,50 m²,
N = 32 espiras en serie por fase, devanado ideal.

**Potencia del agua:**  P_h = ρ·g·Q·H   (ρ = 1000 kg/m³, g = 9,81 m/s²)       **Eficiencia global:**  η = P_el/P_h
**Frecuencia y polos:**  f = p·n/120   (p = número de polos)
**Flujo por polo:**  Φ = B_med·A_polo      **FEM por fase (Faraday):**  ℰ_pico = N·Φ·ω_e ,  ω_e = 2π·f ,  ℰ_rms = ℰ_pico/√2
**Par en el eje:**  τ = P_el/ω_m ,  ω_m = 2π·n/60

> 🔧 **Aplicación en ingeniería:** Hidroituango tiene 8 unidades de 300 MW (2400 MW, cerca del 17 % de la demanda
> eléctrica del país). Una turbina Francis con 197 m de salto gira lento (180 rpm), así que para entregar 60 Hz el
> rotor necesita muchos polos: por eso los generadores hidráulicos tienen rotores de gran diámetro, mientras que un
> turbogenerador térmico de 2 polos gira a 3600 rpm. Cuando la demanda sube, crece la corriente del estátor y con
> ella el par de frenado electromagnético (ley de Lenz, taller Inducción I): el regulador de velocidad debe abrir el paso de
> agua para mantener las 180 rpm y, con ellas, los 60 Hz de la red.
> *Fuentes de los datos públicos:* hidroituango.com.co (300 MW y 168,75 m³/s por unidad); EPM, citada por El Tiempo
> y abceconomia (180 rpm, rotor de unos 40 polos); GlobalData y NS Energy (salto neto 197 m).

## Punto 3 — Sensor inductivo de posición (encoder de reluctancia variable, VRS) (Ing. Mecatrónica/Automotriz)  ★★★  [PRC, UCM, ULC]
Al girar una rueda dentada frente a un imán con bobina, dientes y valles alternan el entrehierro y
modulan el flujo. El flujo varía como Φ(t) = Φ_med + (ΔΦ/2)·cos(2π·f_d·t), y solo la parte oscilante
induce FEM.

**Datos:** N = 1200 vueltas,  Φ_máx = 8,00×10⁻⁶ Wb,  Φ_mín = 2,00×10⁻⁶ Wb,  Z = 60 dientes,
n = 3000 rpm,  R_s = 860 Ω,  R_ECU = 10,0 kΩ.

**Frecuencias:**  f_rot = n/60 ,  f_d = Z·f_rot       **Flujo:**  Φ_med = (Φ_máx+Φ_mín)/2 ,  ΔΦ = Φ_máx−Φ_mín
**FEM (Faraday):**  ℰ(t) = −N·dΦ/dt   ⟹   **ℰ_pico = N·π·f_d·ΔΦ**       **Divisor:**  V_ECU = ℰ_pico·R_ECU/(R_s+R_ECU)

> 🔧 **Aplicación en ingeniería:** los sensores VRS miden velocidad y posición en sistemas ABS, inyección
> electrónica y transmisiones automáticas. La frecuencia de la señal es proporcional a las rpm, así que el
> ECU cuenta pulsos para conocer la velocidad en tiempo real. Un límite real: a muy bajas rpm la amplitud
> de la FEM cae (ℰ_pico ∝ f_d ∝ n), lo que dificulta la detección justo cuando el vehículo casi se detiene
> — por eso el ABS moderno complementa o reemplaza los VRS con sensores de efecto Hall a baja velocidad.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# Calificación formativa del taller
calificacion_final(total_preguntas=15)
```
