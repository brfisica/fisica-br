## Punto 1 — Generador síncrono de imanes permanentes (PMSG) (Ing. Eléctrica/Energética)  ★★★  [PRC, UCM, ULC]
El rotor de un PMSG lleva imanes permanentes; al girar con velocidad angular ω, el campo que atraviesa
la bobina del estátor varía como B(t) = B₀·cos(ωt), induciendo una FEM alterna sinusoidal.

**Datos:** N = 200 espiras,  A = 50,0 cm² = 50,0×10⁻⁴ m²,  B₀ = 0,800 T,  n = 1500 rpm,
R_bob = 4,00 Ω,  R_L = 20,0 Ω.

**Velocidad angular:**  ω = 2πn/60       **Flujo concatenado:**  Λ(t) = N·B₀·A·cos(ωt)
**FEM (Faraday):**  ℰ(t) = −dΛ/dt = N·B₀·A·ω·sen(ωt)   ⟹   **ℰ_pico = N·B₀·A·ω**
**Circuito resistivo:**  I_pico = ℰ_pico/(R_bob+R_L) ,  V_L,pico = I_pico·R_L ,  P = I²·R

> 🔧 **Aplicación en ingeniería:** los PMSG son el corazón de los aerogeneradores de pequeña y mediana
> potencia y de muchos vehículos eléctricos. La FEM crece con la velocidad de giro (ℰ_pico ∝ ω), por eso
> los aerogeneradores de velocidad variable ajustan las rpm para entregar el voltaje deseado. Que la
> resistencia de carga sea mucho mayor que la de la bobina (R_L ≫ R_bob) es clave para que la mayor
> parte de la potencia llegue a la red y no se pierda como calor en el devanado.

## Punto 2 — Generador de una central hidroeléctrica: Hidroituango (Ing. Eléctrica/Civil/Energética)  ★★★  [PRC, UCM, ULC]
En la central Ituango (río Cauca, Antioquia) cada unidad tiene una turbina Francis unida por el eje a un generador
síncrono de eje vertical. El agua del embalse mueve la turbina; el rotor del generador —un electroimán con muchos
polos N y S alternados— gira dentro del estátor, y el flujo de cada polo que barre las bobinas del estátor induce
la FEM (ley de Faraday). Es el mismo principio del Punto 1, a escala de cientos de megavatios.

**Datos públicos de una unidad:**  P_el = 300 MW,  Q = 168,75 m³/s,  salto neto H = 197 m,  n = 180 rpm,
frecuencia de la red colombiana f = 60,0 Hz.
**Datos didácticos (supuestos, no del fabricante):**  B_med = 0,800 T en la cara de cada polo,  A_polo = 1,50 m²,
N = 32 espiras en serie por fase, devanado ideal.

**Potencia del agua:**  P_h = ρ·g·Q·H   (ρ = 1000 kg/m³, g = 9,81 m/s²)       **Eficiencia global:**  η = P_el/P_h
**Frecuencia y polos:**  f = p·n/120   (p = número de polos)
**Flujo por polo:**  Φ = B_med·A_polo      **FEM por fase (Faraday):**  ℰ_pico = N·Φ·ω_e ,  ω_e = 2π·f ,  ℰ_rms = ℰ_pico/√2
**Par en el eje:**  τ = P_el/ω_m ,  ω_m = 2π·n/60

> 🔧 **Aplicación en ingeniería:** Hidroituango tiene 8 unidades de 300 MW (2400 MW, cerca del 17 % de la demanda
> eléctrica del país). Una turbina Francis con 197 m de salto gira lento (180 rpm), así que para entregar 60 Hz el
> rotor necesita muchos polos: por eso los generadores hidráulicos tienen rotores de gran diámetro, mientras que un
> turbogenerador térmico de 2 polos gira a 3600 rpm. Cuando la demanda sube, crece la corriente del estátor y con
> ella el par de frenado electromagnético (ley de Lenz, taller Inducción I): el regulador de velocidad debe abrir el paso de
> agua para mantener las 180 rpm y, con ellas, los 60 Hz de la red.
> *Fuentes de los datos públicos:* hidroituango.com.co (300 MW y 168,75 m³/s por unidad); EPM, citada por El Tiempo
> y abceconomia (180 rpm, rotor de unos 40 polos); GlobalData y NS Energy (salto neto 197 m).

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---


def _c3(x):
    # 3 cifras significativas conservando ceros finales (8,00 y no 8); notación ×10ⁿ desde 1000
    if x != 0 and abs(x) >= 1000: return _sci(x)
    t = f"{x:#.3g}"
    return t[:-1] if t.endswith('.') else t

def _sci(x):
    e = int(np.floor(np.log10(abs(x)))); m = x/10**e
    if round(m, 2) >= 10: m /= 10; e += 1
    return f"{m:.2f}×10" + str(e).translate(str.maketrans('-0123456789', '⁻⁰¹²³⁴⁵⁶⁷⁸⁹'))

def esquema():
    fig,(ax1,ax2)=plt.subplots(1,2,figsize=(9.5,3.8))
    th=np.linspace(0,2*np.pi,300)
    ax1.plot(1.65*np.cos(th),1.65*np.sin(th),color='#34495e',lw=6,alpha=0.6)   # estátor
    ax1.plot(1.05*np.cos(th),1.05*np.sin(th),color='#7f8c8d',lw=1)            # cubo del rotor
    for k in range(40):
        an=2*np.pi*k/40; col='#c0392b' if k%2==0 else '#2980b9'
        ax1.plot([1.08*np.cos(an),1.42*np.cos(an)],[1.08*np.sin(an),1.42*np.sin(an)],color=col,lw=3.2,solid_capstyle='butt')
    ax1.plot([0],[1.65],'s',color='#e67e22',ms=12)
    ax1.text(0,1.95,'bobina del estátor',ha='center',fontsize=8,color='#e67e22')
    ax1.annotate('',xy=(1.75,-0.9),xytext=(1.75,0.9),arrowprops=dict(arrowstyle='->',color='#27ae60',lw=2,connectionstyle='arc3,rad=-0.45'))
    ax1.text(1.55,1.2,'giro',color='#27ae60',fontsize=9,fontweight='bold',va='center')
    ax1.text(0,0,'40 polos\n180 rpm',ha='center',va='center',fontsize=10,fontweight='bold',color='#2c3e50')
    ax1.text(0,-2.15,'rojo = N   ·   azul = S',ha='center',fontsize=8,color='#7f8c8d')
    ax1.set_xlim(-2.1,2.1); ax1.set_ylim(-2.35,2.2); ax1.axis('off'); ax1.set_aspect('equal')
    ax1.set_title('Rotor de polos salientes (vista superior)',fontsize=10)
    ang=np.linspace(0,360,2000)
    ax2.plot(ang,np.cos(np.radians(20*ang)),color='#8e44ad',lw=1.2)
    ax2.set_xlabel('ángulo girado por el rotor (°)'); ax2.set_ylabel('Φ en la bobina / Φ_polo')
    ax2.set_title('Una vuelta del rotor = 20 ciclos eléctricos',fontsize=10)
    ax2.set_xlim(0,360)
    plt.tight_layout(); plt.show()
esquema()

ETA = 0.920   # eficiencia global calculada con los datos públicos (300 MW / 326 MW); se asume constante
def hidro(Q=168.75, H=197.0, n_rpm=180.0, polos=40, B=0.80, Ap=1.50, N=32):
    Ph=1000*9.81*Q*H; Pel=ETA*Ph
    f=polos*n_rpm/120; we=2*np.pi*f; wm=2*np.pi*n_rpm/60
    Phi=B*Ap; Ep=N*Phi*we; tau=Pel/wm
    print_es(f"P_h = ρ·g·Q·H = {_c3(Ph/1e6)} MW   |   P_el = η·P_h = {_c3(Pel/1e6)} MW  (η = 0,920 supuesta constante)")
    print_es(f"f = p·n/120 = {_c3(f)} Hz   |   ω_e = {_c3(we)} rad/s   |   ω_m = {_c3(wm)} rad/s")
    print_es(f"Φ_polo = {_c3(Phi)} Wb   |   ℰ_pico = {_c3(Ep/1e3)} kV   |   ℰ_rms = {_c3(Ep/np.sqrt(2)/1e3)} kV")
    print_es(f"τ = P_el/ω_m = {_c3(tau/1e6)} MN·m")
    if abs(f-60.0) > 0.05:
        print_es(f"⚠ f = {_c3(f)} Hz ≠ 60,0 Hz: este generador no se puede sincronizar con la red colombiana.")
    fig,(a1,a2)=plt.subplots(1,2,figsize=(9,3.4))
    a1.bar(['P del agua','P eléctrica'],[Ph/1e6,Pel/1e6],color=['#2980b9','#27ae60'])
    for i,v in enumerate([Ph/1e6,Pel/1e6]): a1.text(i,v*1.01,f'{v:.0f} MW',ha='center',fontsize=9,fontweight='bold')
    a1.set_ylabel('MW'); a1.set_title('Del agua a la red',fontsize=10)
    ns=np.linspace(60,600,200)
    a2.plot(ns,polos*ns/120,color='#8e44ad',lw=2,label=f'{polos} polos')
    a2.axhline(60,ls='--',color='#c0392b',label='red: 60 Hz'); a2.scatter([n_rpm],[f],color='#2c3e50',zorder=5)
    a2.set_xlabel('n (rpm)'); a2.set_ylabel('f (Hz)'); a2.set_ylim(0,200); a2.legend(fontsize=8)
    a2.set_title('f = p·n/120',fontsize=10)
    plt.tight_layout(); plt.show()


def solucion():
    Q=168.75; H=197.0; P=300e6; n=180.0; f=60.0
    Ph=1000*9.81*Q*H
    print_es(f"a) P_h = 1000·9,81·168,75·197 = {_sci(Ph)} W = {Ph/1e6:.0f} MW ;  η = 300/{Ph/1e6:.0f} = {P/Ph:.3f} ({P/Ph*100:.1f} %)")
    print_es(f"   El {100-P/Ph*100:.0f} % restante se pierde en la turbina y el generador (H ya es el salto NETO).")
    p=120*f/n
    print_es(f"b) p = 120·f/n = 120·60,0/180 = {p:.0f} polos (coincide con los ~40 polos informados por EPM).")
    print_es(f"   Con 2 polos haría falta n = 120·60,0/2 = 3600 rpm: imposible para esta turbina.")
    Phi=0.800*1.50
    print_es(f"c) Φ_polo = B_med·A_polo = 0,800·1,50 = {Phi:.2f} Wb")
    we=2*np.pi*f; wm=2*np.pi*n/60; Ep=32*Phi*we
    print_es(f"d) ω_e = 2π·60,0 = {we:.0f} rad/s ;  ℰ_pico = N·Φ·ω_e = 32·1,20·{we:.0f} = {_sci(Ep)} V = {Ep/1e3:.1f} kV")
    print_es(f"   ℰ_rms = ℰ_pico/√2 = {Ep/np.sqrt(2)/1e3:.1f} kV")
    print_es(f"   Ojo: con ω_m = 2π·180/60 = {wm:.1f} rad/s saldría {32*Phi*wm:.0f} V, 20 veces menos. Con 40 polos el flujo")
    print_es(f"   en cada bobina completa 20 ciclos por vuelta: ω_e = (p/2)·ω_m = 20·(2π·180/60) = {20*wm:.0f} rad/s.")
    tau=P/wm
    print_es(f"e) τ = P_el/ω_m = 300×10⁶/(2π·180/60) = {_sci(tau)} N·m = {tau/1e6:.1f} MN·m")
    print_es(f"   Es el par de frenado electromagnético (Lenz) que la turbina debe vencer en cada instante.")
    print_es(f"Verificación: ℰ_rms = 4,44·f·N·Φ = 4,44·60,0·32·1,20 = {4.44*60*32*1.2/1e3:.1f} kV ✓ (fórmula clásica de máquinas, devanado ideal)")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_numerico("Potencia hidráulica disponible en una unidad, P_h = ρ·g·Q·H, con Q = 168,75 m³/s y H = 197 m (en MW):",
    326,0.03,"MW","P_h=1000·9,81·168,75·197=3,26×10⁸ W=326 MW; η=300/326=0,920.", id_preg="Induc2_p2_1", peso=2.0)
quiz_numerico("¿Cuántos polos necesita el rotor para generar a 60,0 Hz girando a 180 rpm? (f = p·n/120)",
    40,0.02,"polos","p=120·60,0/180=40,0 polos.", id_preg="Induc2_p2_2", peso=2.0)
quiz_numerico("FEM de pico por fase ℰ_pico = N·Φ·ω_e con N = 32, Φ = 1,20 Wb y f = 60,0 Hz (en kV):",
    14.5,0.04,"kV","ω_e=2π·60,0=377 rad/s; ℰ_pico=32·1,20·377=1,45×10⁴ V=14,5 kV.", id_preg="Induc2_p2_3", peso=3.0)
quiz_numerico("Par electromagnético en el eje, τ = P_el/ω_m, con P_el = 300 MW y n = 180 rpm (en MN·m):",
    15.9,0.04,"MN·m","ω_m=2π·180/60=18,8 rad/s; τ=300×10⁶/18,8=1,59×10⁷ N·m=15,9 MN·m.", id_preg="Induc2_p2_4", peso=2.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: un estudiante usa la fórmula del Punto 1 con ω = 2π·n/60 = 18,8 rad/s, obtiene ℰ_pico = 724 V y concluye: 'el generador de Hidroituango produce menos de 1 kV; por eso la central necesita transformadores elevadores'. ¿Correcto o solo convincente?",
    ['Solo convincente: con 40 polos el flujo de cada bobina oscila 20 veces por vuelta; en Faraday va ω_e = 377 rad/s',
     'Correcto: la FEM crece con la velocidad del rotor, y 180 rpm es una velocidad de giro muy baja',
     'Correcto: por eso la central tiene transformadores que elevan la tensión hacia la subestación de 500 kV',
     'Solo convincente: el error está en el flujo; con 40 polos hay que multiplicar Φ_polo por 40'],
    0,
    'En Faraday entra la rapidez con que cambia el flujo en la BOBINA, no la del eje: ω_e = (p/2)·ω_m. Los transformadores elevadores existen, pero porque transmitir a alta tensión reduce las pérdidas I²·R, no porque el generador dé 724 V. La opción «el error está en el flujo…» acierta el veredicto con una corrección falsa.',
    id_preg='Induc2_p2_concep_2', peso=2.0)

def _explora_sin_cifras(fn):
    # Opción 3: exploración cualitativa; las cifras de este punto solo están en «Mostrar solución».
    import functools, io, contextlib, re as _re
    _pat = _re.compile(r'(?<![\w.,/])[-−]?\d+(?:[.,]\d+)?(?:\s?×\s?10[⁻⁰¹²³⁴⁵⁶⁷⁸⁹]+|[eE][+-]?\d+)?(?![\wπ·(]|[.,]\d)')
    def _m(t): return _pat.sub('…', t)
    def _ocultar():
        for num in plt.get_fignums():
            fig = plt.figure(num)
            textos = list(fig.texts) + ([fig._suptitle] if getattr(fig, '_suptitle', None) else [])
            for ax in fig.axes:
                textos += [ax.title, ax._left_title, ax._right_title] + list(ax.texts)
                if ax.get_legend(): textos += list(ax.get_legend().get_texts())
            for t in textos: t.set_text(_m(t.get_text()))
    @functools.wraps(fn)
    def envuelta(*a, **k):
        _show = plt.show
        def _show2(*sa, **sk):
            _ocultar(); _show(*sa, **sk)
        buf = io.StringIO(); plt.show = _show2
        try:
            with contextlib.redirect_stdout(buf): fn(*a, **k)
            _ocultar()
        finally:
            plt.show = _show
        for linea in buf.getvalue().splitlines():
            print(linea if ('⚠' in linea or '⛔' in linea) else _m(linea))
        print("(Exploración cualitativa: observa tendencias. Las cifras de este punto están en «Mostrar solución».)")
    return envuelta

print("\n--- 🔬 Explora después de responder ---")
print("Los controles NO arrancan en los datos del enunciado. Primero resuelve y responde;\nluego explora tendencias, casos límite y comprueba tu razonamiento.")
interact(_explora_sin_cifras(hidro),
    Q=FloatSlider(value=218.75,min=50,max=250,step=1.25,description='Q (m³/s)'),
    H=FloatSlider(value=335.0,min=50,max=600,step=1,description='H (m)'),
    n_rpm=FloatSlider(value=300.0,min=60,max=600,step=10,description='n (rpm)'),
    polos=widgets.IntSlider(value=24,min=2,max=80,step=2,description='polos p'),
    B=FloatSlider(value=1.00,min=0.4,max=1.2,step=0.05,description='B_med (T)'),
    Ap=FloatSlider(value=2.1,min=0.5,max=3.0,step=0.1,description='A_polo (m²)'),
    N=widgets.IntSlider(value=46,min=8,max=64,step=2,description='N espiras'));
```

## Punto 3 — Transformador elevador: de la unidad de Hidroituango a la red de 500 kV (Ing. Eléctrica/Potencia)  ★★★  [PRC, UCM, ULC]
La energía que sale del generador del Punto 2 no viaja a su tensión de generación: cada unidad de Ituango tiene
tres transformadores monofásicos (uno por fase) que elevan la tensión hacia la subestación de 500 kV, desde donde
sale la línea al Sistema Interconectado Nacional. Cada transformador toma un tercio de la potencia de la unidad.

**Datos públicos:**  P_unidad = 300 MW,  tres transformadores monofásicos por unidad,  red de transmisión de 500 kV.
**Datos didácticos (supuestos, no del fabricante):**  tensión del lado de baja V_p = 10,0 kV (rms, del orden del ℰ_rms

## Punto 4 — Sensor inductivo de posición (encoder de reluctancia variable, VRS) (Ing. Mecatrónica/Automotriz)  ★★★  [PRC, UCM, ULC]
Al girar una rueda dentada frente a un imán con bobina, dientes y valles alternan el entrehierro y
modulan el flujo. El flujo varía como Φ(t) = Φ_med + (ΔΦ/2)·cos(2π·f_d·t), y solo la parte oscilante
induce FEM.

**Datos:** N = 1200 vueltas,  Φ_máx = 8,00×10⁻⁶ Wb,  Φ_mín = 2,00×10⁻⁶ Wb,  Z = 60 dientes,
n = 3000 rpm,  R_s = 860 Ω,  R_ECU = 10,0 kΩ.

**Frecuencias:**  f_rot = n/60 ,  f_d = Z·f_rot       **Flujo:**  Φ_med = (Φ_máx+Φ_mín)/2 ,  ΔΦ = Φ_máx−Φ_mín
**FEM (Faraday):**  ℰ(t) = −N·dΦ/dt   ⟹   **ℰ_pico = N·π·f_d·ΔΦ**       **Divisor:**  V_ECU = ℰ_pico·R_ECU/(R_s+R_ECU)

> 🔧 **Aplicación en ingeniería:** los sensores VRS miden velocidad y posición en sistemas ABS, inyección
> electrónica y transmisiones automáticas. La frecuencia de la señal es proporcional a las rpm, así que el
> ECU cuenta pulsos para conocer la velocidad en tiempo real. Un límite real: a muy bajas rpm la amplitud
> de la FEM cae (ℰ_pico ∝ f_d ∝ n), lo que dificulta la detección justo cuando el vehículo casi se detiene
> — por eso el ABS moderno complementa o reemplaza los VRS con sensores de efecto Hall a baja velocidad.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# Calificación formativa del taller
calificacion_final(total_preguntas=20)
```
