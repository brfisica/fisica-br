velocidad  v₁ = 2,00 m/s; luego el fluido acelera a  v₂ = 5,00 m/s. Se busca la presión  P₂.

**Bernoulli (horizontal):**  P₁ + ½·ρ·v₁² = P₂ + ½·ρ·v₂²

> 🔧 **Aplicación en ingeniería:** Bernoulli explica que donde el fluido va más rápido, la presión baja.
> Es la base de la sustentación de alas, los medidores de flujo (Venturi), los atomizadores y el
> diseño de tuberías. Un error de presión puede causar cavitación y dañar bombas.
