## Punto 1 — Terminal esférico de alta tensión (Ing. Eléctrica/Electrónica)
Una terminal esférica conductora de radio  R = 10,0 cm  acumula carga  Q = +50,0 nC.

**Potencial (carga puntual/esfera):**   V(r) = k·Q / r      (para r ≥ R)

> 🔧 **Aplicación en ingeniería:** en terminales de alta tensión y pararrayos, el potencial en la superficie determina el riesgo de descarga. Calcularlo permite fijar distancias de seguridad y aislamiento según la normativa RETIE.

## Punto 2 — Superposición de dos cargas (Ing. Electrónica/Biomédica)

> 🔧 **Aplicación en ingeniería:** el mapa de potencial de dos cargas modela configuraciones reales como electrodos, líneas de transmisión paralelas o sensores capacitivos. Entender la superposición es clave para diseñar y blindar estos sistemas.

En una dimensión, E = −dV/dx. Donde las equipotenciales están más juntas, el campo es más intenso.

**Relación fundamental:**  E = −dV/dx  (unidad V/m = N/C)

## Punto 6 — Superposición escalar: potencial y energía de un sistema de cargas  ★★  [UCM, PRC]
El potencial es **escalar**: para varias cargas se suman los valores algebraicamente (con su signo), sin
componentes ni ángulos. Esto es mucho más simple que sumar campos vectoriales.

**Potencial de n cargas:**  V = k·Σ(qᵢ/rᵢ)   ·   **Energía del sistema:**  U = k·Σ(qᵢqⱼ/rᵢⱼ)

> 🔧 **Aplicación en ingeniería:** la energía de un sistema de cargas es la base para entender la energía
> almacenada en capacitores y en configuraciones de electrodos. El trabajo para ensamblar las cargas desde
> el infinito es exactamente esa energía, clave en el diseño de dispositivos de alta tensión.

## Punto 8 — Precipitador electrostático: limpieza de gases industriales (Ing. Ambiental/Eléctrica)  ★★★  [PRC, UCM, ULC]
Un precipitador electrostático captura partículas (ceniza, polvo) de los gases de una chimenea. Su
geometría es un **cilindro coaxial**: un alambre central delgado a alto voltaje negativo y un tubo
colector externo conectado a tierra. El alto campo cerca del alambre ioniza el aire (**efecto corona**),
las partículas se cargan y el campo las empuja hacia la pared, donde se adhieren y luego se retiran.

**Datos:** alambre a = 1,00 mm de radio,  tubo b = 15,0 cm de radio,  rigidez dieléctrica del aire
E_ruptura ≈ 3,00×10⁶ V/m.

**Campo en un cilindro coaxial:**  E(r) = V / [r·ln(b/a)]   (máximo en la superficie del alambre, r = a)
**Voltaje de inicio de corona:**  V_corona = E_ruptura · a · ln(b/a)
**Carga de saturación de una partícula (radio d):**  q ≈ 3ε₀·[κ/(κ+2)]·π·d²·E
**Velocidad de migración (arrastre de Stokes):**  w = q·E / (3π·η·d)

> 🔧 **Aplicación en ingeniería:** los precipitadores electrostáticos son el sistema de control de
> material particulado en termoeléctricas (p. ej. las del Cesar), cementeras y calderas industriales,
> con eficiencias superiores al 99 %. El diseño equilibra dos límites: el voltaje debe ser suficiente
> para iniciar y sostener la corona que carga las partículas, pero sin llegar a la ruptura total del
> gas entre alambre y pared (arco eléctrico), que cortocircuitaría el equipo. Por eso se opera en una
> ventana de voltaje cuidadosamente controlada.
