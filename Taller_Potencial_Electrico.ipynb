## Punto 1 — Terminal esférico de alta tensión (Ing. Eléctrica/Electrónica)
Una terminal esférica conductora de radio  R = 10,0 cm  acumula una carga  Q = +50,0 nC.

**Potencial de una carga puntual/esfera:**   V(r) = k·Q / r      (para r ≥ R)

Se estudia el potencial alrededor de la terminal y el trabajo para acercar cargas.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))
# --- fin setup ---


k = 8.99e9  # N m^2 / C^2

def potencial_esfera(Q_nC=50.0, R_cm=10.0, r_cm=30.0):
    Q = Q_nC*1e-9; R = R_cm/100; rp = r_cm/100
    r = np.linspace(R, 1.0, 500)
    V = k*Q/r
    Vp = k*Q/max(rp, R)
    Vsurf = k*Q/R
    plt.figure(figsize=(9,5))
    plt.plot(r*100, V, lw=2, color='darkred')
    plt.axvline(R_cm, color='gray', ls='--', label=f'superficie R={R_cm:.0f} cm')
    plt.scatter([max(r_cm,R_cm)],[Vp], color='blue', s=60, zorder=5, label=f'r={r_cm:.0f} cm')
    plt.xlabel('r (cm)'); plt.ylabel('V(r)  (V)')
    plt.title('Potencial alrededor de la terminal esférica')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()
    print(f"V en la superficie (r=R) = {Vsurf:.1f} V ≈ {Vsurf:.3g} V")
    print(f"V en r={r_cm:.0f} cm = {Vp:.1f} V")
    # trabajo para traer q=+2nC desde el infinito hasta r
    q = 2e-9
    print(f"Trabajo para traer q=+2,00 nC desde ∞ hasta r={r_cm:.0f} cm:  W = q·V = {q*Vp:.3e} J")

interact(potencial_esfera,
    Q_nC=FloatSlider(value=50.0, min=-100, max=100, step=5, description='Q (nC)'),
    R_cm=FloatSlider(value=10.0, min=5, max=30, step=1, description='R (cm)'),
    r_cm=FloatSlider(value=30.0, min=5, max=90, step=5, description='r (cm)'));

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("a) Potencial en la superficie (r=R=10 cm, Q=50 nC):", 4495.9, 0.03, "V",
    "V = kQ/R = 8,99e9·50e-9/0,10 ≈ 4,50×10³ V.")
quiz_numerico("a) Potencial en r=30 cm:", 1498.6, 0.03, "V",
    "V = kQ/r = 8,99e9·50e-9/0,30 ≈ 1,50×10³ V.")
quiz_opcion_multiple("b) Si duplicas la distancia r, el potencial V:",
    ["Se duplica", "Se reduce a la mitad (V ∝ 1/r)", "No cambia", "Se reduce a un cuarto"],
    1, "El potencial va como 1/r: al doblar r, V cae a la mitad. (El campo sí va como 1/r².)")
```

## Punto 2 — Superposición de dos cargas (Ing. Electrónica/Biomédica)
Dos cargas puntuales:  Q₁ = +30,0 nC  en  x = −10 cm  y  Q₂ = −30,0 nC  en  x = +10 cm
(un dipolo, modelo de electrodos o sensores).

**Potencial total (superposición):**   V = k·Q₁/r₁ + k·Q₂/r₂

donde r₁ y r₂ son las distancias del punto a cada carga.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))
# --- fin setup ---


k = 8.99e9

def dos_cargas(Q1_nC=30.0, Q2_nC=-30.0, x_cm=5.0, y_cm=0.0):
    Q1 = Q1_nC*1e-9; Q2 = Q2_nC*1e-9
    x1, x2 = -0.10, 0.10  # posiciones de las cargas
    px, py = x_cm/100, y_cm/100
    r1 = np.hypot(px-x1, py); r2 = np.hypot(px-x2, py)
    r1 = max(r1,1e-3); r2 = max(r2,1e-3)
    V = k*Q1/r1 + k*Q2/r2

    # Mapa de potencial
    xs = np.linspace(-0.3,0.3,200); ys = np.linspace(-0.2,0.2,150)
    X,Y = np.meshgrid(xs,ys)
    R1 = np.hypot(X-x1,Y); R2 = np.hypot(X-x2,Y)
    R1[R1<1e-3]=1e-3; R2[R2<1e-3]=1e-3
    Vmap = k*Q1/R1 + k*Q2/R2
    plt.figure(figsize=(9,5))
    lim = 4000
    cf = plt.contourf(X*100,Y*100,np.clip(Vmap,-lim,lim),levels=30,cmap='RdBu_r')
    plt.colorbar(cf,label='V (V)')
    plt.scatter([x1*100],[0],color='red',s=80,marker='+',label='Q₁')
    plt.scatter([x2*100],[0],color='blue',s=80,marker='_',label='Q₂')
    plt.scatter([x_cm],[y_cm],color='black',s=50,zorder=5,label='punto')
    plt.xlabel('x (cm)'); plt.ylabel('y (cm)')
    plt.title('Mapa de potencial de dos cargas'); plt.legend(); plt.tight_layout(); plt.show()
    print(f"En el punto ({x_cm:.0f}, {y_cm:.0f}) cm:  r₁={r1*100:.1f} cm, r₂={r2*100:.1f} cm")
    print(f"V total = {V:.3g} V")

interact(dos_cargas,
    Q1_nC=FloatSlider(value=30.0, min=-50, max=50, step=5, description='Q₁ (nC)'),
    Q2_nC=FloatSlider(value=-30.0, min=-50, max=50, step=5, description='Q₂ (nC)'),
    x_cm=FloatSlider(value=5.0, min=-25, max=25, step=1, description='x (cm)'),
    y_cm=FloatSlider(value=0.0, min=-15, max=15, step=1, description='y (cm)'));

print("\n--- Cuestionario Punto 2 ---")
quiz_opcion_multiple("a) ¿Cuánto vale V en la mediatriz (puntos equidistantes de ambas cargas)?",
    ["Máximo", "Cero: los potenciales +Q y −Q se cancelan", "Infinito", "Igual a kQ/r"],
    1, "En la mediatriz r₁=r₂ y como Q₂=−Q₁, V = kQ/r − kQ/r = 0.")
quiz_numerico("a) V total en el punto (x=5 cm, y=0):", -3600, 0.08, "V",
    "V = kQ₁/r₁ + kQ₂/r₂ con r₁=15 cm, r₂=5 cm ≈ −3,6×10³ V.")
quiz_opcion_multiple("b) El potencial es una magnitud:",
    ["Vectorial (se suma con componentes)", "Escalar (se suma directamente con signo)",
     "Siempre positiva", "Sin unidades"],
    1, "V es escalar: se suman los aportes con su signo, sin ángulos ni componentes.")
```

## Punto 3 — Capacitor de placas paralelas (Ing. Mecánica/Sistemas)
Dos placas paralelas separadas  d = 2,00 cm  con una diferencia de potencial  V = 100 V
(modelo de sensores capacitivos, pantallas, memorias).

**Relación campo–potencial (campo uniforme):**   E = V / d

**Trabajo sobre una carga:**   W = q · V

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))
# --- fin setup ---


def capacitor(V=100.0, d_cm=2.0, q_nC=5.0):
    d = d_cm/100
    E = V/d
    q = q_nC*1e-9
    W = q*V
    # Perfil de potencial lineal entre placas
    x = np.linspace(0, d, 100)
    Vx = V*(1 - x/d)  # de V en x=0 a 0 en x=d
    plt.figure(figsize=(9,5))
    plt.plot(x*100, Vx, lw=2, color='teal')
    plt.axvline(0, color='red', ls='--', label='placa + (V)')
    plt.axvline(d_cm, color='blue', ls='--', label='placa − (0 V)')
    plt.xlabel('posición entre placas (cm)'); plt.ylabel('Potencial (V)')
    plt.title('Potencial lineal dentro del capacitor')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()
    print(f"Campo eléctrico uniforme:  E = V/d = {E:.1f} V/m ≈ {E:.3g} V/m")
    print(f"Trabajo para mover q={q_nC:.1f} nC de una placa a otra:  W = q·V = {W:.3e} J")

interact(capacitor,
    V=FloatSlider(value=100.0, min=10, max=500, step=10, description='V (voltios)'),
    d_cm=FloatSlider(value=2.0, min=0.5, max=10, step=0.5, description='d (cm)'),
    q_nC=FloatSlider(value=5.0, min=1, max=50, step=1, description='q (nC)'));

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("a) Campo eléctrico E entre placas (V=100 V, d=2 cm):", 5000, 0.03, "V/m",
    "E = V/d = 100/0,02 = 5,00×10³ V/m.")
quiz_numerico("a) Trabajo para mover q=5 nC entre las placas:", 5.0e-7, 0.05, "J",
    "W = q·V = 5e-9·100 = 5,0×10⁻⁷ J.")
quiz_opcion_multiple("b) Si acercas las placas (d menor) manteniendo V fija, el campo E:",
    ["Disminuye", "Aumenta (E = V/d, inversamente proporcional a d)", "No cambia", "Se anula"],
    1, "Con V fija, al reducir d el campo E=V/d aumenta.")
```
