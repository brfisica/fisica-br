## Punto 1 — Terminal esférico de alta tensión (Ing. Eléctrica/Electrónica)
Una terminal esférica conductora de radio  R = 10,0 cm  acumula carga  Q = +50,0 nC.

**Potencial (carga puntual/esfera):**   V(r) = k·Q / r      (para r ≥ R)

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    """Botón que revela la solución solo al pulsarlo."""
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))

k = 8.99e9  # constante de Coulomb, N m^2 / C^2
# --- fin setup ---


def esquema():
    fig, ax = plt.subplots(figsize=(4,4))
    ax.add_patch(plt.Circle((0,0), 1.0, color='goldenrod', alpha=0.6))
    for a in np.linspace(0,2*np.pi,12,endpoint=False):
        ax.text(1.08*np.cos(a),1.08*np.sin(a),'+',color='red',ha='center',va='center',fontsize=9)
    ax.annotate('', xy=(1,0), xytext=(0,0), arrowprops=dict(arrowstyle='->',lw=1.5))
    ax.text(0.45,0.12,'R',fontsize=13); ax.plot(0,0,'k+',ms=10)
    ax.set_xlim(-1.6,1.6); ax.set_ylim(-1.6,1.6); ax.set_aspect('equal'); ax.axis('off')
    ax.set_title('Terminal esférica conductora (carga Q en la superficie)')
    plt.tight_layout(); plt.show()
esquema()

def potencial_esfera(Q_nC=50.0, R_cm=10.0, r_cm=30.0):
    Q=Q_nC*1e-9; R=R_cm/100; rp=r_cm/100
    r=np.linspace(R,1.0,500); V=k*Q/r
    plt.figure(figsize=(9,4.5)); plt.plot(r*100,V,lw=2,color='darkred')
    plt.axvline(R_cm,color='gray',ls='--',label=f'superficie R={R_cm:.0f} cm')
    plt.scatter([max(r_cm,R_cm)],[k*Q/max(rp,R)],color='blue',s=60,zorder=5,label=f'r={r_cm:.0f} cm')
    plt.xlabel('r (cm)'); plt.ylabel('V(r) (V)'); plt.title('Potencial alrededor de la terminal')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(potencial_esfera,
    Q_nC=FloatSlider(value=50.0,min=-100,max=100,step=5,description='Q (nC)'),
    R_cm=FloatSlider(value=10.0,min=5,max=30,step=1,description='R (cm)'),
    r_cm=FloatSlider(value=30.0,min=5,max=90,step=5,description='r (cm)'));

def solucion():
    Q=50e-9; R=0.10; rp=0.30
    print(f"V superficie (r=R=10 cm) = kQ/R = {k*Q/R:.1f} V ≈ {k*Q/R:.3g} V")
    print(f"V en r=30 cm = kQ/r = {k*Q/rp:.1f} V")
    print(f"Trabajo para traer q=+2 nC desde ∞ a r=30 cm: W=qV = {2e-9*k*Q/rp:.3e} J")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Potencial en la superficie (r=R=10 cm, Q=50 nC):",4495.9,0.03,"V",
    "V=kQ/R≈4,50×10³ V.")
quiz_numerico("Potencial en r=30 cm:",1498.6,0.03,"V","V=kQ/r≈1,50×10³ V.")
quiz_opcion_multiple("Si duplicas la distancia r, el potencial V:",
    ["Se duplica","Se reduce a la mitad (V ∝ 1/r)","No cambia","Se reduce a un cuarto"],1,
    "V va como 1/r: al doblar r, V cae a la mitad.")
```

## Punto 2 — Superposición de dos cargas (Ing. Electrónica/Biomédica)
Q₁ = +30,0 nC en x = −10 cm  y  Q₂ = −30,0 nC en x = +10 cm (dipolo/electrodos).

**Potencial total (superposición):**   V = k·Q₁/r₁ + k·Q₂/r₂

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    """Botón que revela la solución solo al pulsarlo."""
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))

k = 8.99e9  # constante de Coulomb, N m^2 / C^2
# --- fin setup ---


def esquema():
    fig, ax = plt.subplots(figsize=(5,2.5))
    ax.scatter([-1],[0],s=300,color='red'); ax.text(-1,0,'+',color='white',ha='center',va='center',fontsize=16,fontweight='bold')
    ax.scatter([1],[0],s=300,color='blue'); ax.text(1,0,'−',color='white',ha='center',va='center',fontsize=16,fontweight='bold')
    ax.text(-1,-0.4,'Q₁ (x=−10cm)',ha='center'); ax.text(1,-0.4,'Q₂ (x=+10cm)',ha='center')
    ax.annotate('',xy=(1,0.3),xytext=(-1,0.3),arrowprops=dict(arrowstyle='<->'))
    ax.text(0,0.4,'20 cm',ha='center')
    ax.set_xlim(-2,2); ax.set_ylim(-0.8,0.8); ax.axis('off'); ax.set_title('Par de cargas (dipolo)')
    plt.tight_layout(); plt.show()
esquema()

def dos_cargas(Q1_nC=30.0, Q2_nC=-30.0, x_cm=5.0, y_cm=0.0):
    Q1=Q1_nC*1e-9; Q2=Q2_nC*1e-9; x1,x2=-0.10,0.10
    xs=np.linspace(-0.3,0.3,200); ys=np.linspace(-0.2,0.2,150); X,Y=np.meshgrid(xs,ys)
    R1=np.hypot(X-x1,Y); R2=np.hypot(X-x2,Y); R1[R1<1e-3]=1e-3; R2[R2<1e-3]=1e-3
    Vmap=k*Q1/R1+k*Q2/R2
    plt.figure(figsize=(9,4.5))
    cf=plt.contourf(X*100,Y*100,np.clip(Vmap,-4000,4000),levels=30,cmap='RdBu_r')
    plt.colorbar(cf,label='V (V)')
    plt.scatter([x1*100],[0],color='darkred',s=80,marker='o'); plt.scatter([x2*100],[0],color='darkblue',s=80,marker='o')
    plt.scatter([x_cm],[y_cm],color='black',s=50,zorder=5,label='punto')
    plt.xlabel('x (cm)'); plt.ylabel('y (cm)'); plt.title('Mapa de potencial'); plt.legend(); plt.tight_layout(); plt.show()

interact(dos_cargas,
    Q1_nC=FloatSlider(value=30.0,min=-50,max=50,step=5,description='Q₁ (nC)'),
    Q2_nC=FloatSlider(value=-30.0,min=-50,max=50,step=5,description='Q₂ (nC)'),
    x_cm=FloatSlider(value=5.0,min=-25,max=25,step=1,description='x (cm)'),
    y_cm=FloatSlider(value=0.0,min=-15,max=15,step=1,description='y (cm)'));

def solucion():
    Q1=30e-9; Q2=-30e-9; px,py=0.05,0.0
    r1=np.hypot(px+0.10,py); r2=np.hypot(px-0.10,py)
    print(f"En (5,0) cm: r₁={r1*100:.0f} cm, r₂={r2*100:.0f} cm")
    print(f"V = kQ₁/r₁ + kQ₂/r₂ = {k*Q1/r1+k*Q2/r2:.3g} V")
    print("En la mediatriz (r₁=r₂) el potencial es 0 por cancelación.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_opcion_multiple("¿Cuánto vale V en la mediatriz (equidistante de ambas cargas)?",
    ["Máximo","Cero: se cancelan +Q y −Q","Infinito","Igual a kQ/r"],1,
    "r₁=r₂ y Q₂=−Q₁ ⟹ V=0.")
quiz_numerico("V total en el punto (x=5 cm, y=0):",-3600,0.08,"V",
    "r₁=15 cm, r₂=5 cm ⟹ V≈−3,6×10³ V.")
quiz_opcion_multiple("El potencial es una magnitud:",
    ["Vectorial","Escalar (se suma con signo)","Siempre positiva","Sin unidades"],1,
    "V es escalar: se suman aportes con su signo.")
```

## Punto 3 — Capacitor de placas paralelas (Ing. Mecánica/Sistemas)
Placas separadas  d = 2,00 cm  con diferencia de potencial  V = 100 V.

**Campo (uniforme):**  E = V / d       **Trabajo:**  W = q · V

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    """Botón que revela la solución solo al pulsarlo."""
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))

k = 8.99e9  # constante de Coulomb, N m^2 / C^2
# --- fin setup ---


def esquema():
    fig, ax = plt.subplots(figsize=(5,3))
    ax.plot([0,0],[-1,1],lw=4,color='red'); ax.plot([2,2],[-1,1],lw=4,color='blue')
    ax.text(-0.1,1.15,'placa + (V)',color='red',ha='center'); ax.text(2.1,1.15,'placa − (0V)',color='blue',ha='center')
    for yy in np.linspace(-0.8,0.8,5):
        ax.annotate('',xy=(1.8,yy),xytext=(0.2,yy),arrowprops=dict(arrowstyle='->',color='gray'))
    ax.text(1,-1.25,'E = V/d',ha='center',fontsize=12)
    ax.annotate('',xy=(2,-1.5),xytext=(0,-1.5),arrowprops=dict(arrowstyle='<->')); ax.text(1,-1.7,'d',ha='center')
    ax.set_xlim(-0.5,2.5); ax.set_ylim(-2,1.5); ax.axis('off'); ax.set_title('Capacitor de placas paralelas')
    plt.tight_layout(); plt.show()
esquema()

def capacitor(V=100.0, d_cm=2.0, q_nC=5.0):
    d=d_cm/100; x=np.linspace(0,d,100); Vx=V*(1-x/d)
    plt.figure(figsize=(9,4.5)); plt.plot(x*100,Vx,lw=2,color='teal')
    plt.axvline(0,color='red',ls='--',label='placa + (V)'); plt.axvline(d_cm,color='blue',ls='--',label='placa − (0V)')
    plt.xlabel('posición (cm)'); plt.ylabel('Potencial (V)'); plt.title('Potencial lineal dentro del capacitor')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(capacitor,
    V=FloatSlider(value=100.0,min=10,max=500,step=10,description='V (voltios)'),
    d_cm=FloatSlider(value=2.0,min=0.5,max=10,step=0.5,description='d (cm)'),
    q_nC=FloatSlider(value=5.0,min=1,max=50,step=1,description='q (nC)'));

def solucion():
    V=100.0; d=0.02; q=5e-9
    print(f"E = V/d = {V/d:.1f} V/m ≈ {V/d:.3g} V/m")
    print(f"W = q·V = {q*V:.3e} J")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("Campo E entre placas (V=100 V, d=2 cm):",5000,0.03,"V/m","E=V/d=5,00×10³ V/m.")
quiz_numerico("Trabajo para mover q=5 nC entre placas:",5.0e-7,0.05,"J","W=qV=5,0×10⁻⁷ J.")
quiz_opcion_multiple("Si acercas las placas (d menor) con V fija, el campo E:",
    ["Disminuye","Aumenta (E=V/d)","No cambia","Se anula"],1,"Con V fija, menor d ⟹ mayor E.")
```

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    """Botón que revela la solución solo al pulsarlo."""
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info')
    salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try:
                v = float(caja.value.replace(",", "."))
            except ValueError:
                print("\u26a0 Escribe un n\u00famero (punto o coma decimal)."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else:
                print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar)
    display(VBox([titulo, caja, boton, salida]))

k = 8.99e9  # constante de Coulomb, N m^2 / C^2
# --- fin setup ---


def esquema():
    fig = plt.figure(figsize=(5,3)); ax=fig.add_subplot(111,projection='3d')
    th=np.linspace(0,2*np.pi,60); Rr=1.0
    ax.plot(Rr*np.cos(th),Rr*np.sin(th),0,lw=3,color='purple')
    ax.plot([0,0],[0,0],[0,1.5],'k--')
    ax.scatter([0],[0],[1.2],color='black',s=40)
    ax.text(0.1,0,1.2,'punto (eje z)'); ax.text(1.0,0,0.15,'anillo R, Q')
    ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_title('Anillo cargado y punto sobre el eje')
    plt.tight_layout(); plt.show()
esquema()

def anillo(Q_nC=20.0, R_cm=5.0, z_cm=10.0):
    Q=Q_nC*1e-9; R=R_cm/100
    z=np.linspace(-0.3,0.3,400); V=k*Q/np.sqrt(R**2+z**2)
    zp=z_cm/100; Vp=k*Q/np.sqrt(R**2+zp**2)
    plt.figure(figsize=(9,4.5)); plt.plot(z*100,V,lw=2,color='purple')
    plt.scatter([z_cm],[Vp],color='red',s=60,zorder=5,label=f'z={z_cm:.0f} cm')
    plt.axvline(0,color='gray',ls=':',label='centro del anillo')
    plt.xlabel('z (cm)'); plt.ylabel('V(z) (V)'); plt.title('Potencial en el eje del anillo')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(anillo,
    Q_nC=FloatSlider(value=20.0,min=-50,max=50,step=5,description='Q (nC)'),
    R_cm=FloatSlider(value=5.0,min=2,max=15,step=1,description='R (cm)'),
    z_cm=FloatSlider(value=10.0,min=-25,max=25,step=1,description='z (cm)'));

def solucion():
    Q=20e-9; R=0.05
    Vc=k*Q/R; Vz=k*Q/np.sqrt(R**2+0.10**2)
    print(f"V en el centro (z=0) = kQ/R = {Vc:.3g} V")
    print(f"V en z=10 cm = kQ/√(R²+z²) = {Vz:.3g} V")
    print("El potencial es máximo en el centro y decae al alejarse por el eje.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Potencial en el centro del anillo (z=0, R=5 cm, Q=20 nC):",3596,0.04,"V",
    "V=kQ/R=8,99e9·20e-9/0,05≈3,60×10³ V.")
quiz_numerico("Potencial en z=10 cm sobre el eje:",1608,0.05,"V",
    "V=kQ/√(R²+z²)=kQ/√(0,05²+0,10²)≈1,61×10³ V.")
quiz_opcion_multiple("¿Dónde es máximo el potencial sobre el eje?",
    ["Muy lejos","En el centro del anillo (z=0)","En el infinito","Nunca es máximo"],1,
    "En z=0 el denominador √(R²+z²) es mínimo, así que V es máximo.")
```
