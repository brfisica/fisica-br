*Física interactiva para ingeniería.* Cuando ocurre un sismo, la ruptura en una falla libera energía
en forma de **ondas mecánicas** que viajan por la corteza terrestre. En una primera aproximación, la
onda primaria (**onda P**) se propaga con velocidad promedio casi constante: un **Movimiento
Rectilíneo Uniforme (MRU)**. Los sistemas de **alerta sísmica temprana** aprovechan que la señal de
radio (a la velocidad de la luz) le gana la carrera a la onda destructiva. Este taller usa ese
fenómeno —muy presente en Colombia— para dominar el MRU.

**Convenciones:** tres cifras significativas · coma decimal · distancias en km, tiempos en s.

**Ecuaciones del modelo (MRU):**
- Posición:  x(t) = x₀ + v·(t − t₀)
- Velocidad constante:  v = Δx/Δt    ⟹    t = x/v
- Velocidades típicas:  onda P ≈ 6,0 km/s · onda S ≈ 3,5 km/s · señal de radio ≈ 300 000 km/s

## Punto 1 — La carrera contra el sismo: alerta temprana (Ing. Civil/Telecom)  ★★  [CL, UCM]
Un epicentro sísmico está a  x = 180 km  de una ciudad. En el instante del sismo (t₀ = 0), una
estación junto al epicentro emite una **señal de radio** de alerta hacia la ciudad. La onda P viaja a
vₚ = 6,0 km/s  y la señal de radio a  v_radio = 300 000 km/s. Se busca el tiempo de cada una y el
**tiempo de alerta** con que cuenta la ciudad antes de sentir el sismo.

**Tiempos (MRU):**  t = x/v    **Alerta:**  t_alerta = tₚ − t_radio

> 🔧 **Aplicación en ingeniería:** este es el principio de los sistemas de alerta sísmica temprana. La
> señal electromagnética viaja millones de veces más rápido que la onda mecánica, así que llega casi
> instantáneamente y "avisa" segundos antes de que llegue el temblor. Esos segundos permiten detener
> trenes, cerrar válvulas de gas y ponerse a cubierto.

## Punto 2 — Onda P contra onda S (Ing. Geotécnica/Sísmica)  ★★  [CL, UFT]
En un sismo se generan dos ondas internas: la **onda P** (primaria, rápida, vₚ = 6,0 km/s) y la
**onda S** (secundaria, lenta pero más destructiva, v_s = 3,5 km/s). Para un epicentro a  x = 180 km,
se busca cuándo llega cada una y la diferencia de tiempo entre ambas.

**Tiempos:**  tₚ = x/vₚ  ,  t_s = x/v_s    **Diferencia:**  Δt = t_s − tₚ

> 🔧 **Aplicación en ingeniería:** la onda P llega primero pero es suave; la onda S llega después y
> causa el sacudón fuerte. Esa diferencia de tiempo es la que aprovecha la alerta: al detectar la
> onda P, el sistema sabe que la S (destructiva) viene en camino y activa el aviso.

## Punto 3 — ¿A qué distancia está el epicentro? (Ing. Sísmica/Instrumentación)  ★★★  [PRC, UCM]
Los sismógrafos miden la diferencia de tiempo Δt entre la llegada de la onda P y la S, y con ella
calculan la distancia al epicentro. Si una estación mide  Δt = 8,0 s  entre P y S (con vₚ = 6,0 km/s
y v_s = 3,5 km/s), se busca la distancia al epicentro.

**Distancia por diferencia P-S:**  x = Δt / (1/v_s − 1/vₚ)

> 🔧 **Aplicación en ingeniería:** así es como se localiza un sismo en la vida real. Una sola estación
> da la distancia (con este cálculo); combinando tres estaciones se triangula el epicentro exacto. Es

## Punto 5 — Velocidad promedio en capas distintas (Ing. Geofísica)  ★★★  [PRC, ULC]
La onda P no viaja a velocidad única: cambia según la roca. Suponga que recorre un primer tramo de
d₁ = 100 km  a  v₁ = 6,0 km/s  y un segundo tramo de  d₂ = 80 km  a  v₂ = 8,0 km/s. Se busca el
tiempo total y la **velocidad promedio** del recorrido.

**Tiempo total:**  t = d₁/v₁ + d₂/v₂    **Velocidad promedio:**  v_prom = (d₁+d₂)/t

> 🔧 **Aplicación en ingeniería:** la velocidad promedio no es el promedio simple de las velocidades:
> hay que sumar los tiempos de cada tramo. Este cálculo es esencial en geofísica para modelar cómo
> viajan las ondas por capas de distinta densidad, y así interpretar correctamente los sismogramas.
