## Punto 1 — Fuerza magnética sobre una carga (Ing. Electrónica/Física)  ★★★  [PRC, UCM]
Un electrón se mueve a  v = 2,00×10⁶ m/s  perpendicular a un campo magnético  B = 0,500 T. Se busca
la fuerza magnética sobre él.

**Fuerza magnética:**  F = q·v·B    (con v perpendicular a B)

> 🔧 **Aplicación en ingeniería:** esta fuerza desvía cargas en movimiento y es la base de los tubos
> de rayos, espectrómetros de masas, y el confinamiento de plasmas. La fuerza es siempre
> perpendicular a la velocidad, por lo que curva la trayectoria sin cambiar la rapidez.

## Punto 2 — De la carga al conductor: catenaria de un tranvía en Bogotá (Ing. Eléctrica/Transporte)  ★★  [PRC, UCM]
Cada portador de carga que viaja por un conductor siente q·v_d × B. Sumando sobre todos los portadores
de un tramo se obtiene la fuerza sobre el conductor:  **F = I·L × B**, con  I = n·q·v_d·A.

Un tranvía de corriente continua toma su energía de un hilo de contacto de cobre de **107 mm²**
(0,959 kg/m, calculado con ρ = 8,96 g/cm³). En un vano de **40,0 m** circula **I = 1,20 kA**. El campo
terrestre en Bogotá (modelos IGRF-14 y WMM2025, octubre de 2026) tiene una componente horizontal
**H = 26,8 µT** hacia el norte magnético y una vertical **Z = 13,2 µT** hacia abajo (inclinación 26,3°;
la declinación, 8,79° al oeste, no interviene si se mide todo respecto al norte magnético).

Se busca la fuerza sobre el vano si la línea va de sur a norte y si va de oeste a este, y compararla
con su peso.

**Fuerza sobre el conductor:**  F = I·L × B    **Peso del vano:**  W = (m/L)·L·g

> 🔧 **Aplicación en ingeniería:** los tranvías y trenes ligeros de corriente continua se alimentan
> por catenaria. Saber qué fuerzas magnéticas son despreciables (la del campo terrestre) y cuáles no
> (las fuerzas entre conductores durante un cortocircuito de decenas de kA) es parte del diseño
> mecánico de la línea.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---

from ipywidgets import IntSlider, Checkbox

def _fmt(x, cs=3):
    # 3 cifras significativas, notación ×10ⁿ si hace falta, coma decimal
    if x == 0: return "0"
    _sup = '⁰¹²³⁴⁵⁶⁷⁸⁹'
    exp = int(np.floor(np.log10(abs(x))))
    if -2 <= exp <= 3:
        dec = max(cs-1-exp, 0)
        t = f"{x:.{dec}f}"
    else:
        mant = x/10**exp
        if abs(round(mant, cs-1)) >= 10: mant, exp = mant/10, exp+1
        e = ('⁻' if exp < 0 else '') + ''.join(_sup[int(d)] for d in str(abs(exp)))
        t = f"{mant:.{cs-1}f}×10{e}"
    return _num_coma(t)

g = 9.81
H_T = 26.8e-6     # T, componente horizontal del campo terrestre (hacia el norte magnético)
Z_T = 13.2e-6     # T, componente vertical (hacia abajo)
I_c = 1200.0      # A, corriente continua en el hilo de contacto
L_v = 40.0        # m, vano
m_lin = 8960*107e-6   # kg/m, cobre de 107 mm² (ρ = 8,96 g/cm³)

def _F_vec(az_deg, I=I_c, L=L_v):
    # componentes (Este, Norte, Arriba) de F = I·L × B, con la corriente apuntando al azimut dado
    az = np.radians(az_deg)
    Lhat = np.array([np.sin(az), np.cos(az), 0.0])
    B = np.array([0.0, H_T, -Z_T])
    return I*L*np.cross(Lhat, B)

def esquema():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.2))
    # Vista en planta
    ax1.annotate('', xy=(0, 1.6), xytext=(0, 0), arrowprops=dict(arrowstyle='-|>', color='#2980b9', lw=2.5))
    ax1.text(0.08, 1.45, 'B horizontal\n(norte magnético)', color='#2980b9', fontsize=8.5, fontweight='bold')
    ax1.plot([-1.4, 1.4], [-0.9, -0.9], color='#c0392b', lw=3)
    ax1.annotate('', xy=(1.55, -0.9), xytext=(1.0, -0.9), arrowprops=dict(arrowstyle='-|>', color='#c0392b', lw=2.5))
    ax1.text(0, -1.25, 'línea oeste → este', ha='center', color='#c0392b', fontsize=9)
    ax1.plot([-1.2, -1.2], [-0.4, 1.3], color='#8e44ad', lw=3)
    ax1.annotate('', xy=(-1.2, 1.5), xytext=(-1.2, 1.0), arrowprops=dict(arrowstyle='-|>', color='#8e44ad', lw=2.5))
    ax1.text(-1.15, -0.6, 'línea sur → norte', color='#8e44ad', fontsize=9)
    ax1.annotate('', xy=(1.6, 1.6), xytext=(1.6, 1.0), arrowprops=dict(arrowstyle='-|>', color='black', lw=1.2))
    ax1.annotate('', xy=(2.2, 1.0), xytext=(1.6, 1.0), arrowprops=dict(arrowstyle='-|>', color='black', lw=1.2))
    ax1.text(1.55, 1.68, 'N', fontsize=10, fontweight='bold'); ax1.text(2.25, 0.95, 'E', fontsize=10, fontweight='bold')
    ax1.set_xlim(-1.8, 2.5); ax1.set_ylim(-1.5, 1.9); ax1.set_aspect('equal'); ax1.axis('off')
    ax1.set_title('Vista en planta')
    # Vista lateral (norte a la derecha)
    inc = np.radians(26.3)
    ax2.axhline(0, color='#7f8c8d', lw=1)
    ax2.annotate('', xy=(1.6*np.cos(inc), -1.6*np.sin(inc)), xytext=(0, 0),
                 arrowprops=dict(arrowstyle='-|>', color='#2980b9', lw=2.5))
    ax2.annotate('', xy=(1.6*np.cos(inc), 0), xytext=(0, 0), arrowprops=dict(arrowstyle='-|>', color='#2980b9', lw=1.2, ls='--'))
    ax2.annotate('', xy=(0, -1.6*np.sin(inc)), xytext=(0, 0), arrowprops=dict(arrowstyle='-|>', color='#2980b9', lw=1.2, ls='--'))
    ax2.text(1.5, 0.1, 'H = 26,8 µT', color='#2980b9', fontsize=9)
    ax2.text(0.05, -0.85, 'Z = 13,2 µT\n(hacia abajo)', color='#2980b9', fontsize=9)
    ax2.text(0.55, -0.12, 'I = 26,3°', fontsize=9)
    ax2.text(1.75, -0.95, 'B', color='#2980b9', fontsize=12, fontweight='bold', style='italic')
    ax2.text(1.9, 0.25, 'norte →', fontsize=9); ax2.text(-0.25, 0.55, '↑ arriba', fontsize=9)
    ax2.set_xlim(-0.4, 2.4); ax2.set_ylim(-1.2, 0.8); ax2.set_aspect('equal'); ax2.axis('off')
    ax2.set_title('Campo terrestre en Bogotá (vista lateral)')
    plt.suptitle('Catenaria de corriente continua dentro del campo geomagnético', fontweight='bold')
    plt.tight_layout(); plt.show()
esquema()

def explorar_catenaria(azimut=90.0, I_kA=1.20):
    F = _F_vec(azimut, I=I_kA*1000)
    F = np.where(np.abs(F) < 1e-9, 0.0, F)
    peso = m_lin*L_v*g
    fig, ax = plt.subplots(figsize=(9, 3.8))
    nombres = ['F hacia el este', 'F hacia el norte', 'F hacia arriba']
    cols = ['#c0392b', '#8e44ad', '#27ae60']
    ax.barh(nombres, F, color=cols, alpha=0.75)
    for k, v in enumerate(F):
        ax.text(v, k, '  ' + _fmt(v) + ' N', va='center', fontsize=10, fontweight='bold')
    ax.axvline(0, color='#2c3e50', lw=1)
    lim = 2.6*I_kA/1.2
    ax.set_xlim(-lim, lim)
    ax.set_xlabel('fuerza sobre el vano de 40,0 m (N)')
    ax.set_title(f'Azimut {azimut:.0f}° · |F| = {_fmt(np.linalg.norm(F))} N · peso del vano = {_fmt(peso)} N')
    plt.tight_layout(); plt.show()

interact(explorar_catenaria,
    azimut=FloatSlider(value=90.0, min=0.0, max=180.0, step=5.0, description='azimut (°)'),
    I_kA=FloatSlider(value=1.20, min=0.20, max=2.00, step=0.10, readout_format='.2f', description='I (kA)'));

def solucion():
    Fns = _F_vec(0.0); Feo = _F_vec(90.0)
    peso = m_lin*L_v*g
    print_es("DE LA CARGA AL CONDUCTOR:")
    print_es("   Cada portador siente q·v_d × B. En un tramo de longitud L y sección A hay n·A·L")
    print_es("   portadores: F = (n·A·L)·q·v_d × B = (n·q·v_d·A)·L × B = I·L × B.")
    print_es("")
    print_es("CAMPO EN BOGOTÁ (marco del norte magnético; Este, Norte, Arriba):  B = (0; 26,8; −13,2) µT")
    print_es("")
    print_es("a) Línea sur → norte (L hacia el norte):")
    print_es("   L × B = ĵ × (H ĵ − Z k̂) = −Z î  →  solo actúa la componente vertical.")
    print_es(f"   F = I·L·Z = 1200·40,0·13,2×10⁻⁶ = {_fmt(abs(Fns[0]))} N, hacia el OESTE (horizontal).")
    print_es("b) Línea oeste → este (L hacia el este):")
    print_es("   L × B = î × (H ĵ − Z k̂) = H k̂ + Z ĵ")
    print_es(f"   Hacia arriba: I·L·H = {_fmt(Feo[2])} N · hacia el norte: I·L·Z = {_fmt(Feo[1])} N · total {_fmt(np.linalg.norm(Feo))} N")
    print_es("")
    print_es(f"c) Peso del vano: m·g = 0,959 kg/m · 40,0 m · 9,81 m/s² = {_fmt(peso)} N")
    print_es(f"   La mayor componente vertical ({_fmt(Feo[2])} N) es el {_fmt(Feo[2]/peso*100)} % del peso: despreciable.")
    print_es("   Verificación de unidades: A·m·T = A·m·N/(A·m) = N.")
    print_es("")
    print_es("d) En corriente alterna de 60 Hz, la fuerza (con el campo terrestre, que es estático) cambia")
    print_es("   de sentido 120 veces por segundo y su promedio es cero: produce una vibración minúscula,")
    print_es("   no una carga sostenida. Lo que sí importa en diseño son las fuerzas entre conductores")
    print_es("   durante un cortocircuito, cuando las corrientes llegan a decenas de kA.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_numerico("Para el vano de 40,0 m orientado de oeste a este (corriente hacia el este, I = 1,20 kA), encuentre la componente VERTICAL de la fuerza magnética (en N).",
    1.29, 0.03, "N", "F_vertical = I·L·H = 1200·40,0·26,8×10⁻⁶ = 1,29 N, hacia arriba.", id_preg="FTM_p2_1", peso=2.0)
quiz_numerico("Para el mismo vano orientado de sur a norte (corriente hacia el norte), encuentre la magnitud de la fuerza magnética (en N).",
    0.634, 0.03, "N", "Solo actúa la componente vertical del campo: F = I·L·Z = 1200·40,0·13,2×10⁻⁶ = 0,634 N.", id_preg="FTM_p2_2", peso=2.0)
quiz_opcion_multiple("En el vano sur → norte (corriente hacia el norte), ¿hacia dónde apunta la fuerza magnética?",
    ["Hacia el este, porque F sigue la regla de la mano izquierda", "Hacia el oeste: ĵ × (−k̂) = −î", "Hacia arriba, igual que en el vano oeste → este", "Hacia abajo, en la dirección de la componente Z del campo"], 1,
    "L = ĵ (norte) y la componente útil del campo es −Z k̂ (hacia abajo): ĵ × (−k̂) = −î, es decir, hacia el oeste. La componente H es paralela a L y no aporta.",
    id_preg="FTM_p2_3", peso=2.0)
quiz_opcion_multiple("La velocidad de deriva en este hilo es del orden de 8×10⁻⁴ m/s. ¿Cómo puede resultar una fuerza medible sobre el conductor?",
    ["Porque la fuerza no depende de v_d sino solo de la corriente", "Porque hay ~10²⁹ portadores/m³: fuerzas diminutas sobre cada uno que se suman", "Porque los electrones viajan a la velocidad de la luz dentro del cable", "Porque el campo terrestre es mucho más intenso dentro del cobre"], 1,
    "Cada portador siente una fuerza ínfima, pero el cobre tiene unos 8,5×10²⁸ electrones libres por m³: I = n·q·v_d·A agrupa ese número enorme. La fuerza sí depende de v_d, a través de I.",
    id_preg="FTM_p2_4", peso=2.0)


quiz_opcion_multiple('DEFIENDE TU CONCLUSIÓN: un estudiante afirma que la fuerza del campo terrestre sobre la catenaria es despreciable para el diseño mecánico, aunque la corriente sea de 1,20 kA. ¿Correcto o solo convincente?',
    ['Correcto: la mayor componente vertical es 1,29 N, el 0,342 % de los 376 N de peso',
     'Correcto, porque el campo terrestre no ejerce fuerza sobre corrientes continuas',
     'Solo convincente: con 1,20 kA la fuerza es comparable al peso y hay que tensar más el hilo',
     'Solo convincente: la fuerza crece con el cuadrado de la corriente y domina en las horas pico'],
    0,
    'Esta vez el estudiante TIENE razón: 1,29 N frente a 376 N. La opción «no ejerce fuerza sobre corrientes continuas» acierta el veredicto con una razón falsa: la fuerza existe (la calculaste), solo es pequeña. Además F es lineal en I, no cuadrática.',
    id_preg='FTM_p2_concep_2', peso=2.0)
```

## Punto 3 — Balanza de corriente (Metrología/Máquinas eléctricas)  ★★  [PRC, UFT]
Un imán en U descansa sobre una balanza digital tarada en cero. Un conductor rígido, sostenido por un
soporte que **no** toca la balanza, pasa horizontalmente entre los polos: el tramo dentro del campo
mide  L = 5,00 cm  y es perpendicular a B. Se toma  I > 0  cuando la corriente **sale** de la página.
La tabla de datos aparece debajo del esquema.

**Fuerza sobre el conductor:**  F = I·L × B  →  F = B·I·L    **Lo que mide la balanza:**  F = Δm·g

**Responde y argumenta:**
- **(a)** La fuerza actúa sobre el *conductor*. ¿Por qué cambia la lectura de la balanza, que sostiene al *imán*?
- **(b)** Determina el sentido de la fuerza cuando I > 0 y verifica que es coherente con el signo de Δm.
- **(c)** Grafica F = Δm·g vs I, ajusta una recta y obtén B de la pendiente. ¿Qué significaría un intercepto distinto de cero?
- **(d)** Diseño: la balanza tiene resolución de 0,01 g y la fuente da como máximo 2,00 A. ¿Qué corriente hace cambiar la lectura 0,800 g? ¿Cuál es el campo más pequeño que el montaje detecta?

## Punto 4 — Torque sobre una bobina (Ing. Mecánica/Eléctrica)  ★★  [CL, PRC]
Una bobina rectangular de  **N = 25** vueltas, con lados  a = 4,00 cm  y  b = 6,00 cm, puede girar
alrededor de un eje paralelo a los lados b. Conduce  **I = 1,50 A**  en un campo uniforme horizontal
**B = 0,400 T**, perpendicular al eje. El momento magnético µ forma un ángulo  **φ = 40,0°**  con B.

**Momento magnético:**  µ = N·I·A    **Torque:**  τ = µ × B  (τ = µ·B·sen φ)    **Energía:**  U = −µ·B·cos φ

Se busca la fuerza sobre cada lado, el torque (por el par de fuerzas y por µ × B), las posiciones de
equilibrio y el trabajo del torque magnético al pasar de φ = 40,0° a φ = 0.

> 🔧 **Aplicación en ingeniería:** el torque sobre una espira es el principio de los motores, los
> instrumentos de aguja y los actuadores de bobina móvil. En plantas industriales y refinerías, los
> actuadores eléctricos de válvulas convierten justamente este torque en el giro que abre o cierra el paso.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---


# Calificación formativa del taller
calificacion_final(total_preguntas=29)
```
