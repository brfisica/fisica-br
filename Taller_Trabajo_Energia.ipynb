## Punto 3 — Conservación de energía en un descenso (Ing. Mecánica/Transporte)  ★★★  [CL, UCM, PRC]
Un vagón de  m = 500 kg  desciende desde una altura  h = 40,0 m  por una vía sin fricción
(aproximación ideal). Se busca su rapidez al llegar abajo usando conservación de energía.

**Conservación:**  Ep_arriba = Ec_abajo  →  m·g·h = ½·m·v²  →  v = √(2·g·h)

> 🔧 **Aplicación en ingeniería:** el diseño de sistemas de transporte por gravedad (funiculares,
