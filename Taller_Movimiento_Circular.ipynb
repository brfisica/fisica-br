## Punto 1 — Secadora rotatoria de café (Ing. Agroindustrial/Mecánica)  ★★  [CL, UCM]
Una secadora rotatoria de café pergamino (tipo guardiola) es un tambor horizontal de **1,40 m de diámetro**
que gira a **12,0 rpm** mientras entra aire caliente. El grano sube con la pared y, cerca de lo alto, se
despega y cae en cascada a través del aire: así se voltea y se seca de manera uniforme.

Se busca: (i) la aceleración centrípeta de un grano en la pared, como fracción de g; (ii) las rpm mínimas a
las que un grano quedaría pegado a la pared en el punto más alto; (iii) qué pasaría con el secado si se
superan esas rpm.

**Segunda Ley en la dirección radial:**  arriba: N + m·g = m·ω²·R    ·    abajo: N − m·g = m·ω²·R
El grano sigue pegado arriba solo si N ≥ 0, es decir, si ω²·R ≥ g.

> 🔧 **Aplicación en ingeniería:** en las secadoras de café de los departamentos cafeteros, la velocidad del
> tambor no se elige "al ojo": si gira muy lento, el grano se arrastra por el fondo; si supera el umbral
> ω²R = g, el grano se queda pegado a la pared y deja de voltearse. El mismo criterio vale para molinos de
> bolas en la minería y para hornos rotatorios de cemento.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))
# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---

def _f3(x):
    import math
    if x == 0: return "0"
    e = int(math.floor(math.log10(abs(x))))
    if e >= 3 or e <= -3:
        m = x/10**e
        if round(abs(m), 2) >= 10: m /= 10; e += 1
        sup = '⁰¹²³⁴⁵⁶⁷⁸⁹'
        ee = ('⁻' if e < 0 else '') + ''.join(sup[int(d)] for d in str(abs(e)))
        return f"{m:.2f}".replace('.', ',') + '×10' + ee
    d = max(0, 2 - e)
    s = f"{x:.{d}f}"
    # si el redondeo sube de orden (p. ej. 99,96 → 100,0) se recalcula
    if len(s.replace('-', '').replace('.', '').lstrip('0')) > 3 and d > 0:
        s = f"{x:.{d-1}f}"
    return s.replace('.', ',')

def esquema():
    fig, ax = plt.subplots(figsize=(6, 5.2))
    R = 1.0; ang = np.linspace(0, 2*np.pi, 200)
    ax.plot(R*np.cos(ang), R*np.sin(ang), color='#34495e', lw=3)
    ax.scatter([0], [0], s=60, color='#2c3e50', zorder=5)
    # sentido de giro
    a2 = np.linspace(np.radians(200), np.radians(320), 40)
    ax.plot(0.35*np.cos(a2), 0.35*np.sin(a2), color='#7f8c8d', lw=1.5)
    ax.annotate('', xy=(0.35*np.cos(a2[-1]), 0.35*np.sin(a2[-1])), xytext=(0.35*np.cos(a2[-3]), 0.35*np.sin(a2[-3])),
                arrowprops=dict(arrowstyle='-|>', color='#7f8c8d', lw=1.5))
    ax.text(0.0, -0.55, 'ω', ha='center', fontsize=12, color='#7f8c8d')
    # grano que sube con la pared (lado derecho) y cae en cascada
    for th in np.radians([-60, -30, 0, 30, 60]):
        ax.scatter([0.93*np.cos(th)], [0.93*np.sin(th)], s=35, color='#8e5a2b', zorder=6)
    xs = np.linspace(0.45, -0.2, 7); ys = 0.85 - 1.6*(0.45 - xs)**1.2
    ax.scatter(xs, ys, s=22, color='#8e5a2b', alpha=0.6, zorder=6)
    ax.text(-0.05, 0.25, 'cascada', fontsize=9, color='#8e5a2b', rotation=-60)
    # DCL en el punto más alto y en el más bajo
    for y0, lab in [(R, 'arriba'), (-R, 'abajo')]:
        ax.scatter([0], [y0], s=90, color='#e67e22', zorder=7, edgecolor='white')
        ax.annotate('', xy=(0, y0-0.35), xytext=(0, y0), arrowprops=dict(arrowstyle='-|>', color='#e74c3c', lw=2.2))
        ax.text(0.06, y0-0.32, 'mg', color='#e74c3c', fontsize=10, fontweight='bold')
        sgn = -1 if y0 > 0 else 1
        ax.annotate('', xy=(-0.22, y0+sgn*0.35), xytext=(-0.22, y0), arrowprops=dict(arrowstyle='-|>', color='#27ae60', lw=2.2))
        ax.text(-0.42, y0+sgn*0.3, 'N', color='#27ae60', fontsize=10, fontweight='bold')
        ax.text(-0.95, y0+0.08*(-sgn), lab, fontsize=9, color='#2c3e50')
    ax.annotate('', xy=(R*np.cos(np.radians(150)), R*np.sin(np.radians(150))), xytext=(0, 0),
                arrowprops=dict(arrowstyle='->', color='#95a5a6', lw=1))
    ax.text(-0.5, 0.38, 'R', color='#7f8c8d', fontsize=10)
    ax.text(0, -1.55, 'Ambas fuerzas van hacia el centro arriba; abajo la normal debe vencer el peso',
            ha='center', fontsize=9, color='#2c3e50', fontweight='bold')
    ax.set_xlim(-1.6, 1.6); ax.set_ylim(-1.7, 1.5); ax.set_aspect('equal'); ax.axis('off')
    ax.set_title('Secadora rotatoria: DCL del grano en la pared')
    plt.tight_layout(); plt.show()
esquema()

def guardiola(n=12.0, R=0.70):
    g = 9.81
    ns = np.linspace(0, 60, 300); ws = ns*2*np.pi/60
    arriba = ws**2*R/g - 1; abajo = ws**2*R/g + 1
    n_crit = np.sqrt(g/R)*60/(2*np.pi)
    plt.figure(figsize=(9, 4))
    plt.plot(ns, arriba, lw=2, color='#e67e22', label='N/(mg) en el punto más alto')
    plt.plot(ns, abajo, lw=2, color='#2980b9', label='N/(mg) en el punto más bajo')
    plt.axhline(0, color='#2c3e50', lw=1)
    plt.fill_between(ns, -1.2, 0, where=ns < n_crit, color='#f5cba7', alpha=0.35,
                     label='arriba N<0: el grano se despega (cascada)')
    plt.axvline(n_crit, color='#e74c3c', ls='--', lw=1.5, label='umbral ω²R = g')
    w = n*2*np.pi/60
    plt.scatter([n], [w**2*R/g - 1], color='#e67e22', s=60, zorder=5)
    plt.scatter([n], [w**2*R/g + 1], color='#2980b9', s=60, zorder=5)
    plt.ylim(-1.2, 4); plt.xlabel('velocidad de giro (rpm)'); plt.ylabel('N / (m·g)')
    plt.title('¿Cascada o grano pegado a la pared?'); plt.legend(fontsize=8, loc='upper left'); plt.tight_layout(); plt.show()
    print_es(f"a_c/g = {_f3(w**2*R/g)}   ·   umbral = {_f3(n_crit)} rpm   ·   " +
             ("el grano CAE en cascada (secado uniforme)" if n < n_crit else "el grano QUEDA PEGADO (no se voltea)"))

interact(guardiola,
    n=FloatSlider(value=12.0, min=1, max=60, step=1, description='giro (rpm)'),
    R=FloatSlider(value=0.70, min=0.30, max=1.20, step=0.05, description='R (m)'));

def solucion():
    g = 9.81; R = 0.700; w = 12.0*2*np.pi/60
    print_es(f"ω = 12,0·2π/60 = {_f3(w)} rad/s")
    print_es(f"(i) a_c = ω²·R = {_f3(w)}²·0,700 = {_f3(w**2*R)} m/s²  →  a_c/g = {_f3(w**2*R/g)}")
    wc = np.sqrt(g/R)
    print_es(f"(ii) Arriba, N = m(ω²R − g) ≥ 0  →  ω_mín = √(g/R) = √(9,81/0,700) = {_f3(wc)} rad/s = {_f3(wc*60/(2*np.pi))} rpm")
    print_es("     La masa se cancela: el umbral es el mismo para cualquier grano.")
    print_es(f"     Verificación (otro camino): con v² = gR, T = 2πR/v = {_f3(2*np.pi*R/np.sqrt(g*R))} s → 60/T = {_f3(60/(2*np.pi*R/np.sqrt(g*R)))} rpm")
    w40 = 40*2*np.pi/60
    print_es(f"(iii) A 40,0 rpm, a_c/g = {_f3(w40**2*R/g)} > 1: el grano queda pegado en todo el recorrido y deja de voltearse.")
    print_es("     A 12,0 rpm solo el 11 % de g 'sostiene' al grano: se despega antes de lo alto y cae en cascada.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("A 12,0 rpm, ¿qué fracción de g es la aceleración centrípeta de un grano en la pared del tambor (R = 0,700 m)? Escriba a_c/g.",
    0.1127, 0.05, "", "ω = 1,26 rad/s; a_c = ω²R = 1,11 m/s²; a_c/g = 0,113.", id_preg="Circ_p1_1", peso=2.0)
quiz_numerico("¿Cuáles son las rpm mínimas para que un grano quede pegado a la pared en el punto más alto?",
    35.75, 0.04, "rpm", "Arriba N = m(ω²R − g) ≥ 0 → ω = √(g/R) = 3,74 rad/s = 35,7 rpm.", id_preg="Circ_p1_2", peso=2.0)
quiz_opcion_multiple("Para secar más rápido, un operario propone subir la secadora de 12,0 a 40,0 rpm. ¿Qué le responde?",
    ['Empeora el secado: a 40,0 rpm ω²R vale 1,25 g, el grano queda pegado a la pared y deja de voltearse',
     'Mejora el secado: más rpm significa que el grano pasa más veces por el aire caliente en cada minuto',
     'No cambia nada: las rpm solo cambian la duración de cada vuelta, no la forma en que cae el grano',
     'Empeora el secado: a 40,0 rpm el grano pesa más y la pared del tambor no resiste la carga'],
    0, "40,0 rpm supera el umbral de 35,7 rpm. El café pegado a la pared no cae en cascada: se seca la capa expuesta y el resto no. El peso mg no cambia con las rpm.",
    id_preg="Circ_p1_3", peso=2.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: un estudiante dice que un lote de café húmedo, que pesa más, necesita más rpm para quedar pegado arriba que un lote seco. ¿Correcto o solo convincente?",
    ['Solo convincente: arriba se pega si ω²R ≥ g; la masa se cancela, así que el umbral es 35,7 rpm para ambos lotes',
     'Correcto: el lote húmedo tiene más peso, y para equilibrarlo hace falta una fuerza hacia afuera mayor, o sea, más rpm',
     'Solo convincente: el lote húmedo necesita menos rpm, porque su mayor masa le da más inercia para seguir la pared',
     'Correcto: la normal arriba es m(ω²R − g); con más masa ese término crece y por eso hay que subir ω'],
    0, "N = m(ω²R − g) se anula cuando ω²R = g, sin importar m. La opción «la normal arriba es m(ω²R − g)…» parte de la ecuación correcta pero concluye mal: m multiplica todo, no corre el umbral. La inercia no ayuda a seguir la curva: es lo que hace que el grano tienda a seguir en línea recta.",
    id_preg="Circ_p1_concep_1", peso=2.0)
```

## Punto 4 — Desbalance del ventilador de una torre de enfriamiento (Ing. Mecánica/Mantenimiento)  ★★★  [PRC, ULC]
En la torre de enfriamiento de una planta de proceso en Barrancabermeja, el ventilador gira a **180 rpm**. Una

cuántas veces es su peso; (ii) en temporada seca, el variador sube el ventilador a **220 rpm**: la nueva fuerza;
(iii) **sin limpiar el aspa**, ¿hasta cuántas rpm se puede operar respetando el límite de 400 N?

**Segunda Ley radial sobre la incrustación:**  F = m·ω²·r. Por la Tercera Ley, la incrustación tira del rotor
hacia afuera con la misma magnitud; esa fuerza gira con el rotor y llega a los rodamientos y a la estructura.

> 🔧 **Aplicación en ingeniería:** el análisis de vibraciones del mantenimiento predictivo detecta el
> desbalance como una vibración a la frecuencia de giro (aquí 3,00 Hz). Como la fuerza crece con ω², un
> desbalance tolerable a baja velocidad puede ser destructivo a alta velocidad: por eso se limpia y se balancea
> antes de subir las rpm, no después.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))
# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---

def _f3(x):
    import math
    if x == 0: return "0"
    e = int(math.floor(math.log10(abs(x))))
    if e >= 3 or e <= -3:
        m = x/10**e
        if round(abs(m), 2) >= 10: m /= 10; e += 1
        sup = '⁰¹²³⁴⁵⁶⁷⁸⁹'
        ee = ('⁻' if e < 0 else '') + ''.join(sup[int(d)] for d in str(abs(e)))
        return f"{m:.2f}".replace('.', ',') + '×10' + ee
    d = max(0, 2 - e)
    s = f"{x:.{d}f}"
    # si el redondeo sube de orden (p. ej. 99,96 → 100,0) se recalcula
    if len(s.replace('-', '').replace('.', '').lstrip('0')) > 3 and d > 0:
        s = f"{x:.{d-1}f}"
    return s.replace('.', ',')

def esquema():
    fig, ax = plt.subplots(figsize=(6, 5.2))
    ang = np.linspace(0, 2*np.pi, 200)
    ax.plot(1.3*np.cos(ang), 1.3*np.sin(ang), color='#bdc3c7', lw=1.2, ls=':')
    for k in range(4):
        th = np.radians(20 + 90*k)
        bl = np.array([[0.15, -0.08], [1.25, -0.18], [1.25, 0.18], [0.15, 0.08]])
        rot = np.array([[np.cos(th), -np.sin(th)], [np.sin(th), np.cos(th)]])
        pts = bl @ rot.T
        ax.fill(pts[:, 0], pts[:, 1], color='#d5dbdb', edgecolor='#7f8c8d')
    ax.scatter([0], [0], s=260, color='#2c3e50', zorder=5)
    th = np.radians(20); p = 1.0*np.array([np.cos(th), np.sin(th)])
    ax.scatter([p[0]], [p[1]], s=150, color='#8e5a2b', zorder=6, edgecolor='white')
    ax.text(p[0]+0.05, p[1]+0.15, 'incrustación', color='#8e5a2b', fontsize=9, fontweight='bold')
    ax.annotate('', xy=p*0.55, xytext=p, arrowprops=dict(arrowstyle='-|>', color='#27ae60', lw=2.4))
    ax.text(0.45, -0.05, 'F del aspa\nsobre el lodo', color='#27ae60', fontsize=8, fontweight='bold')
    q = -0.0*p
    ax.annotate('', xy=(0.42*np.cos(th), 0.42*np.sin(th)), xytext=(0, 0), arrowprops=dict(arrowstyle='-|>', color='#e74c3c', lw=2.4))
    ax.text(-0.75, 0.3, 'tirón sobre\nel eje (3.ª ley)', color='#e74c3c', fontsize=8, fontweight='bold')
    a2 = np.linspace(np.radians(60), np.radians(150), 40)
    ax.plot(1.5*np.cos(a2), 1.5*np.sin(a2), color='#7f8c8d', lw=1.5)
    ax.annotate('', xy=(1.5*np.cos(a2[-1]), 1.5*np.sin(a2[-1])), xytext=(1.5*np.cos(a2[-3]), 1.5*np.sin(a2[-3])),
                arrowprops=dict(arrowstyle='-|>', color='#7f8c8d', lw=1.5))
    ax.text(-0.2, 1.62, 'ω (el tirón gira con el rotor)', color='#7f8c8d', fontsize=9)
    ax.text(0, -1.65, 'Una fuerza que gira = una vibración a la frecuencia de giro',
            ha='center', fontsize=9, color='#2c3e50', fontweight='bold')
    ax.set_xlim(-1.8, 1.8); ax.set_ylim(-1.8, 1.9); ax.set_aspect('equal'); ax.axis('off')
    ax.set_title('Desbalance en el ventilador (vista superior)')
    plt.tight_layout(); plt.show()
esquema()

def desbalance(n=180.0, m=0.800):
    r, Flim = 1.80, 400.0
    ns = np.linspace(0, 260, 300)
    Fs = m*(ns*2*np.pi/60)**2*r
    F0 = m*(n*2*np.pi/60)**2*r
    plt.figure(figsize=(9, 4))
    plt.plot(ns, Fs, lw=2, color='#8e44ad', label='F = m·ω²·r')
    plt.axhline(Flim, color='#e74c3c', ls='--', lw=1.5, label='límite de los rodamientos (400 N)')
    plt.scatter([n], [F0], s=70, zorder=5, color='red' if F0 > Flim else 'green')
    plt.xlabel('velocidad de giro (rpm)'); plt.ylabel('fuerza por desbalance (N)')
    plt.title('La fuerza crece con el cuadrado de las rpm'); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
    n_lim = np.sqrt(Flim/(m*r))*60/(2*np.pi)
    print_es(f"F = {_f3(F0)} N  ·  rpm máximas con esta masa = {_f3(n_lim)} rpm  ·  " +
             ("dentro del límite" if F0 <= Flim else "SUPERA el límite"))

interact(desbalance,
    n=FloatSlider(value=180, min=20, max=260, step=5, description='giro (rpm)'),
    m=FloatSlider(value=0.800, min=0.1, max=1.5, step=0.05, description='masa (kg)'));

def solucion():
    g = 9.81; m, r = 0.800, 1.80
    w1 = 180*2*np.pi/60; F1 = m*w1**2*r
    print_es(f"(i) ω = {_f3(w1)} rad/s;  F = m·ω²·r = 0,800·{_f3(w1)}²·1,80 = {_f3(F1)} N  →  F/(m·g) = {_f3(F1/(m*g))} veces el peso")
    print_es(f"    Verificación: F = m·4π²·f²·r con f = 3,00 Hz → {_f3(m*4*np.pi**2*9*r)} N ✔.  Ya supera el límite de 400 N.")
    w2 = 220*2*np.pi/60; F2 = m*w2**2*r
    print_es(f"(ii) F(220 rpm) = {_f3(F2)} N.  Razón: (220/180)² = {_f3((220/180)**2)} → la fuerza sube {_f3(((220/180)**2-1)*100)} % mientras las rpm suben {_f3((220/180-1)*100)} %")
    wl = np.sqrt(400/(m*r))
    print_es(f"(iii) ω_máx = √(F_lím/(m·r)) = √(400/(0,800·1,80)) = {_f3(wl)} rad/s = {_f3(wl*60/(2*np.pi))} rpm")
    print_es(f"     Verificación por proporción: 180·√(400/{_f3(F1)}) = {_f3(180*np.sqrt(400/F1))} rpm ✔")
    print_es("     Conclusión: con el lodo, el ventilador ni siquiera cumple a 180 rpm. Subir a 220 rpm sin limpiar es")
    print_es("     inaceptable; la solución de fondo es limpiar y balancear, no bajar la ventilación de la torre.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("A 180 rpm, ¿qué fuerza debe ejercer el rotor sobre la incrustación (0,800 kg a 1,80 m) para mantenerla en su círculo?",
    511.6, 0.04, "N", "ω = 18,8 rad/s; F = mω²r = 0,800·18,8²·1,80 = 512 N (65,2 veces su peso).", id_preg="Circ_p4_1", peso=3.0)
quiz_numerico("Si el variador sube el ventilador a 220 rpm sin limpiar el aspa, ¿cuál es la nueva fuerza?",
    764.3, 0.04, "N", "F ∝ ω²: 512·(220/180)² = 764 N.", id_preg="Circ_p4_2", peso=3.0)
quiz_numerico("Sin limpiar el aspa, ¿cuántas rpm como máximo respetan el límite de 400 N de los rodamientos?",
    159.15, 0.04, "rpm", "ω = √(400/(0,800·1,80)) = 16,7 rad/s = 159 rpm.", id_preg="Circ_p4_3", peso=3.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: un estudiante dice que, si el ventilador sube de 180 a 220 rpm (22 % más), la fuerza por desbalance sobre los rodamientos también sube 22 %. ¿Correcto o solo convincente?",
    ['Solo convincente: F = mω²r crece con el cuadrado de ω; (220/180)² = 1,49, así que la fuerza sube 49,4 %',
     'Correcto: la fuerza es proporcional a la rapidez de la incrustación, y esa rapidez sube 22 % junto con las rpm',
     'Correcto: la masa y el radio no cambian, así que la fuerza sigue exactamente el mismo cambio que las rpm',
     'Solo convincente: la fuerza no cambia, porque la incrustación sigue estando a la misma distancia del eje'],
    0, "F = mω²r = mv²/r: depende del CUADRADO de la rapidez. Que m y r no cambien significa que F ∝ ω², no F ∝ ω. Un aumento 'modesto' de rpm casi duplica el castigo a los rodamientos.",
    id_preg="Circ_p4_concep_4", peso=2.0)
```

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# Calificación formativa del taller
calificacion_final(total_preguntas=16)
```
