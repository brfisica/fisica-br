## Punto 1 — Tracción de un vagón del Metro de Medellín  ★★  [CL, UCM]
Un vagón del Metro de Medellín tiene una masa de  m = 40 000 kg  (40 toneladas). Al salir de
una estación, el motor debe acelerarlo uniformemente a  a = 1,10 m/s².

**2ª ley:**  F = m·a

> 🔧 **Aplicación en ingeniería:** dimensionar el sistema de tracción (motores, potencia,
> consumo eléctrico) parte de conocer la fuerza necesaria para acelerar la masa del tren.
> Con pasajeros la masa sube (~48 t), y el motor debe entregar más fuerza: por eso se diseña
> con margen. Es dinámica aplicada al transporte masivo.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


g = 9.81

def esquema():
    fig,ax=plt.subplots(figsize=(7,4))
    ax.axhline(0,color='#8b7355',lw=2,alpha=0.4)
    # vagón como caja
    ax.add_patch(plt.Rectangle((-1,0),2,1.2,color='#2980b9',alpha=0.8,zorder=3))
    ax.text(0,0.6,'vagón',ha='center',va='center',color='white',fontsize=10,fontweight='bold')
    # centro para los vectores
    cx,cy=0,0.6
    # tracción (F, derecha)
    ax.annotate('',xy=(cx+2.5,cy),xytext=(cx,cy),arrowprops=dict(arrowstyle='-|>',color='#27ae60',lw=3))
    ax.text(cx+2.6,cy,'F (tracción)',color='#27ae60',fontsize=10,fontweight='bold',va='center')
    # peso (abajo)
    ax.annotate('',xy=(cx,cy-2),xytext=(cx,cy),arrowprops=dict(arrowstyle='-|>',color='#2c3e50',lw=2.5))
    ax.text(cx+0.15,cy-1.5,'W=mg',color='#2c3e50',fontsize=9)
    # normal (arriba)
    ax.annotate('',xy=(cx,cy+2),xytext=(cx,cy),arrowprops=dict(arrowstyle='-|>',color='#e67e22',lw=2.5))
    ax.text(cx+0.15,cy+1.5,'N (normal)',color='#e67e22',fontsize=9)
    # aceleración resultante
    ax.annotate('',xy=(cx+2,cy-1.6),xytext=(cx+1,cy-1.6),arrowprops=dict(arrowstyle='-|>',color='#e74c3c',lw=2))
    ax.text(cx+1.5,cy-2,'a',color='#e74c3c',fontsize=11,fontweight='bold',ha='center')
    ax.text(0,3,'Diagrama de cuerpo libre: la fuerza NETA horizontal (F) produce la aceleración',
            ha='center',fontsize=9,color='#2c3e50',fontweight='bold')
    ax.text(0,-2.5,'Horizontal: F = m·a  ·  Vertical: N = W (equilibrio)',ha='center',fontsize=9,color='#7f8c8d')
    ax.set_xlim(-4,5); ax.set_ylim(-3,3.3); ax.axis('off')
    ax.set_title('Diagrama de cuerpo libre del vagón (2ª ley)')
    plt.tight_layout(); plt.show()
esquema()

def traccion(m_ton=40.0, a=1.10):
    m=m_ton*1000; F=m*a
    aa=np.linspace(0,2,100); Ff=m*aa
    plt.figure(figsize=(9,4.2)); plt.plot(aa,Ff/1000,lw=2,color='darkred')
    plt.scatter([a],[F/1000],color='blue',s=60,zorder=5,label=f'a={a:.2f} m/s² → F={F/1000:.1f} kN')
    plt.xlabel('aceleración a (m/s²)'); plt.ylabel('fuerza de tracción F (kN)')
    plt.title(f'Fuerza vs aceleración (m={m_ton:.0f} t)'); plt.legend(); plt.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(traccion,
    m_ton=FloatSlider(value=40.0,min=30,max=55,step=1,description='masa (t)'),
    a=FloatSlider(value=1.10,min=0.5,max=2.0,step=0.1,description='a (m/s²)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    m=40000.0; a=1.10; F=m*a
    _pc(f"F = m·a = 40000·1,10 = {F:.3g} N = {F/1000:.3g} kN")
    _pc("Con pasajeros (48 t): F = 48000·1,10 = 52,8 kN → el motor debe dar más fuerza.")
    _pc("Por eso el sistema de tracción se dimensiona para la masa MÁXIMA con margen.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Un vagón del Metro de 40,0 toneladas debe acelerar a 1,10 m/s². Encuentre la fuerza de tracción necesaria.",44,0.04,"kN","F=ma=44,0 kN.", id_preg="Newton_p1_1", peso=2.0)
quiz_opcion_multiple("Si el vagón lleva pasajeros y aumenta su masa, para lograr la misma aceleración, ¿qué ocurre con la fuerza necesaria?",
    ["Disminuye","Aumenta (F=ma)","No cambia","Se anula"],1,
    "F=ma: más masa exige más fuerza. Se diseña para la carga máxima.", id_preg="Newton_p1_2", peso=2.0)
quiz_opcion_multiple("Un vagón vacío (40,0 t) y el mismo vagón lleno (48,0 t) deben alcanzar la MISMA aceleración. ¿Qué se necesita para el vagón lleno?",["La misma fuerza","Una fuerza mayor, proporcional al aumento de masa (F=ma)","Menos fuerza","La aceleración no depende de la masa"],1,"Por F=ma, para igual aceleración la fuerza crece con la masa. El vagón lleno pesa 20% más, así que exige 20% más de tracción. La masa es la resistencia al cambio de movimiento.", id_preg="Newton_p1_3", peso=2.0)

# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---
```

## Punto 4 — Tensión del cable de una grúa torre (Ing. Civil/Construcción)  ★★★  [PRC, UCM, ULC]
Una grúa torre en una obra iza una carga de  m = 1500 kg  sujeta con un estrobo (elemento de
izaje) de  m = 300 kg. El conjunto sube con aceleración  a = 0,80 m/s². Se busca la tensión en
el cable principal (superior) y en el estrobo (inferior).

**Cable superior (sostiene todo):**  T = (m_carga + m_estrobo)·(g + a)
**Estrobo inferior (solo la carga):**  T = m_carga·(g + a)

> 🔧 **Aplicación en ingeniería:** al izar acelerando, la tensión SUPERA el peso (T = m(g+a) > mg).
> Dimensionar el cable, el gancho y el estrobo con la tensión real —no solo con el peso estático—
> es esencial para la seguridad. Ignorar el término de aceleración subestima la carga del cable.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


g = 9.81

def esquema():
    fig,ax=plt.subplots(figsize=(6,4.8))
    # cable
    ax.plot([0,0],[1,4],color='#34495e',lw=2)
    ax.text(0.2,3.5,'cable',fontsize=8,color='#34495e')
    # carga
    ax.add_patch(plt.Rectangle((-0.7,0),1.4,1,color='#c0392b',alpha=0.8,zorder=3))
    ax.text(0,0.5,'carga',ha='center',va='center',color='white',fontsize=10,fontweight='bold')
    cx,cy=0,0.5
    # tensión (arriba)
    ax.annotate('',xy=(cx,cy+2.2),xytext=(cx,cy),arrowprops=dict(arrowstyle='-|>',color='#27ae60',lw=3))
    ax.text(cx+0.2,cy+1.6,'T (tensión)',color='#27ae60',fontsize=10,fontweight='bold')
    # peso (abajo)
    ax.annotate('',xy=(cx,cy-1.8),xytext=(cx,cy),arrowprops=dict(arrowstyle='-|>',color='#2c3e50',lw=2.5))
    ax.text(cx+0.2,cy-1.3,'W=mg',color='#2c3e50',fontsize=9)
    # aceleración
    ax.annotate('',xy=(cx-1.3,cy+1),xytext=(cx-1.3,cy),arrowprops=dict(arrowstyle='-|>',color='#e74c3c',lw=2))
    ax.text(cx-1.6,cy+0.5,'a',color='#e74c3c',fontsize=11,fontweight='bold',ha='right')
    ax.text(0,-2.4,'Izaje acelerado: T − W = m·a  ⟹  T = m(g+a)',ha='center',fontsize=10,color='#2c3e50',fontweight='bold')
    ax.text(0,-3,'La tensión SUPERA el peso porque además debe acelerar la carga',ha='center',fontsize=8,color='#7f8c8d')
    ax.set_xlim(-3,3); ax.set_ylim(-3.3,4.3); ax.axis('off')
    ax.set_title('Diagrama de cuerpo libre: izaje con grúa')
    plt.tight_layout(); plt.show()
esquema()

def izaje(m_carga=1500.0, m_estrobo=300.0, a=0.80):
    M=m_carga+m_estrobo; Tsup=M*(g+a); Tinf=m_carga*(g+a)
    accs=np.linspace(0,2,100); Ts=M*(g+accs)
    plt.figure(figsize=(9,4.2)); plt.plot(accs,Ts/1000,lw=2,color='darkorange')
    plt.axhline(M*g/1000,color='gray',ls='--',alpha=.7,label=f'peso estático = {M*g/1000:.1f} kN')
    plt.scatter([a],[Tsup/1000],color='red',s=60,zorder=5,label=f'a={a:.2f} → T={Tsup/1000:.1f} kN')
    plt.xlabel('aceleración de izaje a (m/s²)'); plt.ylabel('tensión cable superior (kN)')
    plt.title('Tensión crece con la aceleración (T = M(g+a))'); plt.legend(); plt.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(izaje,
    m_carga=FloatSlider(value=1500.0,min=500,max=3000,step=100,description='carga (kg)'),
    m_estrobo=FloatSlider(value=300.0,min=100,max=600,step=50,description='estrobo (kg)'),
    a=FloatSlider(value=0.80,min=0,max=2.0,step=0.1,description='a (m/s²)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    m_carga=1500.0; m_est=300.0; a=0.80; M=m_carga+m_est
    _pc(f"Cable superior (sostiene carga+estrobo): T = (1500+300)·(9,81+0,80) = {M*(g+a):.3g} N = {M*(g+a)/1000:.3g} kN")
    _pc(f"Estrobo inferior (solo la carga): T = 1500·(9,81+0,80) = {m_carga*(g+a):.3g} N = {m_carga*(g+a)/1000:.3g} kN")
    _pc(f"En reposo el cable superior tendría solo {M*g/1000:.3g} kN; al acelerar sube a {M*(g+a)/1000:.3g} kN.")
    _pc("Por eso el cable se dimensiona con la tensión dinámica, no solo con el peso.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Una grúa iza un conjunto de 1800 kg (carga más estrobo) con aceleración de 0,80 m/s² hacia arriba. Encuentre la tensión en el cable superior.",19.1,0.04,"kN","T=M(g+a)≈19,1 kN.", id_preg="Newton_p4_10", peso=3.0)
quiz_opcion_multiple("Una grúa iza carga+estrobo con aceleración hacia arriba. ¿Por qué la tensión del cable superior es MAYOR que si la carga estuviera quieta?",["Porque el cable pesa más","Porque además de sostener el peso, debe aportar la fuerza extra para acelerar (T=m(g+a))","Porque la gravedad aumenta","La tensión sería igual"],1,"Al acelerar hacia arriba, el cable no solo carga el peso mg, sino que añade ma para el impulso: T=m(g+a). Por eso izar con aceleración tensa más el cable que sostener en reposo.", id_preg="Newton_p4_11", peso=3.0)
quiz_opcion_multiple("Al izar una carga acelerando hacia arriba, ¿cómo es la tensión del cable comparada con el peso de la carga?",
    ["Es igual al peso","Es mayor que el peso (T=m(g+a))","Es menor","Es cero"],1,
    "T=m(g+a)>mg: acelerar hacia arriba aumenta la tensión. Clave para la seguridad.", id_preg="Newton_p4_12", peso=3.0)
```
