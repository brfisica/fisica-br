## Punto 2 — Peso y plano inclinado (Ing. Civil/Estructural)  ★★  [CL, UFT]
Un bloque de  m = 50,0 kg  está sobre una rampa. Se estudia su peso y las componentes en un
plano inclinado  θ = 30,0°  (aplicación: rampas, taludes, transporte en obra).

**Peso:**  W = m·g       **Plano θ:**  W∥ = W·sen θ  (baja),  W⊥ = W·cos θ  (normal)

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80

def esquema():
    fig,ax=plt.subplots(figsize=(5,3.2))
    ax.plot([0,3],[0,0],'k-'); ax.plot([0,3],[0,1.73],'k-'); ax.plot([3,3],[0,1.73],'k--',alpha=.3)
    ax.add_patch(plt.Rectangle((1.3,0.75),0.5,0.4,angle=30,color='saddlebrown'))
    ax.annotate('',xy=(1.55,0.3),xytext=(1.55,0.95),arrowprops=dict(arrowstyle='->',color='red',lw=2))
    ax.text(1.75,0.5,'W',color='red')
    ax.text(0.5,0.1,'θ=30°',fontsize=9)
    ax.set_xlim(-0.3,3.5); ax.set_ylim(-0.3,2); ax.axis('off'); ax.set_title('Bloque en plano inclinado')
    plt.tight_layout(); plt.show()
esquema()

def plano(m=50.0, theta=30.0):
    W=m*g; th=np.radians(theta); Wp=W*np.sin(th); Wn=W*np.cos(th)
    ths=np.linspace(0,90,100); Wps=W*np.sin(np.radians(ths)); Wns=W*np.cos(np.radians(ths))
    plt.figure(figsize=(9,4.2))
    plt.plot(ths,Wps,lw=2,color='crimson',label='W∥ = W·sen θ')
    plt.plot(ths,Wns,lw=2,color='navy',label='W⊥ = W·cos θ')
    plt.scatter([theta,theta],[Wp,Wn],color='green',s=50,zorder=5)
    plt.xlabel('ángulo θ (°)'); plt.ylabel('componente (N)'); plt.title(f'θ={theta:.0f}°: W∥={Wp:.0f} N, W⊥={Wn:.0f} N')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(plano,
    m=FloatSlider(value=50.0,min=10,max=100,step=5,description='m (kg)'),
    theta=FloatSlider(value=30.0,min=0,max=90,step=5,description='θ (°)'));

def solucion():
    m=50.0; W=m*g; th=np.radians(30)
    print(f"Peso: W = m·g = 50,0·9,80 = {W:.3g} N")
    print(f"Componente paralela: W∥ = W·sen30° = {W*np.sin(th):.3g} N")
    print(f"Componente normal: W⊥ = W·cos30° = {W*np.cos(th):.3g} N")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_numerico("Peso del bloque (m=50,0 kg):",490,0.03,"N","W=mg=490 N.")
quiz_numerico("Componente paralela al plano (θ=30°):",245,0.04,"N","W∥=W·sen30°=245 N.")
quiz_numerico("Componente normal al plano (θ=30°):",424,0.04,"N","W⊥=W·cos30°≈424 N.")
```

## Punto 3 — Fricción y aceleración (Ing. Mecánica/Mantenimiento)  ★★★  [PRC, ULC]
El bloque de  m = 50,0 kg  descansa en piso horizontal con coeficiente de fricción  μ = 0,40.
Se le aplica una fuerza horizontal de  F = 250 N.

**Fricción:**  f = μ·N  (con N = W en piso horizontal)     **2ª ley:**  a = (F − f)/m

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80

def esquema():
    fig,ax=plt.subplots(figsize=(6,2.4))
    ax.plot([0,6],[0,0],'k-',lw=2)
    ax.add_patch(plt.Rectangle((1.5,0.05),1.2,0.6,color='slategray'))
    ax.annotate('',xy=(3.6,0.35),xytext=(2.8,0.35),arrowprops=dict(arrowstyle='->',color='red',lw=2)); ax.text(3.2,0.55,'F',color='red')
    ax.annotate('',xy=(0.9,0.15),xytext=(1.5,0.15),arrowprops=dict(arrowstyle='->',color='blue',lw=2)); ax.text(1.0,0.3,'f',color='blue')
    ax.set_xlim(-0.3,6.3); ax.set_ylim(-0.3,1); ax.axis('off'); ax.set_title('Fuerza aplicada vs fricción')
    plt.tight_layout(); plt.show()
esquema()

def friccion(m=50.0, mu=0.40, F=250.0):
    N=m*g; f=mu*N; a=(F-f)/m
    Fs=np.linspace(0,600,100); acc=np.clip((Fs-f)/m,0,None)
    plt.figure(figsize=(9,4.2)); plt.plot(Fs,acc,lw=2,color='purple')
    plt.axvline(f,color='blue',ls='--',alpha=.6,label=f'umbral f={f:.0f} N')
    plt.scatter([F],[max(a,0)],color='red',s=60,zorder=5,label=f'F={F:.0f}N → a={max(a,0):.2f} m/s²')
    plt.xlabel('fuerza aplicada F (N)'); plt.ylabel('aceleración a (m/s²)'); plt.title('Aceleración vs fuerza (con fricción)')
    plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(friccion,
    m=FloatSlider(value=50.0,min=10,max=100,step=5,description='m (kg)'),
    mu=FloatSlider(value=0.40,min=0.1,max=0.9,step=0.05,description='μ'),
    F=FloatSlider(value=250.0,min=0,max=600,step=25,description='F (N)'));

def solucion():
    m=50.0; mu=0.40; N=m*g; f=mu*N; F=250.0
    print(f"Normal: N = W = m·g = {N:.3g} N")
    print(f"Fricción: f = μ·N = 0,40·490 = {f:.3g} N")
    print(f"Aceleración: a = (F−f)/m = (250−196)/50,0 = {(F-f)/m:.3g} m/s²")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("Fuerza de fricción (μ=0,40, m=50,0 kg):",196,0.04,"N","f=μmg=196 N.")
quiz_numerico("Aceleración resultante con F=250 N:",1.08,0.06,"m/s²","a=(F−f)/m≈1,08 m/s².")
quiz_opcion_multiple("Si la fuerza aplicada es menor que la fricción máxima, el bloque:",
    ["Acelera igual","No se mueve (a=0)","Retrocede","Vuela"],1,
    "Si F < f_máx, no hay movimiento: la aceleración es cero.")
```

## Punto 4 — Sistema de bloques y tensión (Ing. Mecánica/Sistemas)  ★★★  [PRC, UCM, ULC]
Dos bloques conectados por una cuerda,  m_A = 3,00 kg  y  m_B = 2,00 kg, son empujados por una
fuerza  F = 20,0 N  sobre superficie sin fricción. Halla la aceleración del sistema y la tensión.

**Sistema:**  a = F/(m_A + m_B)       **Tensión (sobre B):**  T = m_B·a

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80

def esquema():
    fig,ax=plt.subplots(figsize=(6,2))
    ax.plot([0,6],[0,0],'k-',lw=2)
    ax.add_patch(plt.Rectangle((1,0.05),0.9,0.6,color='steelblue')); ax.text(1.45,0.35,'A',ha='center',color='white')
    ax.add_patch(plt.Rectangle((2.4,0.05),0.7,0.6,color='firebrick')); ax.text(2.75,0.35,'B',ha='center',color='white')
    ax.plot([1.9,2.4],[0.35,0.35],'k-',lw=1.5)  # cuerda
    ax.annotate('',xy=(1.0,0.35),xytext=(0.3,0.35),arrowprops=dict(arrowstyle='->',color='red',lw=2)); ax.text(0.5,0.55,'F',color='red')
    ax.set_xlim(-0.3,6.3); ax.set_ylim(-0.3,1); ax.axis('off'); ax.set_title('Bloques conectados (empuje sobre A)')
    plt.tight_layout(); plt.show()
esquema()

def sistema(mA=3.0, mB=2.0, F=20.0):
    a=F/(mA+mB); T=mB*a
    plt.figure(figsize=(7,4))
    plt.bar(['a sistema\n(m/s²)','Tensión T\n(N)','Fuerza F\n(N)'],[a,T,F],color=['green','orange','red'])
    plt.title(f'a={a:.2f} m/s², T={T:.2f} N'); plt.grid(alpha=.3,axis='y'); plt.tight_layout(); plt.show()
    print(f"a = {a:.3g} m/s²  |  T = {T:.3g} N")

interact(sistema,
    mA=FloatSlider(value=3.0,min=1,max=8,step=0.5,description='m_A (kg)'),
    mB=FloatSlider(value=2.0,min=1,max=8,step=0.5,description='m_B (kg)'),
    F=FloatSlider(value=20.0,min=5,max=60,step=5,description='F (N)'));

def solucion():
    mA=3.0; mB=2.0; F=20.0; a=F/(mA+mB)
    print(f"Aceleración del sistema: a = F/(m_A+m_B) = 20,0/5,00 = {a:.3g} m/s²")
    print(f"Tensión en la cuerda (fuerza neta sobre B): T = m_B·a = 2,00·{a:.1f} = {mB*a:.3g} N")
    print("La cuerda solo debe acelerar a B, por eso T = m_B·a (menor que F).")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Aceleración del sistema (m_A=3, m_B=2 kg, F=20,0 N):",4.0,0.04,"m/s²","a=F/(mA+mB)=4,00 m/s².")
quiz_numerico("Tensión en la cuerda:",8.0,0.05,"N","T=mB·a=8,00 N.")
quiz_opcion_multiple("La tensión T comparada con la fuerza F aplicada es:",
    ["Igual a F","Menor que F (solo acelera a B)","Mayor que F","Cero"],1,
    "T=mB·a solo debe mover a B, así que T<F.")
```
