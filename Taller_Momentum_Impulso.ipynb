## Punto 1 — Cantidad de movimiento de un vehículo (Ing. Mecánica/Automotriz)  ★★  [CL, UCM]
Un automóvil de  m = 1500 kg  viaja a  v = 20,0 m/s.

**Cantidad de movimiento:**  p = m·v

> 🔧 **Aplicación en ingeniería:** el momentum mide "cuánto movimiento" lleva un cuerpo y es lo que
> debe detenerse en una colisión. Los sistemas de seguridad (barreras, zonas de deformación,
> airbags) se diseñan para absorber esta cantidad de movimiento de forma controlada.

## Punto 3 — Choque perfectamente inelástico (Ing. Civil/Seguridad vial)  ★★★  [CL, UCM, PRC]
Un camión de  m_a = 2000 kg  a  v_a = 15,0 m/s  choca de frente contra un auto de  m_b = 1000 kg
detenido, y quedan enganchados (choque perfectamente inelástico).

**Conservación de momentum:**  m_a·v_a + m_b·v_b = (m_a + m_b)·v_f

> 🔧 **Aplicación en ingeniería:** en un choque inelástico se conserva el momentum pero NO la energía
> cinética (parte se pierde en deformación y calor). El análisis de accidentes (reconstrucción vial)
