**Impulso:**  J = Δp = m·(v₁ − v₀)    **Fuerza media:**  F = J/Δt   (tomando el rebote como negativo)

> el tiempo de impacto reduce la fuerza (por eso sirven los airbags, las zonas de deformación y los

## Punto 5 — Centro de masa de un sistema (Ing. Mecánica/Estructural)  ★★  [CL, PRC]

Se busca la posición del centro de masa.

**Centro de masa:**  x_cm = (m₁·x₁ + m₂·x₂) / (m₁ + m₂)

> 🔧 **Aplicación en ingeniería:** el centro de masa determina el equilibrio y la estabilidad de
