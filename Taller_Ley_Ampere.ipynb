## Punto 1 — Campo de un alambre recto por la ley de Ampère (Ing. Eléctrica/Instalaciones)  ★★  [CL, UCM]
Un alambre recto largo conduce una corriente  I = 20,0 A. Usando la ley de Ampère, se busca el campo
magnético a una distancia  r = 0,100 m.

**Ley de Ampère (alambre recto):**  ∮B·dl = μ₀·I  →  B·(2π·r) = μ₀·I  →  B = μ₀·I/(2π·r)

> 🔧 **Aplicación en ingeniería:** la ley de Ampère da el campo de un conductor sin integrar el aporte
> de cada tramo, aprovechando la simetría circular. Es fundamental para calcular campos e

## Punto 5 — Fuerza entre conductores paralelos (Ing. Eléctrica/Seguridad)  ★★★  [PRC, ULC]
Dos conductores paralelos separados  d = 0,0500 m  llevan cada uno  I = 15,0 A. Se busca la fuerza
magnética por unidad de longitud entre ellos.

**Fuerza por unidad de longitud:**  F/L = μ₀·I₁·I₂/(2π·d)    (atractiva si las corrientes van en el mismo sentido)

> 🔧 **Aplicación en ingeniería:** las corrientes en conductores cercanos ejercen fuerzas entre sí.
> En cortocircuitos, corrientes muy altas producen fuerzas que pueden arrancar barras y cables, por lo
> que el diseño de tableros y subestaciones debe considerar estos esfuerzos electrodinámicos.
