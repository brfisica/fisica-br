## Punto 1 — Campo de un alambre recto por la ley de Ampère (Ing. Eléctrica/Instalaciones)  ★★  [CL, UCM]
Un alambre recto largo conduce una corriente  I = 20,0 A. Usando la ley de Ampère, se busca el campo
magnético a una distancia  r = 0,100 m.

**Ley de Ampère (alambre recto):**  ∮B·dl = μ₀·I  →  B·(2π·r) = μ₀·I  →  B = μ₀·I/(2π·r)

> 🔧 **Aplicación en ingeniería:** la ley de Ampère da el campo de un conductor sin integrar el aporte
> de cada tramo, aprovechando la simetría circular. Es fundamental para calcular campos e
