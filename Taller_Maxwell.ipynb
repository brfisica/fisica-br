**Las cuatro ecuaciones de Maxwell (forma integral):**
1. **Gauss eléctrica:**  ∮E·dA = Q_enc/ε₀    (las cargas crean campo eléctrico)
2. **Gauss magnética:**  ∮B·dA = 0    (no hay monopolos magnéticos: las líneas de B se cierran)
3. **Faraday:**  ∮E·dl = −dΦ_B/dt    (un campo magnético variable crea un campo eléctrico)
4. **Ampère-Maxwell:**  ∮B·dl = μ₀·(I_enc + ε₀·dΦ_E/dt)    (corrientes y campos eléctricos variables crean campo magnético)

**Consecuencias:**  c = 1/√(ε₀·μ₀) en el vacío  ·  v = c/√(ε_r·μ_r) en un medio  ·  en una onda plana E = c·B y λ = c/f

## Punto 1 — La paradoja de Ampère: corriente de desplazamiento en un condensador (Ing. Electrónica/Potencia pulsada)  ★★  [CL, PRC]
Un condensador de placas circulares de radio **R = 5,00 cm** se está cargando. En el instante analizado, por el cable
llega una corriente **I = 2,00 A**. Entre las placas hay vacío: ninguna carga cruza de una placa a la otra.

**Paradoja:** con la ley de Ampère original, un lazo alrededor del cable da ∮B·dl = μ₀·I si la superficie corta el
cable, y 0 si la superficie pasa entre las placas. El mismo lazo no puede tener dos respuestas.
**Corriente de desplazamiento:**  I_d = ε₀·dΦ_E/dt = ε₀·A·dE/dt       (entre las placas, I_d = I)

**Campo B entre las placas (r ≤ R):**  B = μ₀·I·r/(2π·R²)

**Fuera del borde (r ≥ R):**  B = μ₀·I/(2π·r)

> 🔧 **Aplicación en ingeniería:** la corriente de desplazamiento explica por qué un condensador «deja pasar» la
> corriente alterna y por qué los condensadores de potencia pulsada (desfibriladores, soldadura por descarga, láseres)
> producen campos magnéticos donde no hay conductor. Sin este término no habría ondas electromagnéticas.

## Punto 2 — Faraday: el campo eléctrico alrededor del núcleo de un transformador (Ing. Eléctrica/Transformadores)  ★★★  [PRC, UCM]
La columna de un transformador de distribución tiene sección circular de radio **R = 5,00 cm**. Dentro de ella el campo
magnético es uniforme y oscila a 60,0 Hz con amplitud **B₀ = 1,50 T**: B(t) = B₀·sen(2π·f·t). Fuera del núcleo, en
esta idealización, B = 0. Se enrolla una espira de alambre alrededor de la columna.

**Faraday:**  ∮E·dl = −dΦ_B/dt  →  por simetría  E·(2π·r) = |dΦ_B/dt|

**Dentro (r ≤ R):**  E = (r/2)·|dB/dt|

**Fuera (r ≥ R):**  E = (R²/(2r))·|dB/dt|

**Máximo:**  |dB/dt|_máx = 2π·f·B₀

**FEM por espira:**  ε = π·R²·|dB/dt|  (no depende del tamaño de la espira, siempre que rodee la columna)

> 🔧 **Aplicación en ingeniería:** esta es la ley con la que se diseña un transformador: el voltaje por espira lo fija
> el flujo del núcleo (la conocida fórmula V_rms = 4,44·f·N·B₀·A). También explica por qué las láminas del núcleo se
> aíslan entre sí: el campo E inducido dentro del hierro mueve corrientes parásitas que lo calientan.

## Punto 3 — La velocidad de la onda dentro de un cable: localizar una falla (Ing. Electrónica/Telecomunicaciones)  ★★★  [PRC, ULC]
Maxwell combinó sus ecuaciones y obtuvo una onda que viaja a c = 1/√(ε₀·μ₀). Dentro de un material la onda va más
despacio: v = c/√(ε_r·μ_r). El cable coaxial de una estación de radio tiene dieléctrico de polietileno sólido
(**ε_r = 2,25**, μ_r = 1,00, valor típico). Para buscar un daño, un reflectómetro (TDR) envía un pulso por el cable y
recibe el eco de la falla **Δt = 1,50 µs** después.

**Velocidad en el vacío:**  c = 1/√(ε₀·μ₀)

**En el cable:**  v = c/√ε_r ,  factor de velocidad FV = v/c

**Distancia a la falla (ida y vuelta):**  d = v·Δt/2

> 🔧 **Aplicación en ingeniería:** los reflectómetros en el dominio del tiempo localizan cortes, humedad y conectores
> dañados en cables coaxiales, de cobre y de potencia sin excavar ni desmontar. El dato clave que el operador debe
> configurar es el factor de velocidad del cable: con el valor equivocado, la falla aparece en otro lugar.

## Punto 4 — La onda plana: campos, longitud de onda y tamaño de una antena (Ing. de Telecomunicaciones/Radiodifusión)  ★★  [UCM, ULC]
Una emisora de FM transmite en **f = 98,5 MHz**. A varios kilómetros de la antena, la onda que llega a un receptor se
puede tratar como una onda plana con amplitud de campo eléctrico **E₀ = 0,500 V/m**.

**Onda plana:**  E y B perpendiculares entre sí y a la dirección de avance ,  B₀ = E₀/c

**Longitud de onda:**  λ = c/f

**Densidades de energía:**  u_E = ½·ε₀·E² ,  u_B = B²/(2μ₀)

**Antena dipolo de media onda:**  longitud ≈ λ/2

> 🔧 **Aplicación en ingeniería:** el tamaño de una antena se fija con la longitud de onda: un dipolo de media onda,
> un monopolo de cuarto de onda sobre un plano de tierra. Por eso las antenas de FM miden más de un metro y las de wifi,
> unos centímetros. Las antenas reales se recortan unos puntos porcentuales respecto a λ/2 por el grosor del elemento.
