**Las cuatro ecuaciones de Maxwell (forma integral):**
1. **Gauss eléctrica:**  ∮E·dA = Q/ε₀    (las cargas crean campo eléctrico)
2. **Gauss magnética:**  ∮B·dA = 0    (no existen monopolos magnéticos)
3. **Faraday:**  ∮E·dl = −dΦ_B/dt    (un campo magnético variable induce campo eléctrico)
4. **Ampère-Maxwell:**  ∮B·dl = μ₀·(I + ε₀·dΦ_E/dt)    (corrientes y campos eléctricos variables crean campo magnético)

**Consecuencia:**  velocidad de las ondas EM:  c = 1/√(ε₀·μ₀)

Cuarta ecuación de Maxwell. El aporte genial de Maxwell: un campo eléctrico variable también crea
campo magnético, como si hubiera una "corriente de desplazamiento". En un capacitor con área

**Corriente de desplazamiento:**  I_d = ε₀·A·(dE/dt)

la longitud de onda.
