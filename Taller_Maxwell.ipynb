**Las cuatro ecuaciones de Maxwell (forma integral):**
1. **Gauss eléctrica:**  ∮E·dA = Q/ε₀    (las cargas crean campo eléctrico)
2. **Gauss magnética:**  ∮B·dA = 0    (no existen monopolos magnéticos)
3. **Faraday:**  ∮E·dl = −dΦ_B/dt    (un campo magnético variable induce campo eléctrico)
4. **Ampère-Maxwell:**  ∮B·dl = μ₀·(I + ε₀·dΦ_E/dt)    (corrientes y campos eléctricos variables crean campo magnético)

**Consecuencia:**  velocidad de las ondas EM:  c = 1/√(ε₀·μ₀)

Cuarta ecuación de Maxwell. El aporte genial de Maxwell: un campo eléctrico variable también crea
campo magnético, como si hubiera una "corriente de desplazamiento". En un capacitor con área
A = 0,0200 m²  donde el campo cambia a  dE/dt = 1,00×10¹² V/(m·s), se busca esa corriente.

**Corriente de desplazamiento:**  I_d = ε₀·A·(dE/dt)

> 🔧 **Aplicación en ingeniería:** la corriente de desplazamiento explica cómo la corriente "circula"
> a través de un capacitor (aunque no pasen cargas entre las placas) y es lo que permite que existan
> las ondas electromagnéticas. Sin este término, no habría radio, wifi ni telecomunicaciones.

Una onda EM tiene campos eléctrico y magnético perpendiculares, relacionados por  E = c·B. Para una
onda con  E₀ = 300 N/C  y frecuencia  f = 1,00×10⁸ Hz (100 MHz, banda de FM/TV), se buscan  B₀  y
la longitud de onda.

**Relación de campos:**  B₀ = E₀/c    **Longitud de onda:**  λ = c/f

> 🔧 **Aplicación en ingeniería:** todas las telecomunicaciones (radio, TV, wifi, celular, radar) usan
> ondas EM. Conocer la relación entre frecuencia y longitud de onda es esencial para diseñar antenas
> (cuyo tamaño se relaciona con λ) y para asignar bandas del espectro electromagnético.
