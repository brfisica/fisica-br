## Punto 1 — La bobina de un filtro: inductancia y energía de un solenoide (Ing. Electrónica/Fuentes de potencia)  ★★  [CL, UCM]
Para el filtro de una fuente de alimentación se enrolla una bobina de una capa con núcleo de aire:
**N = 800 espiras**, longitud **ℓ = 25,0 cm**, radio **r = 2,00 cm**. Por ella circula una corriente continua **I = 3,00 A**.

**Modelo de solenoide largo:**  B = μ₀·N·I/ℓ ,  L = μ₀·N²·A/ℓ  (A = π·r²)

**Energía:**  U = ½·L·I²

**Fórmula de Wheeler (1928), bobina de una capa:**  L(µH) = r²·N²/(9·r + 10·ℓ)  con r y ℓ en pulgadas

> 🔧 **Aplicación en ingeniería:** las bobinas de filtro de las fuentes conmutadas, los reactores de línea y las bobinas
> de los cargadores almacenan energía en su campo magnético y la devuelven en cada ciclo. Diseñarlas exige saber
> cuánta inductancia da una geometría y cuánta energía guarda a la corriente de trabajo. El modelo de solenoide largo
> supone campo uniforme y sin bordes: es un punto de partida, no el valor de catálogo.

## Punto 2 — ¿Cuánto tarda en cerrar un relé? Transitorio RL (Ing. Mecatrónica/Automatización)  ★★★  [UCM, PRC]
La bobina de un relé de control se conecta a una fuente de **24,0 V** de un tablero. La bobina tiene resistencia
**R = 120 Ω** e inductancia **L = 0,600 H**. El relé cierra sus contactos cuando la corriente alcanza **0,150 A**
(corriente de operación, dato didáctico).

**Conexión (Kirchhoff en la malla):**  ε − I·R − L·dI/dt = 0  →  I(t) = (ε/R)·(1 − e^(−t/τ)) ,  τ = L/R

**Tiempo para llegar a una corriente I₁:**  t₁ = −τ·ln(1 − I₁·R/ε)

> 🔧 **Aplicación en ingeniería:** el retardo de un relé o contactor limita la velocidad de un automatismo, y la misma
> ecuación gobierna el arranque de la corriente en las fases de un motor paso a paso. Para que la corriente suba más
> rápido sin calentar la bobina, los controladores combinan un voltaje mayor con una resistencia en serie (accionamiento
> «L/R») o reducen el voltaje una vez cerrado el relé. Simplificación: en un relé real L aumenta cuando la armadura se
> acerca al núcleo; aquí se toma constante.

## Punto 3 — Abrir un circuito inductivo: el pico de voltaje y el diodo de rueda libre (Ing. Electrónica/Control industrial)  ★★★  [PRC, ULC]
El relé del Punto 2 (R = 120 Ω, L = 0,600 H, I₀ = **0,200 A**) lo enciende y apaga un transistor que soporta como
máximo **100 V** entre sus terminales. El relé suelta sus contactos cuando la corriente baja a **0,0500 A**
(corriente de liberación, dato didáctico). Se comparan tres opciones al apagar el transistor:

1. **Sin protección:** la corriente intenta caer a cero en el tiempo de conmutación del transistor (≈ **1,00 µs**).
2. **Diodo de rueda libre** en paralelo con la bobina: la corriente circula por la bobina y el diodo (caída del diodo despreciable).
3. **Diodo en serie con una resistencia R_s:** la corriente circula por la bobina, el diodo y R_s.

**Autoinducción:**  ε_L = −L·dI/dt

**Desconexión:**  I(t) = I₀·e^(−t/τ') ,  τ' = L/(R + R_s)

**Voltaje en el transistor al abrir (opción 3):**  V_T = ε + I₀·R_s

> 🔧 **Aplicación en ingeniería:** toda bobina que se desconecta (relés, solenoides de válvulas, frenos, motores)
> intenta mantener su corriente. Sin un camino para ella, el voltaje sube hasta que algo cede: el transistor se rompe o
> salta un arco entre contactos (el texto guía, sección 8.3.1, analiza un caso industrial con un motor). El diodo de
> rueda libre da ese camino, a cambio de que el relé tarde más en soltar.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# Signo menos tipográfico (−) en todo texto impreso: «-5» → «−5» (no toca guiones entre palabras, rangos ni 1e-3)
import builtins as _bi_br
if not getattr(_bi_br.print, '_menos_br', False):
    def _print_br(*a, _orig=_bi_br.print, **k):
        import re as _re
        _orig(*[_re.sub(r'(?<![\w.,\-])-(?=\d)', '−', x) if isinstance(x, str) else x for x in a], **k)
    _print_br._menos_br = True
    _bi_br.print = _print_br
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---


def _c3(x):
    # 3 cifras significativas conservando ceros finales (8,00 y no 8); notación ×10ⁿ desde 1000
    if x != 0 and abs(x) >= 1000: return _sci(x)
    t = f"{x:#.3g}"
    return t[:-1] if t.endswith('.') else t

def _sci(x):
    e = int(np.floor(np.log10(abs(x)))); m = x/10**e
    if round(m, 2) >= 10: m /= 10; e += 1
    return f"{m:.2f}×10" + str(e).translate(str.maketrans('-0123456789', '⁻⁰¹²³⁴⁵⁶⁷⁸⁹'))

def _3c(x):
    # 3 cifras significativas con coma decimal; ×10ⁿ fuera de 0,001 ≤ |x| < 1000 (textos de gráficas)
    import math
    try: x = float(x)
    except (TypeError, ValueError): return str(x)
    if x == 0 or not math.isfinite(x): return str(x).replace('.', ',') if x else '0'
    x = float(f"{x:.2e}")
    if abs(x) >= 1000 or abs(x) < 0.001:
        e = int(math.floor(math.log10(abs(x)))); m = x / 10**e
        if round(abs(m), 2) >= 10: m /= 10; e += 1
        return f"{m:.2f}".replace('.', ',') + '×10' + str(e).translate(str.maketrans('-0123456789', '⁻⁰¹²³⁴⁵⁶⁷⁸⁹'))
    t = f"{x:#.3g}"
    return (t[:-1] if t.endswith('.') else t).replace('.', ',')

P3_L, P3_R, P3_I0, P3_V = 0.600, 120.0, 0.200, 24.0

def esquema():
    fig, (a1, a2) = plt.subplots(1, 2, figsize=(10, 3.8))
    t = np.linspace(0, 12, 400)
    for Rs, col, et in ((0, '#27ae60', 'diodo solo'), (380, '#c0392b', 'diodo + 380 Ω')):
        tau = P3_L/(P3_R + Rs)*1000; a1.plot(t, P3_I0*np.exp(-t/tau), color=col, lw=2, label=et)
    a1.axhline(0.05, ls='--', color='#7f8c8d'); a1.text(7, 0.058, 'corriente de liberación', fontsize=8)
    a1.set_xlabel('t después de abrir (ms)'); a1.set_ylabel('I en la bobina (A)'); a1.legend(fontsize=8); a1.set_title('La corriente se extingue por el diodo', fontsize=10)
    Rs = np.linspace(0, 600, 200); a2.plot(Rs, P3_V + P3_I0*Rs, color='#8e44ad', lw=2); a2.axhline(100, color='#2c3e50'); a2.text(20, 104, 'límite del transistor', fontsize=8)
    a2b = a2.twinx(); a2b.plot(Rs, P3_L/(P3_R + Rs)*1000*np.log(4), '--', color='#e67e22', lw=2); a2b.set_ylabel('tiempo de liberación (ms)', color='#e67e22')
    a2.set_xlabel('R_s (Ω)'); a2.set_ylabel('V en el transistor (V)', color='#8e44ad'); a2.set_title('Compromiso: voltaje frente a rapidez', fontsize=10)
    plt.tight_layout(); plt.show()
esquema()

def apertura(Rs=0.0, I0=0.2, V_max=100.0):
    if V_max <= P3_V:
        print_es("⛔ BLOQUEADO: el transistor debe soportar al menos el voltaje de la fuente (24,0 V)."); return
    tau = P3_L/(P3_R + Rs); VT = P3_V + I0*Rs
    if I0 <= 0.05:
        print_es(f"Con I₀ = {_c3(I0)} A la corriente ya está por debajo de la de liberación: el relé suelta de inmediato.  V en el transistor = {_c3(VT)} V"); return
    tl = tau*np.log(I0/0.05)
    print_es(f"τ' = {_c3(tau*1000)} ms   |   suelta a t = {_c3(tl*1000)} ms   |   V en el transistor = {_c3(VT)} V   |   energía a disipar = {_c3(0.5*P3_L*I0**2*1000)} mJ")
    print_es('✅ El voltaje en el transistor está dentro de su límite.' if VT <= V_max else '⚠️ El voltaje en el transistor supera su límite: se rompería.')
    t = np.linspace(0, 5*tau, 300)
    plt.figure(figsize=(8, 3.3)); plt.plot(t*1000, I0*np.exp(-t/tau), color='#8e44ad', lw=2); plt.axhline(0.05, ls='--', color='#7f8c8d')
    plt.xlabel('t después de abrir (ms)'); plt.ylabel('I (A)'); plt.title('Desconexión de la bobina', fontsize=10); plt.tight_layout(); plt.show()

def solucion():
    L, R, I0, V = P3_L, P3_R, P3_I0, P3_V
    print_es(f"a) Sin protección: |ε_L| ≈ L·ΔI/Δt = 0,600·0,200/1,00×10⁻⁶ = {_sci(L*I0/1e-6)} V = {_c3(L*I0/1e-6/1000)} kV.")
    print_es("   Ese voltaje no llega a existir: mucho antes el transistor entra en ruptura o salta un arco. El cálculo dice que algo se va a romper.")
    print_es(f"b) Energía guardada que hay que disipar: U = ½·L·I₀² = {_c3(0.5*L*I0**2*1000)} mJ (en cada apagado)")
    td = (L/R)*np.log(I0/0.05)
    print_es(f"c) Con diodo solo: τ' = L/R = {_c3(L/R*1000)} ms; el transistor ve ≈ 24,0 V; suelta a t = τ'·ln(0,200/0,0500) = {_c3(td*1000)} ms")
    VD = 0.7; tdd = (L/R)*np.log((I0 + VD/R)/(0.05 + VD/R))
    print_es(f"   (con una caída de 0,700 V en el diodo: L·dI/dt = −(I·R + 0,700) → suelta a {_c3(tdd*1000)} ms: la idealización cambia poco)")
    Rs = (100 - V)/I0; tau = L/(R + Rs); tl = tau*np.log(I0/0.05)
    print_es(f"d) Límite del transistor: 24,0 + 0,200·R_s ≤ 100 → R_s ≤ {_c3(Rs)} Ω")
    print_es(f"e) Con R_s = 380 Ω: τ' = 0,600/500 = {_c3(tau*1000)} ms → suelta a {_c3(tl*1000)} ms ({_c3(td/tl)} veces más rápido que con el diodo solo)")
    print_es(f"   R_s se lleva {_c3(100*Rs/(R + Rs))} % de los {_c3(0.5*L*I0**2*1000)} mJ; el resto se disipa en la propia bobina.")
    from scipy.integrate import solve_ivp
    sol = solve_ivp(lambda t, I: [-(R + Rs)*I[0]/L], (0, 0.01), [I0], events=lambda t, I: I[0] - 0.05, rtol=1e-10, atol=1e-12)
    print_es(f"Verificación (solve_ivp con evento I = 0,0500 A): suelta a {_c3(sol.t_events[0][0]*1000)} ms ✓")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("Sin protección, si la corriente cayera de 0,200 A a cero en 1,00 µs, ¿qué voltaje induciría la bobina? (en kV)",
    120, 0.02, "kV", "|ε|=L·ΔI/Δt=0,600·0,200/1,00×10⁻⁶=1,20×10⁵ V=120 kV.", id_preg="RL_p3_1", peso=2.0)
quiz_numerico("Máxima resistencia R_s en serie con el diodo para que el transistor no supere 100 V (en Ω):",
    380, 0.02, "Ω", "24,0+0,200·R_s=100 → R_s=76,0/0,200=380 Ω.", id_preg="RL_p3_2", peso=3.0)
quiz_numerico("Con esa R_s, ¿cuánto tarda el relé en soltar (corriente de 0,200 A a 0,0500 A)? (en ms)",
    1.66, 0.02, "ms", "τ'=0,600/(120+380)=1,20 ms; t=1,20·ln 4=1,66 ms.", id_preg="RL_p3_3", peso=3.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: para que el relé suelte más rápido, el técnico retira el diodo y argumenta 'la fuente es de 24,0 V; un transistor de 100 V tiene margen de sobra'. ¿Correcto o solo convincente?",
    ['Solo convincente: al abrir, el voltaje lo fija L·dI/dt y no la fuente; sin camino para la corriente, los 12,0 mJ de la bobina rompen el transistor',
     'Correcto: ningún voltaje del circuito puede superar el de la fuente',
     'Correcto: sin diodo la energía de la bobina vuelve a la fuente y no hay sobrevoltaje',
     'Solo convincente: sin diodo el relé suelta más despacio, no más rápido'], 0,
    'Que el voltaje de la fuente sea bajo es cierto, y eso lo hace convincente. Pero la bobina se opone al cambio de corriente: si la corriente se corta en microsegundos, ε_L llega a decenas de kilovoltios hasta que algo cede. Para soltar más rápido sin romper nada, se usa R_s (o un zéner) en serie con el diodo.',
    id_preg='RL_p3_concep_3', peso=2.0)

def _interact_coma(fn, _crudos=(), **controles):
    # Como interact, con lectura de sliders en coma decimal. La salida NO se captura con «with salida:»:
    # se calculan el texto y las figuras y se escriben con append_stdout/append_display_data
    # (evita bloqueos de Voilà en Binder y funciona también en JupyterLite).
    import ipywidgets as _w, io as _io, contextlib as _cl
    import matplotlib.pyplot as _plt
    def _fmt(v, paso, crudo=False):
        # Lectura con coma decimal y 3 cifras significativas (×10ⁿ fuera de 0,001 ≤ |v| < 1000).
        # IntSlider: entero exacto. «crudo»: decimales del paso (slider que representa una medición sin redondear).
        import math
        if isinstance(v, int) and not isinstance(v, bool): return str(v)
        if crudo:
            s = repr(float(paso)); dec = 0 if 'e' in s else len(s.split('.')[1].rstrip('0'))
            return f"{v:.{min(dec, 4)}f}".replace('.', ',').replace('-', '−')
        v = float(v)
        if v == 0 or not math.isfinite(v): return '0' if v == 0 else str(v)
        v = float(f"{v:.2e}")
        if abs(v) >= 1000 or abs(v) < 0.001:
            m, e = f"{v:.2e}".split('e')
            return m.replace('.', ',').replace('-', '−') + '×10' + str(int(e)).translate(str.maketrans('-0123456789', '⁻⁰¹²³⁴⁵⁶⁷⁸⁹'))
        t = f"{v:#.3g}"
        return (t[:-1] if t.endswith('.') else t).replace('.', ',').replace('-', '−')
    filas = []
    for nombre, c in controles.items():
        if isinstance(c, (_w.FloatSlider, _w.IntSlider)):
            c.readout = False
            lab = _w.Label(layout=_w.Layout(min_width='90px'))
            def _act(*_, c=c, lab=lab, cr=(nombre in _crudos)): lab.value = _fmt(c.value, c.step, cr)
            c.observe(_act, 'value'); _act()
            filas.append(_w.HBox([c, lab]))
        else:
            filas.append(c)
    salida = _w.Output()
    def _ejecutar(*_):
        partes, buf = [], _io.StringIO()
        def _vaciar_texto():
            t = buf.getvalue()
            if t: partes.append(('t', t)); buf.seek(0); buf.truncate(0)
        def _capturar(*a, **k):
            _vaciar_texto()
            for num in _plt.get_fignums():
                b = _io.BytesIO(); _plt.figure(num).savefig(b, format='png', bbox_inches='tight')
                partes.append(('i', b.getvalue()))
            _plt.close('all')
        _show = _plt.show; _plt.show = _capturar
        try:
            with _cl.redirect_stdout(buf):
                try: fn(**{k: c.value for k, c in controles.items()})
                except Exception as e: print('⚠ No se pudo calcular con estos valores:', e)
            _capturar()
        finally:
            _plt.show = _show
        _vaciar_texto()
        # Reemplazo en UNA sola asignación: la salida nunca queda vacía (sin saltos de la página)
        import base64 as _b64
        nuevas = []
        for tipo, val in partes:
            if tipo == 't': nuevas.append({'output_type': 'stream', 'name': 'stdout', 'text': val})
            else: nuevas.append({'output_type': 'display_data', 'metadata': {},
                                 'data': {'image/png': _b64.b64encode(val).decode('ascii'), 'text/plain': '<Figure>'}})
        salida.outputs = tuple(nuevas)
    for c in controles.values():
        c.observe(_ejecutar, 'value')
    _ejecutar()
    display(_w.VBox(filas + [salida]))

def _explora_sin_cifras(fn):
    # Opción 3: exploración cualitativa; las cifras de este punto solo están en «Mostrar solución».
    import functools, io, contextlib, re as _re
    _pat = _re.compile(r'(?<![\w.,/])[-−]?\d+(?:[.,]\d+)?(?:\s?×\s?10[⁻⁰¹²³⁴⁵⁶⁷⁸⁹]+|[eE][+-]?\d+)?(?![\wπ·(]|[.,]\d)')
    def _m(t): return _pat.sub('…', t)
    def _ocultar():
        for num in plt.get_fignums():
            fig = plt.figure(num)
            textos = list(fig.texts) + ([fig._suptitle] if getattr(fig, '_suptitle', None) else [])
            for ax in fig.axes:
                textos += [ax.title, ax._left_title, ax._right_title] + list(ax.texts)
                if ax.get_legend(): textos += list(ax.get_legend().get_texts())
            for t in textos: t.set_text(_m(t.get_text()))
    @functools.wraps(fn)
    def envuelta(*a, **k):
        _show = plt.show
        def _show2(*sa, **sk):
            _ocultar(); _show(*sa, **sk)
        buf = io.StringIO(); plt.show = _show2
        try:
            with contextlib.redirect_stdout(buf): fn(*a, **k)
            _ocultar()
        finally:
            plt.show = _show
        for linea in buf.getvalue().splitlines():
            print(linea if ('⚠' in linea or '⛔' in linea) else _m(linea))
        print("(Exploración cualitativa: observa tendencias. Las cifras de este punto están en «Mostrar solución».)")
    return envuelta

print("\n--- 🔬 Explora después de responder ---")
print("Los controles NO arrancan en los datos del enunciado. Primero resuelve y responde;\nluego explora tendencias, casos límite y comprueba tu razonamiento.")
_interact_coma(_explora_sin_cifras(apertura),
    Rs=FloatSlider(value=200.0,min=0,max=1000,step=10,description='R_s (Ω)'),
    I0=FloatSlider(value=0.3,min=0.01,max=0.5,step=0.01,description='I₀ (A)'),
    V_max=FloatSlider(value=60.0,min=10,max=400,step=5,description='V_máx (V)'));
```

La etapa de entrada de un receptor de AM es un circuito RLC serie con una bobina de **L = 240 µH** y un condensador
variable. La resistencia de pérdidas de la bobina es **R = 10,0 Ω**. En la Región 2 de la UIT (Américas) la banda de AM
va de 525 a 1 705 kHz, con emisoras cada **10 kHz** (de 530 a 1 700 kHz). Se quiere sintonizar una emisora en
**f₀ = 1 000 kHz**.

**Frecuencia natural:**  f₀ = 1/(2π·√(L·C))  →  C = 1/[(2π·f₀)²·L]

**Calidad y ancho de banda:**  Q = 2π·f₀·L/R ,  Δf = f₀/Q = R/(2π·L)

**Respuesta relativa:**  I/I_máx = 1/√(1 + Q²·(f/f₀ − f₀/f)²)

> 🔧 **Aplicación en ingeniería:** la resonancia LC es la base de la sintonía en radio, de los filtros de las
> telecomunicaciones, de los osciladores y de la transferencia inalámbrica de energía. Un circuito sintonizado debe dejar
> pasar la emisora elegida con su contenido de audio y rechazar las vecinas: el factor de calidad Q fija ese compromiso.
> Los receptores reales (superheterodinos) añaden filtros de frecuencia intermedia; aquí se analiza solo la entrada.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# Signo menos tipográfico (−) en todo texto impreso: «-5» → «−5» (no toca guiones entre palabras, rangos ni 1e-3)
import builtins as _bi_br
if not getattr(_bi_br.print, '_menos_br', False):
    def _print_br(*a, _orig=_bi_br.print, **k):
        import re as _re
        _orig(*[_re.sub(r'(?<![\w.,\-])-(?=\d)', '−', x) if isinstance(x, str) else x for x in a], **k)
    _print_br._menos_br = True
    _bi_br.print = _print_br
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# Calificación formativa del taller
calificacion_final(total_preguntas=16)
```
