## Punto 1 — Apriete de los pernos de una brida en una estación de bombeo (Ing. Mecánica/Mantenimiento)  ★  [CL, UCM]
En una estación de bombeo de un oleoducto, un técnico debe apretar los pernos de una brida a **180 N·m**, el
valor que indica el procedimiento. Su llave mide **0,450 m** desde el centro del perno hasta donde la agarra.

Se busca: (i) la fuerza mínima si tira **perpendicular** al mango; (ii) la fuerza necesaria si una tubería
vecina lo obliga a tirar formando **60,0°** con el mango; (iii) el problema inesperado: el técnico solo puede
ejercer **300 N** en esa dirección incómoda. ¿Qué tan larga debe ser la llave como mínimo?

**Momento de torsión:**  τ = r·F·sen φ    ·    brazo de palanca  d = r·sen φ    ·    τ = F·d
La componente de F a lo largo del mango (F·cos φ) tiene su línea de acción sobre el eje del perno: no lo hace girar.

> 🔧 **Aplicación en ingeniería:** en las bridas de oleoductos y plantas de proceso, un perno flojo deja
> fugar el fluido y uno sobreapretado se estira o daña el empaque. Por eso los procedimientos de
> mantenimiento fijan el torque de apriete, no la fuerza: la fuerza que hace falta depende de la llave y de la
> postura del técnico, el torque no.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))
# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---

def _f3(x):
    import math
    if x == 0: return "0"
    e = int(math.floor(math.log10(abs(x))))
    if e >= 3 or e <= -3:
        m = x/10**e
        if round(abs(m), 2) >= 10: m /= 10; e += 1
        sup = '⁰¹²³⁴⁵⁶⁷⁸⁹'
        ee = ('⁻' if e < 0 else '') + ''.join(sup[int(d)] for d in str(abs(e)))
        return f"{m:.2f}".replace('.', ',') + '×10' + ee
    d = max(0, 2 - e)
    s = f"{x:.{d}f}"
    # si el redondeo sube de orden (p. ej. 99,96 → 100,0) se recalcula
    if len(s.replace('-', '').replace('.', '').lstrip('0')) > 3 and d > 0:
        s = f"{x:.{d-1}f}"
    return s.replace('.', ',')

def _flecha(ax, x0, y0, x1, y1, color, lw=2.4, estilo='-|>'):
    ax.annotate('', xy=(x1, y1), xytext=(x0, y0), arrowprops=dict(arrowstyle=estilo, color=color, lw=lw))

def esquema():
    fig, ax = plt.subplots(figsize=(7.2, 4.6))
    r = 0.45; phi = np.radians(60)
    # perno (hexágono) y llave
    hx = 0.06*np.cos(np.linspace(0, 2*np.pi, 7)); hy = 0.06*np.sin(np.linspace(0, 2*np.pi, 7))
    ax.fill(hx, hy, color='#95a5a6', zorder=4); ax.scatter([0], [0], s=20, color='#2c3e50', zorder=5)
    ax.plot([0, r+0.05], [0, 0], color='#34495e', lw=9, solid_capstyle='round', zorder=3)
    ax.plot([r-0.06, r+0.05], [0, 0], color='#c0392b', lw=11, solid_capstyle='round', zorder=3)
    ax.text(r-0.07, 0.045, 'agarre', ha='center', fontsize=9, color='#c0392b')
    _flecha(ax, 0, -0.30, r, -0.30, '#7f8c8d', lw=1.2, estilo='<->')
    ax.text(r/2, -0.36, 'r = 0,450 m', ha='center', fontsize=10, color='#2c3e50')
    # fuerza a 60° con el mango (el ángulo se mide desde la prolongación de r)
    F = 0.30; fx, fy = F*np.cos(phi), F*np.sin(phi)
    _flecha(ax, r, 0, r+fx, fy, '#e74c3c', lw=2.8)
    ax.text(r+fx+0.01, fy+0.01, 'F', color='#e74c3c', fontsize=13, fontweight='bold')
    a = np.linspace(0, phi, 30); ax.plot(r+0.09*np.cos(a), 0.09*np.sin(a), color='#e74c3c', lw=1)
    ax.plot([r, r+0.22], [0, 0], color='#bdc3c7', lw=1, ls=':')
    ax.text(r+0.115, 0.03, 'φ', color='#e74c3c', fontsize=12)
    # componentes
    _flecha(ax, r, 0, r+fx, 0, '#f39c12', lw=1.8); ax.text(r+fx+0.02, -0.035, 'F·cos φ\n(no gira)', ha='left', fontsize=8.5, color='#d68910')
    _flecha(ax, r+fx, 0, r+fx, fy, '#27ae60', lw=1.8, estilo='-'); ax.text(r+fx+0.02, fy/2, 'F·sen φ\n(gira)', fontsize=8.5, color='#1e8449')
    # línea de acción y brazo de palanca d = r·sen φ
    t = np.linspace(-1.6, 0.6, 2); ax.plot(r+fx*t/F*0.5, fy*t/F*0.5, color='#e74c3c', lw=1, ls='--', alpha=0.6)
    s = r*np.cos(phi)            # pie de la perpendicular desde el perno a la línea de acción
    px, py = r - s*np.cos(phi), -s*np.sin(phi)
    ax.plot([0, px], [0, py], color='#2980b9', lw=2)
    u = np.array([np.cos(phi), np.sin(phi)]); w = np.array([-px, -py])/np.hypot(px, py); q = 0.025
    ax.plot([px+q*u[0], px+q*u[0]+q*w[0], px+q*w[0]], [py+q*u[1], py+q*u[1]+q*w[1], py+q*w[1]], color='#2980b9', lw=1)
    ax.text(px/2-0.16, py/2-0.06, 'd = r·sen φ', color='#2980b9', fontsize=10, fontweight='bold')
    ax.text(0.02, 0.33, 'Solo cuenta la distancia PERPENDICULAR\nentre el eje del perno y la línea de acción de F',
            fontsize=9.5, color='#2c3e50', fontweight='bold')
    ax.set_xlim(-0.25, 0.95); ax.set_ylim(-0.42, 0.45); ax.set_aspect('equal'); ax.axis('off')
    ax.set_title('Llave sobre el perno de una brida: brazo de palanca')
    plt.tight_layout(); plt.show()
esquema()

def llave(F=400.0, r=0.45, phi=90.0):
    tau_obj = 180.0
    tau = r*F*np.sin(np.radians(phi))
    phis = np.linspace(0, 180, 361)
    plt.figure(figsize=(9, 4))
    plt.plot(phis, r*F*np.sin(np.radians(phis)), lw=2, color='#2980b9', label='τ = r·F·sen φ (con tu F y tu r)')
    plt.axhline(tau_obj, color='#27ae60', ls='--', lw=1.5, label='torque del procedimiento: 180 N·m')
    plt.fill_between(phis, tau_obj*1.1, max(1.2*r*F, 1.3*tau_obj), color='#f5b7b1', alpha=0.3, label='sobreapriete (> +10 %)')
    plt.scatter([phi], [tau], s=70, color='#e74c3c', zorder=5)
    plt.xlabel('ángulo φ entre el mango y la fuerza (grados)'); plt.ylabel('τ (N·m)')
    plt.title('¿Llegas al torque de apriete?'); plt.legend(fontsize=8, loc='lower center'); plt.tight_layout(); plt.show()
    if tau < 0.97*tau_obj: est = 'perno FLOJO: falta torque'
    elif tau <= 1.10*tau_obj: est = 'apriete dentro del rango'
    else: est = 'SOBREAPRIETE: riesgo de estirar el perno'
    print_es(f"τ = {_f3(r)}·{_f3(F)}·sen({phi:.0f}°) = {_f3(tau)} N·m   ·   brazo de palanca d = {_f3(r*np.sin(np.radians(phi)))} m   ·   {est}")

interact(llave,
    F=FloatSlider(value=400.0, min=50, max=800, step=10, description='F (N)'),
    r=FloatSlider(value=0.45, min=0.15, max=1.00, step=0.05, description='r (m)'),
    phi=FloatSlider(value=90.0, min=0, max=180, step=5, description='φ (grados)'));

def solucion():
    tau = 180.0; r = 0.450; phi = np.radians(60.0)
    print_es(f"(i) φ = 90° → sen φ = 1:  F = τ/r = 180/0,450 = {_f3(tau/r)} N")
    F2 = tau/(r*np.sin(phi))
    print_es(f"(ii) φ = 60,0°:  F = τ/(r·sen φ) = 180/(0,450·{_f3(np.sin(phi))}) = {_f3(F2)} N")
    print_es(f"     Verificación por componentes: F·sen φ = {_f3(F2*np.sin(phi))} N (la que gira) y F·cos φ = {_f3(F2*np.cos(phi))} N (tira a lo largo del mango, su línea de acción pasa por el eje: τ = 0).")
    print_es(f"     Verificación por brazo de palanca: d = r·sen φ = {_f3(r*np.sin(phi))} m;  F·d = {_f3(F2*r*np.sin(phi))} N·m ✔")
    rmin = tau/(300.0*np.sin(phi))
    print_es(f"(iii) r_mín = τ/(F·sen φ) = 180/(300·{_f3(np.sin(phi))}) = {_f3(rmin)} m")
    print_es(f"     Comprobación: 300·{_f3(rmin)}·sen 60,0° = {_f3(300*rmin*np.sin(phi))} N·m ✔.  Una llave de 0,450 m no alcanza: hace falta una más larga o cambiar de postura.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("¿Qué fuerza mínima debe hacer el técnico si tira perpendicular al mango (r = 0,450 m, τ = 180 N·m)?",
    400.0, 0.03, "N", "F = τ/r = 180/0,450 = 400 N.", id_preg="Torq_p1_1", peso=1.0)
quiz_numerico("Si debe tirar formando 60,0° con el mango, ¿qué fuerza necesita para lograr los mismos 180 N·m?",
    461.88, 0.03, "N", "F = τ/(r·sen φ) = 180/(0,450·0,866) = 462 N. Brazo de palanca: d = 0,390 m.", id_preg="Torq_p1_2", peso=1.0)
quiz_numerico("Problema inesperado: en esa postura (60,0°) el técnico solo puede ejercer 300 N. ¿Cuál es la longitud mínima de la llave (del perno al agarre)?",
    0.69282, 0.03, "m", "r = τ/(F·sen φ) = 180/(300·0,866) = 0,693 m. Con 0,450 m solo lograría 117 N·m.", id_preg="Torq_p1_3", peso=1.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: un estudiante dice que a 60,0° hay que hacer más fuerza que a 90° porque una parte de la fuerza se gasta tirando a lo largo del mango, sin hacerlo girar. ¿Correcto o solo convincente?",
    ['Correcto: la componente F·cos φ va a lo largo del mango y pasa por el eje, su torque es cero; solo F·sen φ hace girar: F = 462 N',
     'Correcto: al inclinar la fuerza, el brazo de palanca crece hasta r/sen φ = 0,520 m, y por eso la fuerza necesaria también aumenta',
     'Solo convincente: el torque depende solo de la fuerza y de la longitud de la llave; la dirección no cambia nada y bastan 400 N',
     'Solo convincente: a 60,0° hace falta menos fuerza, porque la componente a lo largo del mango también ayuda a girar el perno'],
    0, "La afirmación es correcta y la razón también. La opción del brazo de palanca tiene el veredicto bien pero la física al revés: d = r·sen φ = 0,390 m es MENOR que r; por eso se requiere más fuerza. Una componente cuya línea de acción pasa por el eje no produce torque.",
    id_preg="Torq_p1_concep_1", peso=2.0)
```

## Punto 2 — Talanquera de un peaje con contrapeso (Ing. Civil/Electromecánica)  ★★  [UFT, PRC]
En un peaje de una vía nacional, la talanquera es un brazo uniforme de aluminio de **4,00 m** de largo y
**12,0 kg**. Gira sobre un pivote ubicado a **0,500 m** de uno de sus extremos; en ese extremo corto lleva un
**contrapeso** para que el motor no tenga que sostener todo el brazo.

Se busca: (i) el torque del peso del brazo respecto al pivote cuando está horizontal; (ii) la masa del
contrapeso que deja el torque neto en cero; (iii) el problema inesperado: el contrapeso instalado es de
**30,0 kg**. ¿Qué torque debe hacer el motor para mantener el brazo levantado a **60,0°**?

**Torque de un peso respecto al pivote:**  τ = m·g·x·cos θ,  con x la distancia del centro de gravedad al
pivote medida a lo largo del brazo y θ el ángulo del brazo con la horizontal.
**Signos:** el brazo largo tiende a bajar (horario, −); el contrapeso tiende a subirlo (antihorario, +).

## Punto 3 — Aviso vial colgado de una viga con tirante (Ing. Civil/Estructuras)  ★★★  [PRC, ULC]
A la entrada de un puente, un aviso de señalización de **60,0 kg** cuelga de una viga horizontal uniforme de

viga (con su sentido); (iii) el problema inesperado: el cable disponible resiste con seguridad **900 N**.
¿Cuál es el ángulo mínimo que debe formar con la viga?

**Equilibrio:**  ΣFₓ = 0  ·  ΣF_y = 0  ·  Στ_A = 0.  Tomar torques respecto a A elimina las dos componentes

> diseñan con este análisis. La bisagra o el anclaje al muro deben resistir la fuerza calculada, y el ángulo
> del tirante define cuánto se carga el cable: un tirante muy "acostado" obliga a usar cables y anclajes
> mucho más robustos.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))
# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---

def _f3(x):
    import math
    if x == 0: return "0"
    e = int(math.floor(math.log10(abs(x))))
    if e >= 3 or e <= -3:
        m = x/10**e
        if round(abs(m), 2) >= 10: m /= 10; e += 1
        sup = '⁰¹²³⁴⁵⁶⁷⁸⁹'
        ee = ('⁻' if e < 0 else '') + ''.join(sup[int(d)] for d in str(abs(e)))
        return f"{m:.2f}".replace('.', ',') + '×10' + ee
    d = max(0, 2 - e)
    s = f"{x:.{d}f}"
    # si el redondeo sube de orden (p. ej. 99,96 → 100,0) se recalcula
    if len(s.replace('-', '').replace('.', '').lstrip('0')) > 3 and d > 0:
        s = f"{x:.{d-1}f}"
    return s.replace('.', ',')

# Calificación formativa del taller
calificacion_final(total_preguntas=16)
```
