Un disco parte del reposo y acelera angularmente a  α = 2,50 rad/s²  durante  t = 6,00 s. Se busca
su velocidad angular final y el ángulo girado.

**Cinemática angular:**  ω = ω₀ + α·t    ·    θ = ω₀·t + ½·α·t²

> 🔧 **Aplicación en ingeniería:** el arranque de motores, turbinas y ejes sigue esta cinemática

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


def esquema():
    fig,ax=plt.subplots(figsize=(6,5))
    ang=np.linspace(0,2*np.pi,100)
    # disco
    for R in [1,2,3]:
        ax.plot(R*np.cos(ang),R*np.sin(ang),color='#bdc3c7',lw=1,ls=':')
    ax.plot(3*np.cos(ang),3*np.sin(ang),color='#34495e',lw=2)
    ax.scatter([0],[0],s=80,color='#2c3e50',zorder=5)
    # puntos a distintos radios con vector velocidad (tangente, mayor a más radio)
    for R,col in [(1,'#27ae60'),(2,'#e67e22'),(3,'#e74c3c')]:
        px,py=R*np.cos(np.radians(45)),R*np.sin(np.radians(45))
        ax.scatter([px],[py],s=90,color=col,zorder=5,edgecolor='white',linewidth=1.5)
        # velocidad tangencial (perpendicular al radio)
        vlen=R*0.5
        tx,ty=-np.sin(np.radians(45)),np.cos(np.radians(45))
        ax.annotate('',xy=(px+tx*vlen,py+ty*vlen),xytext=(px,py),
                    arrowprops=dict(arrowstyle='-|>',color=col,lw=2.2))
        ax.text(px+tx*vlen*1.3,py+ty*vlen*1.3,f'v={R}ω',color=col,fontsize=9,fontweight='bold')
    ax.annotate('',xy=(2*np.cos(np.radians(20)),2*np.sin(np.radians(20))),xytext=(0,0),
                arrowprops=dict(arrowstyle='->',color='#7f8c8d',lw=1))
    ax.text(1,0.2,'r',color='#7f8c8d',fontsize=10)
    ax.text(0,-3.6,'Mismo ω, pero mayor radio → mayor velocidad lineal (v=ω·r)',
            ha='center',fontsize=9,color='#2c3e50',fontweight='bold')
    ax.set_xlim(-3.5,4.5); ax.set_ylim(-4,3.8); ax.axis('off'); ax.set_aspect('equal')
    ax.set_title('Velocidad lineal según el radio')
    plt.tight_layout(); plt.show()
esquema()

def lineal_angular(w=15.0, r=0.30):
    v=w*r; ac=w**2*r
    ws=np.linspace(1,30,100); vs=ws*r
    plt.figure(figsize=(9,4)); plt.plot(ws,vs,lw=2,color='teal')
    plt.scatter([w],[v],color='red',s=60,zorder=5,label=f'ω={w:.0f} rad/s → v={v:.2f} m/s, a_c={ac:.1f} m/s²')
    plt.xlabel('velocidad angular ω (rad/s)'); plt.ylabel('velocidad lineal v (m/s)')
    plt.title('v = ω·r (proporcional al radio)'); plt.legend(); plt.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(lineal_angular,
    w=FloatSlider(value=15.0,min=1,max=30,step=1,description='ω (rad/s)'),
    r=FloatSlider(value=0.30,min=0.05,max=0.6,step=0.05,description='r (m)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)
    w=15.0; r=0.30
    _pc(f"Velocidad lineal: v = ω·r = 15,0·0,300 = {w*r:.2f} m/s")
    _pc(f"Aceleración centrípeta: a_c = ω²·r = 15,0²·0,300 = {w**2*r:.3g} m/s²")
    _pc("Un punto más alejado del eje (mayor r) se mueve más rápido.")
    _pc("Molinillo: el eje queda quieto y cada borde del mango va con su palma (sin deslizar): u = ω·r_mango")
    _pc(f"  ω = 0,400/0,0120 = {0.4/0.012:.3g} rad/s = {0.4/0.012*60/(2*np.pi):.3g} rpm;  borde de la estrella: v = ω·0,0300 = {0.4/0.012*0.03:.3g} m/s")
    _pc(f"  Verificación: v = u·(3,00/1,20) = {0.4*3/1.2:.3g} m/s ✔;  a_c = v²/r = {(0.4/0.012)**2*0.03:.3g} m/s²")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_numerico("Una rueda gira a 15,0 rad/s y tiene radio 0,300 m. Encuentre la velocidad lineal de un punto del borde.",4.5,0.04,"m/s","v=ω·r=4,50 m/s.", id_preg="Rotaci_p2_4", peso=2.0)
quiz_numerico("Para esa misma rueda, encuentre la aceleración centrípeta de un punto del borde.",67.5,0.04,"m/s²","a_c=ω²·r=67,5 m/s².", id_preg="Rotaci_p2_5", peso=2.0)
quiz_opcion_multiple("En una rueda que gira, si comparamos un punto del borde con uno más cercano al centro, el del borde tiene:",
    ['Menor velocidad lineal (gira menos)', 'Mayor velocidad lineal (v = ω·r)', 'La misma velocidad lineal (misma ω)', 'Velocidad angular distinta'],1,
    "Todos tienen la misma ω, pero v=ω·r es mayor donde r es mayor.", id_preg="Rotaci_p2_6", peso=2.0)
quiz_numerico("Molinillo: las palmas se mueven en sentidos opuestos a 0,400 m/s cada una y el mango (radio 1,20 cm) rueda entre ellas sin deslizar. ¿Qué rapidez tiene el borde de la estrella, de radio 3,00 cm?",
    1.0, 0.04, "m/s", "ω = u/r_mango = 0,400/0,0120 = 33,3 rad/s (318 rpm); v = ω·0,0300 = 1,00 m/s. El borde de la estrella va 2,50 veces más rápido que las manos.", id_preg="Rotaci_p2_16", peso=2.0)
quiz_opcion_multiple("DEFIENDE TU CONCLUSIÓN: un estudiante dice que, en una rueda que gira, un punto del borde y uno cerca del centro tienen la misma velocidad lineal porque 'giran juntos'. ¿Correcto o solo convincente?",
    ['Solo convincente: comparten ω, pero v = ω·r; el punto del borde, con mayor r, se mueve más rápido',
     'Correcto: si giran juntos, dan una vuelta en el mismo tiempo y por eso van igual de rápido',
     'Correcto: la velocidad lineal es la misma porque la velocidad angular es la misma',
     'Solo convincente: el punto cercano al centro va más rápido, porque recorre una circunferencia más cerrada'],
    0,
    "Mismo tiempo por vuelta (misma ω), pero el borde recorre una circunferencia más larga en ese tiempo: v = ω·r. Una circunferencia 'más cerrada' es más corta, no más rápida.",
    id_preg='Rot_p2_concep_2', peso=2.0)
```

## Punto 3 — Momento de inercia y torque (Ing. Mecánica/Diseño)  ★★★  [PRC, ULC]
Un disco sólido de  M = 5,00 kg  y radio  R = 0,200 m  debe acelerar angularmente a  α = 8,00 rad/s².
Se busca su momento de inercia y el torque necesario.

**Momento de inercia (disco sólido):**  I = ½·M·R²    **Torque:**  τ = I·α

> 🔧 **Aplicación en ingeniería:** el momento de inercia es la "resistencia a girar" de una pieza, el
> análogo rotacional de la masa. Dimensionar el torque de un motor para acelerar un volante, una
> polea o un rotor parte de calcular I. Distribuir la masa lejos del eje aumenta I.

> 🤖 **TIP-IA — Traductor de representaciones:**
> Pega τ=I·α a un asistente: *"Explícame qué representa el momento de inercia como 'resistencia a girar', y por qué depende de CÓMO se distribuye la masa (no solo de cuánta hay). Dame la intuición con un ejemplo cotidiano."*

**Conservación (sin torque externo):**  I₁·ω₁ = I₂·ω₂

> 🔧 **Aplicación en ingeniería:** al no haber torque externo, el momento angular se conserva. Esto
> explica por qué un patinador gira más rápido al recoger los brazos, y se aplica en el control de
> orientación de satélites y en el diseño de rotores y giroscopios.

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))


def _num_coma(s):
    import re as _re
    # reemplaza punto decimal por coma en números dentro de un string ya formateado
    return _re.sub(r"(\d)\.(\d)", r"\1,\2", s)
def print_es(*args, **kwargs):
    args2=[_num_coma(a) if isinstance(a,str) else a for a in args]
    print(*args2, **kwargs)

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

try:
    print_es
except NameError:
    def print_es(*args, **kwargs):
        import re as _re
        print(*[_re.sub(r"(\d)\.(\d)", r"\1,\2", a) if isinstance(a, str) else a for a in args], **kwargs)
# --- Cuestionarios Física BR v3: opciones barajadas, crédito por intento, nota sobre el total ---
import random as _rnd_br
try:
    PESOS_BR
except NameError:
    PESOS_BR = {}      # id_preg -> peso de cada pregunta mostrada
try:
    INTENTOS_BR
except NameError:
    INTENTOS_BR = {}   # id_preg -> [intentos, resuelta]
_CREDITO_BR = {1: 1.0, 2: 0.5, 3: 0.25}   # crédito según el intento en que se acierta
_MAX_INT_BR = 3

def _evaluar_intento(id_preg, ok, peso, explicacion, msg_ok, msg_mal):
    est = INTENTOS_BR.setdefault(id_preg, [0, False])
    if est[1]:
        print("✔ Ya resolviste esta pregunta; tu crédito quedó registrado.")
        if explicacion: print_es("→ " + explicacion)
        return
    if est[0] >= _MAX_INT_BR:
        print("⛔ Agotaste los 3 intentos con crédito: esta pregunta ya no suma.")
        if explicacion: print_es("→ " + explicacion)
        return
    est[0] += 1
    n = est[0]
    if ok:
        est[1] = True
        cred = _CREDITO_BR.get(n, 0.0)
        REGISTRO_BR[id_preg] = (cred, float(peso))
        print_es(msg_ok + f"  (intento {n}: crédito {int(cred*100)} %)")
        if explicacion: print_es("→ " + explicacion)
    else:
        REGISTRO_BR.setdefault(id_preg, (0.0, float(peso)))
        if n < _MAX_INT_BR:
            print(msg_mal + f"  Intento {n} de {_MAX_INT_BR}. Revisa tu razonamiento antes de volver a intentar.")
        else:
            print(msg_mal + " Agotaste los 3 intentos: esta pregunta ya no suma. Estudia la explicación:")
            if explicacion: print_es("→ " + explicacion)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    orden = list(range(len(opciones))); _rnd_br.shuffle(orden)   # orden distinto en cada sesión
    correcta = orden.index(indice_correcto)
    letras = "abcdefgh"
    etiquetas = [f"{letras[i]}) {opciones[j]}" for i, j in enumerate(orden)]
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=etiquetas, value=None, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            if radio.index is None:
                print("⚠ Elige una opción."); return
            _evaluar_intento(id_preg, radio.index == correcta, peso, explicacion,
                             "✅ ¡Correcto!", "❌ Incorrecto.")
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    if id_preg is None: id_preg = f"_anon_{len(PESOS_BR)}"
    PESOS_BR[id_preg] = float(peso)
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.strip().replace(",", "."))
            except ValueError: print("⚠ Escribe un número (usa coma o punto decimal)."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            _evaluar_intento(id_preg, ok, peso, explicacion,
                             f"✅ ¡Correcto!  ({valor_correcto:.3g} {unidad})", "❌ Fuera de rango.")
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not PESOS_BR:
                print("No se encontraron cuestionarios en este taller."); return
            peso_total = sum(PESOS_BR.values())
            peso_ok = sum(float(c)*p for c, p in REGISTRO_BR.values())
            nota = 5.0*peso_ok/peso_total if peso_total > 0 else 0.0
            total = len(PESOS_BR)
            resp = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0])[0] > 0)
            prim = sum(1 for k in PESOS_BR if INTENTOS_BR.get(k, [0, False])[1] and INTENTOS_BR[k][0] == 1)
            print("="*46)
            print("          CALIFICACIÓN FORMATIVA")
            print("="*46)
            print(f"  Preguntas respondidas:       {resp} de {total}")
            print(f"  Acertadas al primer intento: {prim}")
            print_es(f"  Puntaje ponderado:           {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*46)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*46)
            if resp < total: print(f"  ⚠ Faltan {total-resp} preguntas: cuentan como 0.")
            if nota >= 3.5: print("  ✅ ¡Buen trabajo! Dominas los conceptos.")
            elif nota >= 3.0: print("  🟡 Vas bien, repasa lo que fallaste.")
            else: print("  🔴 Repasa la teoría y vuelve a los puntos.")
            print("\n  Crédito por pregunta: 100 % al primer intento,")
            print("  50 % al segundo, 25 % al tercero. Predice y")
            print("  razona ANTES de verificar: adivinar cuesta nota.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>📊 Tu calificación del taller</h3>"
        "<i>La nota se calcula sobre TODAS las preguntas del taller y premia acertar al primer intento.</i>"),
        boton, salida]))


# Calificación formativa del taller
calificacion_final(total_preguntas=21)
```
