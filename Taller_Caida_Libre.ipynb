## Punto 1 — Caída libre de una herramienta en obra  ★★  [CL, PRC]
En la construcción de un edificio en Bogotá, una herramienta se suelta accidentalmente
(v₀ = 0) desde un andamio a  h = 60,0 m  de altura.

**Modelo (caída libre desde reposo):**  h = ½·g·t²  →  t = √(2·h/g),   v = g·t

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    fig,ax=plt.subplots(figsize=(3,4.5))
    ax.plot([0,1.5],[4,4],'k-',lw=3); ax.text(0.75,4.1,'andamio',ha='center')
    ax.scatter([0.5],[3.6],s=120,color='dimgray',zorder=5)
    for yy in [3.0,2.3,1.5,0.6]:
        ax.annotate('',xy=(0.5,yy-0.3),xytext=(0.5,yy),arrowprops=dict(arrowstyle='->',color='navy'))
    ax.text(0.7,2.0,'g',color='navy',fontsize=12)
    ax.plot([-0.3,1.5],[0,0],color='brown',lw=3)
    ax.annotate('',xy=(1.3,0),xytext=(1.3,4),arrowprops=dict(arrowstyle='<->')); ax.text(1.4,2,'h',fontsize=11)
    ax.set_xlim(-0.5,2); ax.set_ylim(-0.5,4.7); ax.axis('off'); ax.set_title('Caída libre desde reposo')
    plt.tight_layout(); plt.show()
esquema()

def caida_libre(h=60.0):
    t_fin=np.sqrt(2*h/g); t=np.linspace(0,t_fin,200); y=h-0.5*g*t**2; v=g*t
    fig,(a1,a2)=plt.subplots(1,2,figsize=(10,4))
    a1.plot(t,y,lw=2,color='navy'); a1.set_xlabel('t (s)'); a1.set_ylabel('altura y (m)'); a1.set_title('Altura vs tiempo'); a1.grid(alpha=.3)
    a2.plot(t,v,lw=2,color='crimson'); a2.set_xlabel('t (s)'); a2.set_ylabel('rapidez v (m/s)'); a2.set_title('Rapidez vs tiempo'); a2.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(caida_libre, h=FloatSlider(value=60.0,min=5,max=250,step=5,description='h (m)'));

def solucion():
    h=60.0; t=np.sqrt(2*h/g); v=g*t
    print(f"Tiempo de caída:  t = √(2h/g) = √(2·60,0/9,80) = {t:.3g} s")
    print(f"Velocidad de impacto:  v = g·t = 9,80·{t:.3g} = {v:.3g} m/s = {v*3.6:.3g} km/h")
    print(f"Verificación (v²=2gh):  {v**2:.0f} ≈ {2*g*h:.0f} ✓")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Tiempo de caída desde h=60,0 m:",3.50,0.04,"s","t=√(2h/g)=3,50 s.")
quiz_numerico("Velocidad de impacto:",34.3,0.04,"m/s","v=gt=9,80·3,50≈34,3 m/s.")
quiz_opcion_multiple("Si la herramienta fuera el doble de pesada, el tiempo de caída:",
    ["Se reduce a la mitad","No cambia: la caída no depende de la masa","Se duplica","Aumenta un poco"],1,
    "En caída libre (sin aire) todos los cuerpos caen con la misma g, sin importar su masa.")
```

## Punto 2 — Herramienta lanzada hacia arriba (refinería)  ★★★  [CL, UCM, PRC]
Un operario en la refinería de Cartagena lanza una herramienta **verticalmente hacia arriba**
con  v₀ = 8,50 m/s  desde una plataforma a  y₀ = 6,00 m  sobre el suelo.

**Modelo (tiro vertical):**  altura máx sobre plataforma = v₀²/(2·g);  el tiempo total se halla
resolviendo  0 = y₀ + v₀·t − ½·g·t².

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    fig,ax=plt.subplots(figsize=(3,4.5))
    ax.plot([-0.3,1.3],[2,2],'k-',lw=3); ax.text(0.5,1.7,'plataforma',ha='center')
    ax.annotate('',xy=(0.5,3.3),xytext=(0.5,2.2),arrowprops=dict(arrowstyle='->',color='red',lw=2))
    ax.text(0.65,2.8,'v₀',color='red',fontsize=12)
    ax.plot([-0.5,1.3],[0,0],color='brown',lw=3)
    ax.annotate('',xy=(1.1,0),xytext=(1.1,2),arrowprops=dict(arrowstyle='<->')); ax.text(1.2,1,'y₀',fontsize=11)
    ax.set_xlim(-0.7,1.8); ax.set_ylim(-0.4,3.6); ax.axis('off'); ax.set_title('Lanzamiento vertical hacia arriba')
    plt.tight_layout(); plt.show()
esquema()

def tiro_vertical(v0=8.50, y0=6.00):
    a_,b_,c_=0.5*g,-v0,-y0
    t_tot=(-b_+np.sqrt(b_**2-4*a_*c_))/(2*a_)
    t=np.linspace(0,t_tot,200); y=y0+v0*t-0.5*g*t**2
    plt.figure(figsize=(9,4.5)); plt.plot(t,y,lw=2,color='darkgreen')
    h_max=y0+v0**2/(2*g); t_top=v0/g
    plt.scatter([t_top],[h_max],color='orange',s=60,zorder=5,label=f'altura máx={h_max:.3g} m')
    plt.scatter([t_tot],[0],color='red',s=60,zorder=5,label=f'impacto t={t_tot:.3g} s')
    plt.axhline(0,color='brown',lw=1); plt.xlabel('t (s)'); plt.ylabel('altura sobre el suelo (m)')
    plt.title('Trayectoria vertical'); plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(tiro_vertical,
    v0=FloatSlider(value=8.50,min=1,max=20,step=0.5,description='v₀ (m/s)'),
    y0=FloatSlider(value=6.00,min=0,max=20,step=1,description='y₀ (m)'));

def solucion():
    v0=8.50; y0=6.00
    h_max=y0+v0**2/(2*g)
    a_,b_,c_=0.5*g,-v0,-y0; t_tot=(-b_+np.sqrt(b_**2-4*a_*c_))/(2*a_)
    v_imp=abs(v0-g*t_tot)
    print(f"Altura máxima sobre la plataforma:  v₀²/(2g) = {v0**2/(2*g):.3g} m")
    print(f"Altura máxima sobre el suelo:  y₀ + v₀²/(2g) = {h_max:.3g} m")
    print(f"Tiempo total en el aire:  t = {t_tot:.3g} s")
    print(f"Velocidad de impacto:  |v| = {v_imp:.3g} m/s")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 2 ---")
quiz_numerico("Altura máxima sobre el suelo (v₀=8,50 m/s, y₀=6,00 m):",9.69,0.04,"m",
    "y₀+v₀²/(2g)=6,00+3,69≈9,69 m.")
quiz_numerico("Tiempo total en el aire:",2.27,0.05,"s",
    "Resolviendo 0=6,00+8,50t−4,90t² ⟹ t≈2,27 s.")
quiz_opcion_multiple("En el punto más alto, la velocidad y la aceleración valen:",
    ["Ambas cero","v=0 pero a=g (sigue actuando la gravedad)","v=máxima, a=0","Ambas máximas"],1,
    "En la cima v=0 instantáneamente, pero la aceleración sigue siendo g hacia abajo.")
```

## Punto 3 — Soltar vs lanzar hacia abajo (pozo de construcción)  ★★  [UCM, PRC]
Desde  h = 45,0 m, se comparan dos casos: (A) un objeto **soltado** (v₀ = 0) y
(B) otro **lanzado hacia abajo** con  v₀ = 5,00 m/s.

**Modelo:**  para caída hacia abajo,  h = v₀·t + ½·g·t²  (se resuelve para t).

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    fig,ax=plt.subplots(figsize=(4,4))
    ax.plot([-0.3,2.3],[4,4],'k-',lw=3)
    ax.scatter([0.5],[3.7],s=100,color='gray'); ax.text(0.5,4.15,'A: soltado',ha='center',fontsize=9)
    ax.scatter([1.7],[3.7],s=100,color='navy')
    ax.annotate('',xy=(1.7,3.2),xytext=(1.7,3.7),arrowprops=dict(arrowstyle='->',color='red',lw=2))
    ax.text(1.85,3.4,'v₀',color='red'); ax.text(1.7,4.15,'B: lanzado',ha='center',fontsize=9)
    ax.plot([-0.3,2.3],[0,0],color='brown',lw=3)
    ax.set_xlim(-0.6,2.8); ax.set_ylim(-0.4,4.6); ax.axis('off'); ax.set_title('Dos formas de caer desde h')
    plt.tight_layout(); plt.show()
esquema()

def comparar(h=45.0, v0_down=5.00):
    tA=np.sqrt(2*h/g)
    a_,b_,c_=0.5*g,v0_down,-h; tB=(-b_+np.sqrt(b_**2-4*a_*c_))/(2*a_)
    t=np.linspace(0,max(tA,tB),200)
    yA=h-0.5*g*t**2; yB=h-v0_down*t-0.5*g*t**2
    yA=np.clip(yA,0,None); yB=np.clip(yB,0,None)
    plt.figure(figsize=(9,4.5))
    plt.plot(t,yA,lw=2,color='gray',label=f'A soltado: t={tA:.3g} s')
    plt.plot(t,yB,lw=2,color='navy',label=f'B lanzado: t={tB:.3g} s')
    plt.axhline(0,color='brown',lw=1); plt.xlabel('t (s)'); plt.ylabel('altura (m)')
    plt.title('Comparación de tiempos de caída'); plt.legend(); plt.grid(alpha=.3); plt.tight_layout(); plt.show()

interact(comparar,
    h=FloatSlider(value=45.0,min=10,max=100,step=5,description='h (m)'),
    v0_down=FloatSlider(value=5.00,min=0,max=20,step=1,description='v₀ abajo (m/s)'));

def solucion():
    h=45.0; v0=5.00
    tA=np.sqrt(2*h/g)
    a_,b_,c_=0.5*g,v0,-h; tB=(-b_+np.sqrt(b_**2-4*a_*c_))/(2*a_)
    print(f"A) Soltado (v₀=0):  t = √(2h/g) = {tA:.3g} s")
    print(f"B) Lanzado hacia abajo (v₀=5,00 m/s):  resolviendo 45,0=5,00t+4,90t² ⟹ t = {tB:.3g} s")
    print(f"El lanzado llega antes ({tB:.3g} s < {tA:.3g} s), como se espera.")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 3 ---")
quiz_numerico("Tiempo del objeto SOLTADO (h=45,0 m):",3.03,0.04,"s","t=√(2h/g)=3,03 s.")
quiz_numerico("Tiempo del objeto LANZADO hacia abajo (v₀=5,00 m/s):",2.56,0.05,"s",
    "45,0=5,00t+4,90t² ⟹ t≈2,56 s.")
quiz_opcion_multiple("¿Cuál llega primero al suelo?",
    ["El soltado","El lanzado hacia abajo","Llegan al mismo tiempo","Depende de la masa"],1,
    "El lanzado ya parte con velocidad, así que recorre la altura en menos tiempo.")
```

```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            print("\u2705 \u00a1Correcto!" if radio.index==indice_correcto else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion=""):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            if abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel:
                print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

g = 9.80  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    fig,ax=plt.subplots(figsize=(3,4.5))
    ax.plot([-0.3,1.3],[0,0],color='brown',lw=3); ax.scatter([0.5],[0.15],s=100,color='black')
    ax.annotate('',xy=(0.5,3.5),xytext=(0.5,0.3),arrowprops=dict(arrowstyle='->',color='red',lw=2))
    ax.text(0.65,2,'v₀',color='red',fontsize=12)
    ax.plot([0.3,0.7],[3.5,3.5],'g--'); ax.text(0.9,3.5,'H',color='green')
    ax.set_xlim(-0.5,1.6); ax.set_ylim(-0.4,4); ax.axis('off'); ax.set_title('Tiro vertical desde el suelo')
    plt.tight_layout(); plt.show()
esquema()

def tiro_suelo(v0=20.0):
    t_fl=2*v0/g; t=np.linspace(0,t_fl,200); y=v0*t-0.5*g*t**2; v=v0-g*t
    H=v0**2/(2*g)
    fig,(a1,a2)=plt.subplots(1,2,figsize=(10,4))
    a1.plot(t,y,lw=2,color='navy'); a1.scatter([t_fl/2],[H],color='orange',s=50,zorder=5,label=f'H={H:.3g} m')
    a1.set_xlabel('t (s)'); a1.set_ylabel('altura (m)'); a1.set_title('Altura vs tiempo'); a1.legend(); a1.grid(alpha=.3)
    a2.plot(t,v,lw=2,color='crimson'); a2.axhline(0,color='k',lw=.5)
    a2.set_xlabel('t (s)'); a2.set_ylabel('velocidad (m/s)'); a2.set_title('Velocidad vs tiempo (baja y sube signo)'); a2.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(tiro_suelo, v0=FloatSlider(value=20.0,min=5,max=40,step=1,description='v₀ (m/s)'));

def solucion():
    v0=20.0; H=v0**2/(2*g); t_fl=2*v0/g
    print(f"Altura máxima:  H = v₀²/(2g) = {v0**2:.0f}/(2·9,80) = {H:.3g} m")
    print(f"Tiempo de vuelo total:  t = 2v₀/g = {t_fl:.3g} s")
    print(f"Velocidad al regresar al suelo: {v0:.3g} m/s (misma magnitud, sentido opuesto)")
    print(f"Verificación (v²=2gH):  {v0**2:.0f} ≈ {2*g*H:.0f} ✓")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("Altura máxima (v₀=20,0 m/s):",20.4,0.04,"m","H=v₀²/(2g)=400/19,6≈20,4 m.")
quiz_numerico("Tiempo de vuelo total:",4.08,0.04,"s","t=2v₀/g=40,0/9,80≈4,08 s.")
quiz_opcion_multiple("El tiempo de subida comparado con el de bajada es:",
    ["Mayor","Menor","Igual (movimiento simétrico)","Depende de v₀"],2,
    "Sin aire, subir y bajar toman el mismo tiempo: el movimiento es simétrico.")
```
