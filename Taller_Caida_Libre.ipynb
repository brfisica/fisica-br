## Punto 1 — Caída libre de una herramienta en obra  ★★  [CL, PRC]
En la construcción de un edificio en Bogotá, una herramienta se suelta accidentalmente
(v₀ = 0) desde un andamio a  h = 60,0 m  de altura.

**Modelo (caída libre desde reposo):**  h = ½·g·t²  →  t = √(2·h/g),   v = g·t

> 🔧 **Aplicación en ingeniería:** en obra, una herramienta que cae desde altura es un riesgo grave de seguridad industrial. Calcular el tiempo y la velocidad de impacto sustenta las normas de protección (mallas, zonas de exclusión, casco).


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


g = 9.81  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    from matplotlib.collections import LineCollection
    g=9.81; h=60
    t_v=np.sqrt(2*h/g); t=np.linspace(0,t_v,100)
    y=h-0.5*g*t**2; v=g*t
    fig,(ax1,ax2)=plt.subplots(1,2,figsize=(8,4.2),gridspec_kw={'width_ratios':[1,1.6]})
    # escena
    ax1.axhspan(0,h*1.1,color='#eaf4fb',alpha=0.6)
    ax1.axhspan(-3,0,color='#d9c8a3',alpha=0.6)
    ax1.plot([-1,1],[h,h],color='#7f8c8d',lw=4)
    ax1.text(0,h+2,'andamio',ha='center',fontsize=8,color='#2c3e50')
    ax1.scatter([0],[h*0.7],s=140,color='#34495e',zorder=5,edgecolor='white',linewidth=1.5)
    ax1.annotate('',xy=(0,h*0.45),xytext=(0,h*0.62),arrowprops=dict(arrowstyle='-|>',color='#2980b9',lw=2.5))
    ax1.text(0.15,h*0.55,'g',color='#2980b9',fontsize=12,fontweight='bold')
    ax1.axhline(0,color='#8b7355',lw=2)
    ax1.text(0,-2,'suelo',ha='center',fontsize=8,color='#8b5a2b')
    ax1.set_xlim(-1.5,1.5); ax1.set_ylim(-3,h*1.15); ax1.axis('off')
    ax1.set_title('Escena',fontsize=10)
    # grafico y(t) coloreado por velocidad
    pts=np.array([t,y]).T.reshape(-1,1,2); segs=np.concatenate([pts[:-1],pts[1:]],axis=1)
    lc=LineCollection(segs,cmap='plasma',linewidth=3.5); lc.set_array(v); ax2.add_collection(lc)
    ax2.scatter([0],[h],s=80,color='#34495e',zorder=5,edgecolor='white',linewidth=1.5)
    ax2.scatter([t_v],[0],s=100,marker='X',color='#c0392b',zorder=5,edgecolor='white',linewidth=1.5)
    ax2.annotate('impacto',xy=(t_v,0),xytext=(t_v*0.6,h*0.15),fontsize=8,color='#c0392b')
    ax2.set_xlim(0,t_v*1.05); ax2.set_ylim(-2,h*1.1)
    ax2.set_xlabel('tiempo (s)'); ax2.set_ylabel('altura (m)')
    ax2.set_title('Altura vs tiempo (acelera al caer)',fontsize=10); ax2.grid(alpha=0.2,linestyle='--')
    cb=plt.colorbar(lc,ax=ax2,pad=0.02,fraction=0.045); cb.set_label('rapidez (m/s)',fontsize=8); cb.ax.tick_params(labelsize=7)
    plt.suptitle('Caída libre de una herramienta en obra',fontsize=11,fontweight='bold',color='#2c3e50')
    plt.tight_layout(); plt.show()
esquema()

def caida_libre(h=60.0):
    t_fin=np.sqrt(2*h/g); t=np.linspace(0,t_fin,200); y=h-0.5*g*t**2; v=g*t
    fig,(a1,a2)=plt.subplots(1,2,figsize=(10,4))
    a1.plot(t,y,lw=2,color='navy'); a1.set_xlabel('t (s)'); a1.set_ylabel('altura y (m)'); a1.set_title('Altura vs tiempo'); a1.grid(alpha=.3)
    a2.plot(t,v,lw=2,color='crimson'); a2.set_xlabel('t (s)'); a2.set_ylabel('rapidez v (m/s)'); a2.set_title('Rapidez vs tiempo'); a2.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(caida_libre, h=FloatSlider(value=60.0,min=5,max=250,step=5,description='h (m)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    h=60.0; t=np.sqrt(2*h/g); v=g*t
    _pc(f"Tiempo de caída:  t = √(2h/g) = √(2·60,0/9,81) = {t:.3g} s")
    _pc(f"Velocidad de impacto:  v = g·t = 9,81·{t:.3g} = {v:.3g} m/s = {v*3.6:.3g} km/h")
    _pc(f"Verificación (v²=2gh):  {v**2:.0f} ≈ {2*g*h:.0f} ✓")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 1 ---")
quiz_numerico("Una herramienta cae desde una altura de 60,0 m. Encuentre el tiempo que tarda en llegar al suelo.",3.50,0.04,"s","t=√(2h/g)=3,50 s.", id_preg="Caida_p1_1", peso=2.0)
quiz_opcion_multiple("Una herramienta cae desde 60,0 m y otra idéntica desde 15,0 m (un cuarto de la altura). ¿Cómo se comparan sus velocidades de impacto?",["La primera llega al cuádruple de velocidad","La primera llega al DOBLE de velocidad","Llegan a la misma velocidad","La primera llega 4 veces más rápido"],1,"Como v=√(2gh), la velocidad va con la RAÍZ de la altura. A un cuarto de altura, la velocidad es la mitad: √(1/4)=1/2. Por eso caer desde 4× la altura solo duplica la velocidad de impacto.", id_preg="Caida_p1_2", peso=2.0)
quiz_opcion_multiple("Si la herramienta que cae tuviera el doble de masa, ¿qué ocurriría con su tiempo de caída (despreciando el aire)?",
    ["Se reduce a la mitad","No cambia: la caída no depende de la masa","Se duplica","Aumenta un poco"],1,
    "En caída libre (sin aire) todos los cuerpos caen con la misma g, sin importar su masa.", id_preg="Caida_p1_3", peso=2.0)

# --- Estilo visual Física BR (apariencia global de los esquemas) ---
try:
    import matplotlib.pyplot as _plt_style
    _plt_style.rcParams['axes.spines.top']=False
    _plt_style.rcParams['axes.spines.right']=False
    _plt_style.rcParams['axes.titlesize']=12
    _plt_style.rcParams['axes.titleweight']='bold'
    _plt_style.rcParams['axes.titlecolor']='#2c3e50'
    _plt_style.rcParams['axes.labelsize']=10
    _plt_style.rcParams['axes.labelcolor']='#34495e'
    _plt_style.rcParams['font.size']=10
    _plt_style.rcParams['axes.grid']=True
    _plt_style.rcParams['grid.alpha']=0.25
    _plt_style.rcParams['grid.linestyle']='--'
    _plt_style.rcParams['axes.edgecolor']='#7f8c8d'
    _plt_style.rcParams['xtick.color']='#34495e'
    _plt_style.rcParams['ytick.color']='#34495e'
    _plt_style.rcParams['figure.facecolor']='white'
    _plt_style.rcParams['axes.facecolor']='#fcfcfd'
except Exception:
    pass
# --- fin estilo ---
```

**Modelo:**  altura máxima  H = v₀²/(2·g);  tiempo de vuelo total  t = 2·v₀/g.

> 🔧 **Aplicación en ingeniería:** el ensayo de caída con grúa (drop test) mide alturas y velocidades para certificar materiales y empaques. La relación v²=2gh conecta la altura de ensayo con la velocidad de impacto exigida por la norma.


```python
# --- setup autónomo (no requiere ejecutar otras celdas) ---
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider, Button, Text, VBox, Output, HTML
from IPython.display import display
try:
    from google.colab import output as _cxo
    _cxo.enable_custom_widget_manager()
except Exception:
    pass

# Registro global para la calificación formativa ponderada
try:
    REGISTRO_BR
except NameError:
    REGISTRO_BR = {}
def _registrar(id_preg, acierto, peso):
    REGISTRO_BR[id_preg] = (bool(acierto), float(peso))

def mostrar_solucion(func_solucion):
    boton = Button(description="Mostrar solución", button_style='success', icon='eye')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output(); func_solucion()
    boton.on_click(_click)
    display(boton, salida)

def quiz_opcion_multiple(pregunta, opciones, indice_correcto, explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>")
    radio  = widgets.RadioButtons(options=opciones, layout={'width':'max-content'})
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            ok = (radio.index==indice_correcto)
            print("\u2705 \u00a1Correcto!" if ok else "\u274c Incorrecto. Int\u00e9ntalo de nuevo.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, radio, boton, salida]))

def quiz_numerico(pregunta, valor_correcto, tolerancia_rel=0.05, unidad="", explicacion="", id_preg=None, peso=1.0):
    titulo = HTML(f"<b>{pregunta}</b>  <i>(tolerancia {tolerancia_rel*100:.0f}%)</i>")
    caja   = Text(placeholder=f"valor en {unidad}" if unidad else "tu respuesta")
    boton  = Button(description="Verificar", button_style='info'); salida = Output()
    def revisar(_):
        with salida:
            salida.clear_output()
            try: v = float(caja.value.replace(",", "."))
            except ValueError: print("\u26a0 Escribe un n\u00famero."); return
            ok = abs(v - valor_correcto) <= abs(valor_correcto)*tolerancia_rel
            if ok: print(f"\u2705 \u00a1Correcto!  ({valor_correcto:.3g} {unidad})")
            else: print("\u274c Fuera de rango. Revisa tu procedimiento.")
            if explicacion: print("\u2192 " + explicacion)
            if id_preg is not None: _registrar(id_preg, ok, peso)
    boton.on_click(revisar); display(VBox([titulo, caja, boton, salida]))

def calificacion_final(total_preguntas=None):
    boton = Button(description="Calcular mi nota", button_style='primary', icon='star')
    salida = Output()
    def _click(_):
        with salida:
            salida.clear_output()
            if not REGISTRO_BR:
                print("A\u00fan no has respondido ning\u00fan cuestionario. Responde y vuelve a intentar.")
                return
            peso_total = sum(p for _,p in REGISTRO_BR.values())
            peso_ok = sum(p for ok,p in REGISTRO_BR.values() if ok)
            nota = 5.0*peso_ok/peso_total if peso_total>0 else 0.0
            resp = len(REGISTRO_BR); aciertos = sum(1 for ok,_ in REGISTRO_BR.values() if ok)
            print("="*42)
            print("        CALIFICACI\u00d3N FORMATIVA")
            print("="*42)
            print(f"  Preguntas respondidas: {resp}" + (f" de {total_preguntas}" if total_preguntas else ""))
            print(f"  Respuestas correctas:  {aciertos}")
            print(f"  Puntaje ponderado:     {peso_ok:.1f} / {peso_total:.1f}")
            print("-"*42)
            print(f"  NOTA:  {(f'%.1f'%nota).replace('.',',')}  / 5,0")
            print("="*42)
            if nota>=3.5: print("  \u2705 \u00a1Buen trabajo! Dominas los conceptos.")
            elif nota>=3.0: print("  \U0001f7e1 Vas bien, repasa lo que fallaste.")
            else: print("  \U0001f534 Repasa la teor\u00eda y reintenta los puntos.")
            print("\n  Nota formativa (para tu pr\u00e1ctica). Puedes")
            print("  reintentar los cuestionarios y recalcular.")
    boton.on_click(_click)
    display(VBox([HTML("<h3>\U0001f4ca Tu calificaci\u00f3n del taller</h3>"
        "<i>Responde los cuestionarios y pulsa el bot\u00f3n para ver tu nota ponderada por dificultad.</i>"),
        boton, salida]))


g = 9.81  # m/s²  (valor usado en el libro de texto)
# --- fin setup ---


def esquema():
    from matplotlib.collections import LineCollection
    g=9.81; v0=20.0
    t_v=2*v0/g; t=np.linspace(0,t_v,100)
    y=v0*t-0.5*g*t**2; v=np.abs(v0-g*t)
    fig,(ax1,ax2)=plt.subplots(1,2,figsize=(8,4.2),gridspec_kw={'width_ratios':[1,1.6]})
    ax1.axhspan(0,y.max()*1.2,color='#eaf4fb',alpha=0.6)
    ax1.axhspan(-2,0,color='#d9c8a3',alpha=0.6)
    ax1.axhline(0,color='#8b7355',lw=2); ax1.scatter([0],[0.3],s=110,color='#34495e',zorder=5,edgecolor='white',linewidth=1.5)
    ax1.annotate('',xy=(0,y.max()*0.55),xytext=(0,0.5),arrowprops=dict(arrowstyle='-|>',color='#27ae60',lw=2.5))
    ax1.text(0.2,y.max()*0.3,'v₀',color='#27ae60',fontsize=11,fontweight='bold')
    ax1.axhline(y.max(),color='#f39c12',ls='--',lw=1.5); ax1.text(0,y.max()+0.6,'H máx',ha='center',fontsize=8,color='#b9770e')
    ax1.set_xlim(-1.5,1.5); ax1.set_ylim(-2,y.max()*1.2); ax1.axis('off'); ax1.set_title('Escena',fontsize=10)
    pts=np.array([t,y]).T.reshape(-1,1,2); segs=np.concatenate([pts[:-1],pts[1:]],axis=1)
    lc=LineCollection(segs,cmap='viridis',linewidth=3.5); lc.set_array(v); ax2.add_collection(lc)
    idx=np.argmax(y)
    ax2.scatter([t[idx]],[y[idx]],s=90,color='#f39c12',zorder=5,edgecolor='white',linewidth=2)
    ax2.annotate('altura máxima',xy=(t[idx],y[idx]),xytext=(t[idx]*0.5,y[idx]+0.5),fontsize=8,color='#b9770e')
    ax2.set_xlim(0,t_v*1.05); ax2.set_ylim(0,y.max()*1.15)
    ax2.set_xlabel('tiempo (s)'); ax2.set_ylabel('altura (m)')
    ax2.set_title('Trayectoria vertical (v=0 en la cima)',fontsize=10); ax2.grid(alpha=0.2,linestyle='--')
    cb=plt.colorbar(lc,ax=ax2,pad=0.02,fraction=0.045); cb.set_label('rapidez (m/s)',fontsize=8); cb.ax.tick_params(labelsize=7)
    plt.suptitle('Ensayo de altura con grúa',fontsize=11,fontweight='bold',color='#2c3e50')
    plt.tight_layout(); plt.show()
esquema()

def tiro_suelo(v0=20.0):
    t_fl=2*v0/g; t=np.linspace(0,t_fl,200); y=v0*t-0.5*g*t**2; v=v0-g*t
    H=v0**2/(2*g)
    fig,(a1,a2)=plt.subplots(1,2,figsize=(10,4))
    a1.plot(t,y,lw=2,color='navy'); a1.scatter([t_fl/2],[H],color='orange',s=50,zorder=5,label=f'H={H:.3g} m')
    a1.set_xlabel('t (s)'); a1.set_ylabel('altura (m)'); a1.set_title('Altura vs tiempo'); a1.legend(); a1.grid(alpha=.3)
    a2.plot(t,v,lw=2,color='crimson'); a2.axhline(0,color='k',lw=.5)
    a2.set_xlabel('t (s)'); a2.set_ylabel('velocidad (m/s)'); a2.set_title('Velocidad vs tiempo (baja y sube signo)'); a2.grid(alpha=.3)
    plt.tight_layout(); plt.show()

interact(tiro_suelo, v0=FloatSlider(value=20.0,min=5,max=40,step=1,description='v₀ (m/s)'));

def solucion():
    import re as _re
    def _pc(*a,**k):
        def _f(t):
            if not isinstance(t,str): return t
            _sup='⁰¹²³⁴⁵⁶⁷⁸⁹'
            def _sci(m):
                mant=m.group(1); exp=int(m.group(2))
                e=('⁻' if exp<0 else '')+''.join(_sup[int(d)] for d in str(abs(exp)))
                return mant+'×10'+e
            t=_re.sub(r'(\d+(?:\.\d+)?)[eE]([+-]?\d+)', _sci, t)
            _u=r'(?:s|km/s|km|m/s²|m/s|m|N·m|N|J|W|A|V|T|Ω|C|F|Hz|rad/s²|rad/s|rad|kg·m²/s|kg·m²|kg|Pa|kPa|µF|nC|µC|pF|mWb|Wb|µT|mT|kW|MW|mA|kΩ|MΩ|rpm|°)'
            def _c3(m):
                num=m.group(2); n=int(num)
                if n==0: return m.group(0)
                if len(num)==1: nv=num+'.00'
                elif len(num)==2: nv=num+'.0'
                else: nv=num
                return m.group(1)+nv+m.group(3)
            t=_re.sub(r'(?<![\d.,])([\s=(×])(\d{1,3})(\s'+_u+r'\b)', _c3, t)
            t=_re.sub(r'(\d)\.(\d)', r'\1,\2', t)
            return t
        print(*[_f(x) for x in a], **k)

    v0=20.0; H=v0**2/(2*g); t_fl=2*v0/g
    _pc(f"Altura máxima:  H = v₀²/(2g) = {v0**2:.0f}/(2·9,81) = {H:.3g} m")
    _pc(f"Tiempo de vuelo total:  t = 2v₀/g = {t_fl:.3g} s")
    _pc(f"Velocidad al regresar al suelo: {v0:.3g} m/s (misma magnitud, sentido opuesto)")
    _pc(f"Verificación (v²=2gH):  {v0**2:.0f} ≈ {2*g*H:.0f} ✓")
mostrar_solucion(solucion)

print("\n--- Cuestionario Punto 4 ---")
quiz_numerico("En la refinería, una válvula sale despedida hacia arriba a 20,0 m/s. Encuentre la altura máxima que alcanza.",20.4,0.04,"m","H=v₀²/(2g)=400/19,6≈20,4 m.", id_preg="Caida_p4_10", peso=3.0)
quiz_numerico("Para ese objeto lanzado hacia arriba a 20,0 m/s (que regresa a la altura de lanzamiento), encuentre el tiempo de vuelo total.",4.08,0.04,"s","t=2v₀/g=40,0/9,81≈4,08 s.", id_preg="Caida_p4_11", peso=3.0)
quiz_opcion_multiple("En un lanzamiento vertical sin resistencia del aire, ¿cómo es el tiempo de subida comparado con el de bajada?",
    ["Mayor","Menor","Igual (movimiento simétrico)","Depende de v₀"],2,
    "Sin aire, subir y bajar toman el mismo tiempo: el movimiento es simétrico.", id_preg="Caida_p4_12", peso=3.0)
```
